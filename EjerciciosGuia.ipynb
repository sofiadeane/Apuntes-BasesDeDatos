### EJERCICIO N2. 
Mostrar el código, detalle de todos los artículos vendidos en el año 2012 ordenados por cantidad vendida.

```sql
SELECT 
    SUM(i.item_cantidad) AS cantidad_vendida,
    p.prod_codigo AS codigo,
    p.prod_detalle AS detalle
FROM Producto p
INNER JOIN Item_Factura i ON p.prod_codigo = i.item_producto
INNER JOIN Factura f 
    ON  i.item_tipo     = f.fact_tipo
    AND i.item_sucursal = f.fact_sucursal
    AND i.item_numero   = f.fact_numero
WHERE YEAR(f.fact_fecha) = 2012
GROUP BY 
    p.prod_codigo,
    p.prod_detalle
ORDER BY cantidad_vendida DESC
```

```sql
SELECT 
    SUM(s.stoc_cantidad)    AS cantidad,
    p.prod_codigo           AS codigo,
    p.prod_detalle          AS nombre
FROM Producto p
INNER JOIN Stock s ON p.prod_codigo = s.stoc_producto
GROUP BY 
    p.prod_codigo, 
    p.prod_detalle
ORDER BY p.prod_detalle ASC
```

### EJERCICIO N4. 
Realizar una consulta que muestre para todos los artículos código, detalle y cantidad de artículos que lo componen. Mostrar solo aquellos artículos para los cuales el stock promedio por depósito sea mayor a 100.

```sql
SELECT 
    p.prod_codigo   AS codigo,
    p.prod_detalle  AS detalle,
    SUM(ISNULL(c.comp_cantidad, 0)) AS cantidad_articulos
FROM Producto p
LEFT JOIN Composicion c ON c.comp_producto = p.prod_codigo 
GROUP BY 
    p.prod_codigo, 
    p.prod_detalle
HAVING ( SELECT AVG(s.stoc_cantidad) 
         FROM Stock s 
         WHERE s.stoc_producto = p.prod_codigo ) > 100       -- subselect correlacionado: no multiplica filas
ORDER BY codigo ASC
```

**Por qué un subselect en el `HAVING`:** si uniera `Producto` con `Composicion` **y** con `Stock` en el mismo `FROM`, cada componente se repetiría una vez por cada depósito y `SUM(comp_cantidad)` daría de más (problema de cardinalidad). Calculando el promedio de stock con un subselect correlacionado, el `FROM` solo une con `Composicion` y los componentes se suman bien.

**Solución del profe:** une las tres tablas y después divide para deshacer la repetición: la suma se multiplicó por la cantidad de depósitos, así que la divide por `COUNT(DISTINCT stoc_deposito)`. El `AVG` del stock no se altera porque todas las filas de stock se repiten la misma cantidad de veces.

```sql
SELECT 
    Producto.prod_codigo,
    Producto.prod_detalle,
    ISNULL( SUM(comp_cantidad) / COUNT(DISTINCT stoc_deposito), 0 ) AS Componentes
FROM Producto
LEFT JOIN Composicion ON Producto.prod_codigo = Composicion.comp_producto
JOIN Stock ON Producto.prod_codigo = Stock.stoc_producto
GROUP BY Producto.prod_codigo, Producto.prod_detalle
HAVING AVG(Stock.stoc_cantidad) > 100
```

### EJERCICIO N5. 
Realizar una consulta que muestre código de artículo, detalle y cantidad de egresos de stock que se realizaron para ese artículo en el año 2012 (egresan los productos que fueron vendidos). Mostrar solo aquellos que hayan tenido más egresos que en el 2011.

```sql
SELECT 
    p1.prod_codigo, 
    p1.prod_detalle, 
    SUM(i1.item_cantidad) AS egresos
FROM Producto p1 
JOIN Item_Factura i1 ON p1.prod_codigo = i1.item_producto
JOIN Factura f1 
    ON  i1.item_tipo     = f1.fact_tipo 
    AND i1.item_sucursal = f1.fact_sucursal 
    AND i1.item_numero   = f1.fact_numero
WHERE YEAR(f1.fact_fecha) = 2012
GROUP BY p1.prod_codigo, p1.prod_detalle
HAVING SUM(i1.item_cantidad) > ( SELECT ISNULL(SUM(i2.item_cantidad), 0)         -- ISNULL AFUERA del SUM
                                 FROM Factura f2 
                                 JOIN Item_Factura i2 
                                    ON  i2.item_tipo     = f2.fact_tipo 
                                    AND i2.item_sucursal = f2.fact_sucursal 
                                    AND i2.item_numero   = f2.fact_numero
                                 WHERE YEAR(f2.fact_fecha) = 2011 
                                   AND i2.item_producto = p1.prod_codigo )
ORDER BY 3 DESC
```

Un `ISNULL` adentro del `SUM` no lo evita (no hay filas a las que aplicárselo). Como `número > NULL` nunca es verdadero, ese artículo quedaría afuera aunque en 2012 haya vendido más que "cero" en 2011.

**Otra forma, sin subselect (idea del profe con `CASE`):** recorro las facturas de 2011 y 2012 una sola vez y sumo por separado cada año con un `CASE` dentro del `SUM`. Acá usa `item_cantidad` porque la consigna pide cantidad de egresos.

```sql
SELECT 
    p.prod_codigo,
    p.prod_detalle,
    SUM(CASE WHEN YEAR(f.fact_fecha) = 2012 THEN i.item_cantidad ELSE 0 END) AS egresos_2012
FROM Item_Factura i 
JOIN Factura f 
    ON  i.item_tipo     = f.fact_tipo 
    AND i.item_sucursal = f.fact_sucursal 
    AND i.item_numero   = f.fact_numero
JOIN Producto p ON p.prod_codigo = i.item_producto
WHERE YEAR(f.fact_fecha) IN (2011, 2012)
GROUP BY p.prod_codigo, p.prod_detalle
HAVING SUM(CASE WHEN YEAR(f.fact_fecha) = 2012 THEN i.item_cantidad ELSE 0 END) 
     > SUM(CASE WHEN YEAR(f.fact_fecha) = 2011 THEN i.item_cantidad ELSE 0 END)
ORDER BY p.prod_codigo
```

### EJERCICIO N6. 
Mostrar para todos los rubros de artículos código, detalle, cantidad de artículos de ese rubro y stock total de ese rubro de artículos. Solo tener en cuenta aquellos artículos que tengan un stock mayor al del artículo '00000000' en el depósito '00'.

El stock del artículo se toma como la suma de todos sus depósitos, y se compara contra el stock de `'00000000'` en el depósito `'00'`. El stock se calcula en un subselect con `IN` para no multiplicar filas al unir con `Stock`.

```sql
-- articulos cuyo stock total (suma de todos los depositos) supera al stock del articulo '00000000' en el deposito '00'
-- OJO: rubros sin ningun articulo que cumpla quedan afuera (el filtro esta en el WHERE). Si la consigna los quiere con 0, hay que
-- pasar el filtro al ON de un LEFT JOIN desde Rubro.
SELECT 
    r.rubr_id                   AS codigo,
    r.rubr_detalle              AS detalle,
    COUNT(DISTINCT p.prod_codigo) AS cantidad_articulos,
    SUM(s.stoc_cantidad)        AS stock_total
FROM Rubro r
JOIN Producto p ON p.prod_rubro = r.rubr_id
JOIN Stock s    ON s.stoc_producto = p.prod_codigo
WHERE p.prod_codigo IN ( SELECT stoc_producto 
                         FROM Stock 
                         GROUP BY stoc_producto
                         HAVING SUM(stoc_cantidad) > ( SELECT stoc_cantidad 
                                                       FROM Stock 
                                                       WHERE stoc_producto = '00000000' AND stoc_deposito = '00' ) )
GROUP BY r.rubr_id, r.rubr_detalle
```

### EJERCICIO EXTRA B (de clase)
Mostrar los 10 productos más vendidos en la historia y también los 10 productos menos vendidos en la historia. Además mostrar de esos productos, quién fue el cliente que mayor compra realizó.

```sql
SELECT 
    i1.item_producto,
    ( SELECT TOP 1 f2.fact_cliente                                   -- el cliente que mas unidades de ESE producto compro
      FROM Item_Factura i2 
      JOIN Factura f2 
          ON  i2.item_tipo     = f2.fact_tipo 
          AND i2.item_sucursal = f2.fact_sucursal 
          AND i2.item_numero   = f2.fact_numero
      WHERE i2.item_producto = i1.item_producto
      GROUP BY f2.fact_cliente
      ORDER BY SUM(i2.item_cantidad) DESC ) AS clienteMasComprador
FROM Item_Factura i1
WHERE i1.item_producto IN ( SELECT TOP 10 i2.item_producto           -- los 10 mas vendidos
                            FROM Item_Factura i2
                            GROUP BY i2.item_producto
                            ORDER BY SUM(i2.item_cantidad) DESC )
   OR i1.item_producto IN ( SELECT TOP 10 i2.item_producto           -- los 10 menos vendidos
                            FROM Item_Factura i2
                            GROUP BY i2.item_producto
                            ORDER BY SUM(i2.item_cantidad) ASC )
GROUP BY i1.item_producto
ORDER BY SUM(i1.item_cantidad) DESC
```

### Clase 2: Agregaciones y `GROUP BY`
Ver [Clase2.ipynb](Clase2.ipynb), sección *Ejercicios para practicar* (ejercicios 1 a 5).

```sql
-- CLASE 2 - EJERCICIO 1: por año, cantidad de facturas y monto total
SELECT
    YEAR(fact_fecha) AS anio,
    COUNT(*)         AS cantidad_facturas,
    SUM(fact_total)  AS monto_total
FROM Factura
GROUP BY YEAR(fact_fecha)
ORDER BY anio
```

```sql
-- CLASE 2 - EJERCICIO 2: por vendedor en 2012, solo los que tienen mas de N facturas (N = 100, ajustalo mirando los datos)
SELECT
    fact_vendedor    AS vendedor,
    COUNT(*)         AS cantidad_facturas,
    AVG(fact_total)  AS promedio_por_factura,
    MAX(fact_total)  AS factura_mayor
FROM Factura
WHERE YEAR(fact_fecha) = 2012
GROUP BY fact_vendedor
HAVING COUNT(*) > 100                      -- HAVING filtra grupos, WHERE filtra filas
ORDER BY cantidad_facturas DESC
```

```sql
-- CLASE 2 - EJERCICIO 3: Cliente en una sola fila
SELECT
    COUNT(*)                                AS total_clientes,
    COUNT(clie_telefono)                    AS con_telefono,            -- COUNT(campo) no cuenta los NULL
    COUNT(*) - COUNT(clie_telefono)         AS sin_telefono,
    AVG(clie_limite_credito)                AS limite_promedio,
    MIN(clie_limite_credito)                AS limite_minimo,
    MAX(clie_limite_credito)                AS limite_maximo
FROM Cliente
```

```sql
-- CLASE 2 - EJERCICIO 4: meses de 2012 que facturaron mas que el promedio mensual del año
-- Lo interpreto asi: "promedio" = total del año / cantidad de meses con facturas.
-- NO se puede sin subconsulta: hay que comparar el total de cada mes (un nivel de agrupacion)
-- contra un valor del año completo (otro nivel). Por eso queda para la Clase 4.
SELECT
    MONTH(fact_fecha)           AS mes,
    COUNT(*)                    AS cantidad_facturas,
    SUM(fact_total_impuestos)   AS total_impuestos
FROM Factura
WHERE YEAR(fact_fecha) = 2012
GROUP BY MONTH(fact_fecha)
HAVING SUM(fact_total_impuestos) > ( SELECT SUM(fact_total_impuestos) / COUNT(DISTINCT MONTH(fact_fecha))
                                     FROM Factura
                                     WHERE YEAR(fact_fecha) = 2012 )
ORDER BY mes
```

```sql
-- CLASE 2 - EJERCICIO 5: stock total por producto, solo los que superan 1000, y en cuantos depositos aparecen
SELECT
    stoc_producto                       AS producto,
    SUM(stoc_cantidad)                  AS stock_total,
    COUNT(DISTINCT stoc_deposito)       AS depositos
FROM Stock
GROUP BY stoc_producto
HAVING SUM(stoc_cantidad) > 1000
ORDER BY stock_total DESC
```

```sql
-- CLASE 3 - EJERCICIO 9: código del jefe, código y nombre de cada empleado que lo tiene como jefe, y la cantidad de depósitos que ambos tienen asignados. 
SELECT 
    j.empl_codigo AS jefe,
    e.empl_codigo AS empleado,
    e.empl_nombre AS nombre_empleado,
    COUNT(d.depo_codigo) AS cant_depositos
FROM Empleado e -- en e busco datos de empleados
JOIN Empleado j ON j.empl_codigo = e.empl_jefe -- en j busco los datos del jefe
LEFT JOIN DEPOSITO d ON d.depo_encargado IN (e.empl_codigo, j.empl_codigo) -- depositos del empleado o del jefe
GROUP BY  j.empl_codigo, e.empl_codigo, e.empl_nombre
```

```sql
-- CLASE 3 - EJERCICIO 11: por familia: detalle, cantidad de productos diferentes vendidos y monto de las ventas sin impuestos; ordenado por la familia con más productos diferentes vendidos, solo familias con ventas > $20000 en 2012.

SELECT
    f.fami_detalle                              AS nombre_familia,
    COUNT(DISTINCT i.item_producto)             AS productos_distintos,
    SUM(i.item_cantidad * i.item_precio)        AS monto_sin_impuestos   -- item_precio es unitario
FROM Producto p
JOIN Familia f ON f.fami_id = p.prod_familia
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura fac 
    ON  i.item_tipo     = fac.fact_tipo
    AND i.item_sucursal = fac.fact_sucursal
    AND i.item_numero   = fac.fact_numero
WHERE YEAR(fac.fact_fecha) = 2012
GROUP BY f.fami_id, f.fami_detalle
HAVING SUM(i.item_cantidad * i.item_precio) > 20000
ORDER BY productos_distintos DESC
```

```sql
-- CLASE 3 - EJERCICIO 13: para cada producto con composición: nombre, precio, y la suma de (precio del componente x cantidad) de lo que lo compone. Solo los compuestos por más de 2 productos, ordenados de mayor a menor por cantidad de componentes.

SELECT
    p.prod_detalle                              AS nombre,
    p.prod_precio                               AS precio,
    SUM(c.comp_cantidad * p_comp.prod_precio)   AS precio_componentes
FROM Composicion c
JOIN Producto p      ON p.prod_codigo      = c.comp_producto      -- p = el producto compuesto
JOIN Producto p_comp ON p_comp.prod_codigo = c.comp_componente    -- p_comp = cada componente
GROUP BY c.comp_producto, p.prod_detalle, p.prod_precio
HAVING SUM(c.comp_cantidad) > 2                                   -- unidades que lo componen
ORDER BY SUM(c.comp_cantidad) DESC
```

```sql
-- CLASE 3 - EJERCICIO 15: pares de productos vendidos juntos (misma factura) más de 500 veces: código y detalle de cada uno y cantidad de veces, ordenado por esa cantidad, sin repetir pares. *(Auto-join de `Item_Factura`.)*

SELECT
    i1.item_producto    AS cod_producto_1,
    p1.prod_detalle     AS detalle_1,
    i2.item_producto    AS cod_producto_2,
    p2.prod_detalle     AS detalle_2,
    COUNT(*)            AS veces_juntos
FROM Item_Factura i1
JOIN Item_Factura i2 
    ON  i2.item_tipo     = i1.item_tipo
    AND i2.item_sucursal = i1.item_sucursal
    AND i2.item_numero   = i1.item_numero              -- misma factura
    AND i2.item_producto > i1.item_producto            -- > y no <>: (A,B) y no tambien (B,A)
JOIN Producto p1 ON p1.prod_codigo = i1.item_producto
JOIN Producto p2 ON p2.prod_codigo = i2.item_producto
GROUP BY i1.item_producto, p1.prod_detalle, i2.item_producto, p2.prod_detalle
HAVING COUNT(*) > 500
ORDER BY veces_juntos DESC
```

```sql
-- CLASE 4 - EJERCICIO 7: Generar una consulta que muestre para cada artículo código, detalle, mayor precio, menor
-- precio y % de la diferencia de precios (respecto del menor. Ej.: menor precio = 10, mayor precio = 12 => mostrar
-- 20 %). Mostrar solo aquellos artículos que posean stock.
-- NOTAS: Precios tomados de Item_Factura (lo facturado); % = (MAX - MIN) * 100 / MIN redondeado a 2 decimales;
-- 'posee stock' = existe algún depósito con stoc_cantidad > 0 (aclaración del profe). Devuelve 633 filas en la
-- base (con SUM total de stock > 0 serían 627); ej. 00000109: 8.13 y 7.08 => 14.83 %.
SELECT p.prod_codigo, p.prod_detalle,
       MAX(i.item_precio)                                                 AS mayor_precio,
       MIN(i.item_precio)                                                 AS menor_precio,
       CAST((MAX(i.item_precio) - MIN(i.item_precio)) * 100 / MIN(i.item_precio) AS DECIMAL(12,2)) AS porc_diferencia  -- respecto del menor
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo     -- los precios salen de lo facturado
WHERE EXISTS ( SELECT 1 FROM STOCK s
               WHERE s.stoc_producto = p.prod_codigo AND s.stoc_cantidad > 0 )   -- "posee stock"
GROUP BY p.prod_codigo, p.prod_detalle
ORDER BY p.prod_codigo
```

```sql
-- CLASE 4 - EJERCICIO 12: Mostrar nombre de producto, cantidad de clientes distintos que lo compraron, importe
-- promedio pagado por el producto, cantidad de depósitos en los cuales hay stock del producto y stock actual del
-- producto en todos los depósitos. Se deberán mostrar aquellos productos que hayan tenido operaciones en el año
-- 2012 y los datos deberán ordenarse de mayor a menor por monto vendido del producto.
-- NOTAS: El 2012 solo filtra qué productos salen (EXISTS); clientes, precio promedio (AVG item_precio) y monto
-- vendido son de toda la historia. Depósitos con stock = stoc_cantidad > 0; STOCK va en mandaderos para no
-- multiplicar filas. Devuelve 629 filas; verificado a mano 'Linterna con pilas': 22 clientes, 10 depósitos con
-- stock (12 filas en STOCK), stock 40.
SELECT p.prod_detalle,
       COUNT(DISTINCT f.fact_cliente)                           AS clientes_distintos,
       AVG(i.item_precio)                                       AS precio_promedio,
       ( SELECT COUNT(*) FROM STOCK s                            -- mandadero: STOCK en la mesa multiplicaria los items
         WHERE s.stoc_producto = p.prod_codigo AND s.stoc_cantidad > 0 )             AS depositos_con_stock,
       ( SELECT ISNULL(SUM(s.stoc_cantidad), 0) FROM STOCK s
         WHERE s.stoc_producto = p.prod_codigo )                                      AS stock_total
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE EXISTS ( SELECT 1                                          -- tuvo operaciones en 2012
               FROM Item_Factura i2
               JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
               WHERE i2.item_producto = p.prod_codigo AND YEAR(f2.fact_fecha) = 2012 )
GROUP BY p.prod_codigo, p.prod_detalle
ORDER BY SUM(i.item_cantidad * i.item_precio) DESC              -- monto vendido del producto
```

```sql
-- CLASE 4 - EJERCICIO 14: Escriba una consulta que retorne una estadística de ventas por cliente. Los campos que
-- debe retornar son: código del cliente, cantidad de veces que compró en el último año, promedio por compra en el
-- último año, cantidad de productos diferentes que compró en el último año y monto de la mayor compra que realizó
-- en el último año. Se deberán retornar todos los clientes ordenados por la cantidad de veces que compró en el
-- último año. No se deberán visualizar NULLs en ninguna columna.
-- NOTAS: Versión UNION ALL del profe (Receta 6 b de Parcial-1): 'último año' = 2012 (último año de la base); los
-- que compraron por un lado y los que no (todo en 0) por el otro. Devuelve 3687 filas, ningún NULL; arriba 01742
-- con 190 compras, promedio 876.20, 121 productos y mayor compra 4421.16 (igual que la versión con LEFT JOIN).
SELECT c.clie_codigo,
       COUNT(*)          AS compras,
       AVG(f.fact_total) AS promedio,
       (SELECT COUNT(DISTINCT i2.item_producto)                    -- mandadero: Item_Factura multiplicaria
          FROM Factura f2
          JOIN Item_Factura i2 ON i2.item_tipo = f2.fact_tipo AND i2.item_sucursal = f2.fact_sucursal
                              AND i2.item_numero = f2.fact_numero
         WHERE f2.fact_cliente = c.clie_codigo
           AND YEAR(f2.fact_fecha) = 2012) AS productos_distintos,
       MAX(f.fact_total) AS mayor_compra
FROM Cliente c
JOIN Factura f ON f.fact_cliente = c.clie_codigo
WHERE YEAR(f.fact_fecha) = 2012
GROUP BY c.clie_codigo
UNION ALL                                                          -- los que no compraron, todo en 0
SELECT c.clie_codigo, 0, 0, 0, 0
FROM Cliente c
WHERE NOT EXISTS (SELECT 1 FROM Factura f2
                   WHERE f2.fact_cliente = c.clie_codigo AND YEAR(f2.fact_fecha) = 2012)
ORDER BY 2 DESC, 1                                                 -- por cantidad de compras
```

```sql
-- CLASE 4 - EJERCICIO 17: Escriba una consulta que retorne una estadística de ventas por año y mes para cada
-- producto. La consulta debe retornar: PERIODO: año y mes de la estadística con el formato YYYYMM; PROD: código de
-- producto; DETALLE: detalle del producto; CANTIDAD_VENDIDA: cantidad vendida del producto en el período;
-- VENTAS_AÑO_ANT: cantidad vendida del producto en el mismo mes del período pero del año anterior; CANT_FACTURAS:
-- cantidad de facturas en las que se vendió el producto en el período. La consulta no puede mostrar NULL en
-- ninguna de sus columnas y debe estar ordenada por período y código de producto.
-- NOTAS: PERIODO con CONVERT(CHAR(6), fecha, 112); año anterior en mandadero con ISNULL afuera; columna llamada
-- VENTAS_ANIO_ANT (sin Ñ). Devuelve 3319 filas sin NULLs; como los datos de 2011 arrancan en julio, solo 201207 /
-- 00010200 tiene año anterior distinto de 0 (90 u. vs 1 u. en 201107, verificado).
SELECT CONVERT(CHAR(6), f.fact_fecha, 112)          AS PERIODO,          -- 112 = AAAAMMDD, me quedo con AAAAMM
       p.prod_codigo                                 AS PROD,
       p.prod_detalle                                AS DETALLE,
       SUM(i.item_cantidad)                          AS CANTIDAD_VENDIDA,
       ISNULL(( SELECT SUM(i2.item_cantidad)                              -- mismo mes, año anterior (ISNULL afuera)
                FROM Item_Factura i2
                JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
                WHERE i2.item_producto = p.prod_codigo
                  AND YEAR(f2.fact_fecha)  = YEAR(f.fact_fecha) - 1
                  AND MONTH(f2.fact_fecha) = MONTH(f.fact_fecha) ), 0) AS VENTAS_ANIO_ANT,
       COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)  AS CANT_FACTURAS   -- facturas distintas
FROM Item_Factura i
JOIN Factura f  ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
JOIN Producto p ON p.prod_codigo = i.item_producto
GROUP BY CONVERT(CHAR(6), f.fact_fecha, 112), YEAR(f.fact_fecha), MONTH(f.fact_fecha), p.prod_codigo, p.prod_detalle
ORDER BY PERIODO, PROD
```

```sql
-- CLASE 4 - EJERCICIO 20: Escriba una consulta SQL que retorne un ranking de los mejores 3 empleados del 2012. Se
-- deberá retornar legajo, nombre y apellido, año de ingreso, puntaje 2011 y puntaje 2012. El puntaje de cada
-- empleado se calculará de la siguiente manera: para los que hayan vendido al menos 50 facturas, el puntaje se
-- calculará como la cantidad de facturas que superen los 100 pesos que haya vendido en el año; para los que tengan
-- menos de 50 facturas en el año, el cálculo del puntaje será el 50% de la cantidad de facturas realizadas por sus
-- subordinados directos en dicho año.
-- NOTAS: Ranking por puntaje 2012 (desempata legajo); 'vendió' = fact_vendedor; subordinados directos = empl_jefe
-- = legajo. Devuelve 3 filas: 3 Quiroga (45 / 475), 4 Trap (59 / 471), 1 Perez (36 / 328.5); verificado a mano:
-- Perez tiene < 50 facturas y sus subordinados 2 y 3 hicieron 0 + 657 en 2012 => 328.5.
SELECT TOP 3
       e.empl_codigo                                AS legajo,
       RTRIM(e.empl_nombre) + ' ' + RTRIM(e.empl_apellido) AS nombre_apellido,
       YEAR(e.empl_ingreso)                         AS anio_ingreso,
       CASE WHEN ( SELECT COUNT(*) FROM Factura f
                   WHERE f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2011 ) >= 50
            THEN ( SELECT COUNT(*) FROM Factura f                     -- >= 50 facturas: las que superan $100
                   WHERE f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2011 AND f.fact_total > 100 )
            ELSE ( SELECT COUNT(*) FROM Factura f                     -- < 50: 50% de las facturas de sus subordinados directos
                   JOIN Empleado s ON s.empl_codigo = f.fact_vendedor
                   WHERE s.empl_jefe = e.empl_codigo AND YEAR(f.fact_fecha) = 2011 ) * 0.5
       END                                          AS puntaje_2011,
       CASE WHEN ( SELECT COUNT(*) FROM Factura f
                   WHERE f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2012 ) >= 50
            THEN ( SELECT COUNT(*) FROM Factura f
                   WHERE f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2012 AND f.fact_total > 100 )
            ELSE ( SELECT COUNT(*) FROM Factura f
                   JOIN Empleado s ON s.empl_codigo = f.fact_vendedor
                   WHERE s.empl_jefe = e.empl_codigo AND YEAR(f.fact_fecha) = 2012 ) * 0.5
       END                                          AS puntaje_2012
FROM Empleado e
ORDER BY puntaje_2012 DESC, e.empl_codigo                         -- el ranking es por el puntaje 2012
```

```sql
-- CLASE 4 - EJERCICIO 21: Escriba una consulta SQL que retorne para todos los años en los cuales se haya hecho al
-- menos una factura, la cantidad de clientes a los que se les facturó de manera incorrecta al menos una factura y
-- qué cantidad de facturas se realizaron de manera incorrecta. Se considera que una factura es incorrecta cuando
-- la diferencia entre el total de la factura menos el total de impuestos tiene una diferencia mayor a $1 respecto
-- a la sumatoria de los costos de cada uno de los ítems de dicha factura. Las columnas que se deben mostrar son:
-- año, clientes a los que se les facturó mal en ese año y facturas mal realizadas en ese año.
-- NOTAS: Costo del ítem = item_cantidad * item_precio; factura sin ítems suma 0 (ISNULL). Sin subselect en el
-- FROM: un grupo por año y dos mandaderos con la misma condición. Devuelve 3 filas: 2010 1/1, 2011 16/79, 2012
-- 29/892 (las 892 verificadas con un GROUP BY + HAVING aparte).
SELECT YEAR(f.fact_fecha)                                         AS anio,
       ( SELECT COUNT(DISTINCT f2.fact_cliente)                    -- clientes con al menos una factura mal hecha
         FROM Factura f2
         WHERE YEAR(f2.fact_fecha) = YEAR(f.fact_fecha)
           AND ABS( f2.fact_total - f2.fact_total_impuestos
                    - ISNULL(( SELECT SUM(i.item_cantidad * i.item_precio)
                               FROM Item_Factura i
                               WHERE i.item_tipo = f2.fact_tipo AND i.item_sucursal = f2.fact_sucursal
                                 AND i.item_numero = f2.fact_numero ), 0) ) > 1 )  AS clientes_mal_facturados,
       ( SELECT COUNT(*)                                           -- facturas mal hechas (misma condicion)
         FROM Factura f2
         WHERE YEAR(f2.fact_fecha) = YEAR(f.fact_fecha)
           AND ABS( f2.fact_total - f2.fact_total_impuestos
                    - ISNULL(( SELECT SUM(i.item_cantidad * i.item_precio)
                               FROM Item_Factura i
                               WHERE i.item_tipo = f2.fact_tipo AND i.item_sucursal = f2.fact_sucursal
                                 AND i.item_numero = f2.fact_numero ), 0) ) > 1 )  AS facturas_mal_hechas
FROM Factura f                                                     -- un grupo por cada año con facturas
GROUP BY YEAR(f.fact_fecha)
ORDER BY 1
```

```sql
-- CLASE 4 - EJERCICIO 22: Escriba una consulta sql que retorne una estadistica de venta para todos los rubros por
-- trimestre contabilizando todos los años. Se mostraran como maximo 4 filas por rubro (1 por cada trimestre). Se
-- deben mostrar 4 columnas: Detalle del rubro; Numero de trimestre del año (1 a 4); Cantidad de facturas emitidas
-- en el trimestre en las que se haya vendido al menos un producto del rubro; Cantidad de productos diferentes del
-- rubro vendidos en el trimestre. El resultado debe ser ordenado alfabeticamente por el detalle del rubro y dentro
-- de cada rubro primero el trimestre en el que mas facturas se emitieron. No se deberan mostrar aquellos rubros y
-- trimestres para los cuales las facturas emitidas no superen las 100. En ningun momento se tendran en cuenta los
-- productos compuestos para esta estadistica.
-- NOTAS: Compuestos = productos que figuran como comp_producto (se excluyen del conteo de facturas y de
-- productos). Devuelve 17 filas en la base (ej. ALFAJORES T2: 203 facturas, 15 productos).
SELECT
    r.rubr_detalle                                                         AS rubro,
    DATEPART(QUARTER, f.fact_fecha)                                        AS trimestre,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)          AS cant_facturas,      -- facturas con al menos un producto del rubro
    COUNT(DISTINCT i.item_producto)                                        AS cant_productos      -- productos distintos del rubro
FROM Rubro r
JOIN Producto p     ON p.prod_rubro = r.rubr_id
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE p.prod_codigo NOT IN ( SELECT comp_producto FROM Composicion )      -- nunca se cuentan los compuestos
GROUP BY r.rubr_id, r.rubr_detalle, DATEPART(QUARTER, f.fact_fecha)      -- todos los años juntos: solo el trimestre
HAVING COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero) > 100
ORDER BY r.rubr_detalle, cant_facturas DESC;                              -- dentro del rubro, primero el trimestre con más facturas
```

```sql
-- CLASE 4 - EJERCICIO 24: Escriba una consulta que considerando solamente las facturas correspondientes a los dos
-- vendedores con mayores comisiones, retorne los productos con composicion facturados al menos en cinco facturas.
-- La consulta debe retornar las siguientes columnas: Codigo de Producto; Nombre del Producto; Unidades facturadas.
-- El resultado debera ser ordenado por las unidades facturadas descendente.
-- NOTAS: Vendedores con mayor comision = TOP 2 por empl_comision (empleados 3 y 8, sin empate). Devuelve 5 filas
-- en la base (00001718 Linterna con pilas: 5595 unidades).
SELECT
    p.prod_codigo,
    p.prod_detalle,
    SUM(i.item_cantidad)                                                   AS unidades_facturadas
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE f.fact_vendedor IN ( SELECT TOP 2 e.empl_codigo                     -- los 2 vendedores con mayor comisión
                           FROM Empleado e
                           ORDER BY e.empl_comision DESC )
  AND p.prod_codigo IN ( SELECT comp_producto FROM Composicion )          -- solo productos con composición
GROUP BY p.prod_codigo, p.prod_detalle
HAVING COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero) >= 5 -- facturado en al menos 5 facturas
ORDER BY unidades_facturadas DESC;
```

```sql
-- CLASE 4 - EJERCICIO 29: Se solicita que realice una estadistica de venta por producto para el año 2011, solo
-- para los productos que pertenezcan a las familias que tengan mas de 20 productos asignados a ellas, la cual
-- debera devolver las siguientes columnas: a. Codigo de producto; b. Descripcion del producto; c. Cantidad
-- vendida; d. Cantidad de facturas en la que esta ese producto; e. Monto total facturado de ese producto. Solo se
-- debera mostrar un producto por fila en funcion a los considerandos establecidos antes. El resultado debera ser
-- ordenado por la cantidad vendida de mayor a menor.
-- NOTAS: Monto = SUM(item_cantidad*item_precio) (no fact_total). Devuelve 263 filas en la base (primero 00010395
-- TANG NARANJA: 4298 unidades en 74 facturas).
SELECT
    p.prod_codigo,
    p.prod_detalle,
    SUM(i.item_cantidad)                                                   AS cant_vendida,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)          AS cant_facturas,
    SUM(i.item_cantidad * i.item_precio)                                   AS monto_total     -- monto del producto, no fact_total
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE YEAR(f.fact_fecha) = 2011
  AND p.prod_familia IN ( SELECT p2.prod_familia                          -- familias con más de 20 productos
                          FROM Producto p2
                          GROUP BY p2.prod_familia
                          HAVING COUNT(*) > 20 )
GROUP BY p.prod_codigo, p.prod_detalle                                    -- una fila por producto
ORDER BY cant_vendida DESC;
```

```sql
-- CLASE 4 - EJERCICIO 32: Se desea conocer las familias que sus productos se facturaron juntos en las mismas
-- facturas, para ello se solicita que escriba una consulta sql que retorne los pares de familias que tienen
-- productos que se facturaron juntos. Para ellos debera devolver las siguientes columnas: Codigo de familia;
-- Detalle de familia; Codigo de familia; Detalle de familia; Cantidad de facturas; Total vendido. Los datos
-- deberan ser ordenados por Total vendido y solo se deben mostrar las familias que se vendieron juntas mas de 10
-- veces.
-- NOTAS: Cada par una vez (fami1 < fami2). 'Mas de 10 veces' = mas de 10 facturas distintas. Total vendido = suma
-- de los renglones de ambas familias en las facturas donde aparecen juntas (mandadero para no inflar el SUM con el
-- self-join); orden descendente. Devuelve 335 filas en la base (top: MARLBORO-HOGAR, 569 facturas).
SELECT
    fa1.fami_id, fa1.fami_detalle,
    fa2.fami_id, fa2.fami_detalle,
    COUNT(DISTINCT i1.item_tipo + i1.item_sucursal + i1.item_numero)      AS cant_facturas,   -- facturas donde aparecen juntas
    ( SELECT SUM(i.item_cantidad * i.item_precio)                         -- mandadero: el SUM del self-join saldría multiplicado
      FROM Item_Factura i
      JOIN Producto p ON p.prod_codigo = i.item_producto
      WHERE p.prod_familia IN (fa1.fami_id, fa2.fami_id)                  -- renglones de cualquiera de las dos familias...
        AND EXISTS ( SELECT 1 FROM Item_Factura ia JOIN Producto pa ON pa.prod_codigo = ia.item_producto
                     WHERE ia.item_tipo = i.item_tipo AND ia.item_sucursal = i.item_sucursal AND ia.item_numero = i.item_numero
                       AND pa.prod_familia = fa1.fami_id )                -- ...en facturas que tienen la familia 1
        AND EXISTS ( SELECT 1 FROM Item_Factura ib JOIN Producto pb ON pb.prod_codigo = ib.item_producto
                     WHERE ib.item_tipo = i.item_tipo AND ib.item_sucursal = i.item_sucursal AND ib.item_numero = i.item_numero
                       AND pb.prod_familia = fa2.fami_id ) )              -- ...y también la familia 2
                                                                           AS total_vendido
FROM Item_Factura i1
JOIN Item_Factura i2 ON i2.item_tipo = i1.item_tipo AND i2.item_sucursal = i1.item_sucursal AND i2.item_numero = i1.item_numero  -- misma factura
JOIN Producto p1     ON p1.prod_codigo = i1.item_producto
JOIN Producto p2     ON p2.prod_codigo = i2.item_producto
JOIN Familia fa1     ON fa1.fami_id = p1.prod_familia
JOIN Familia fa2     ON fa2.fami_id = p2.prod_familia
WHERE fa1.fami_id < fa2.fami_id                                           -- cada par una sola vez y sin (A,A)
GROUP BY fa1.fami_id, fa1.fami_detalle, fa2.fami_id, fa2.fami_detalle
HAVING COUNT(DISTINCT i1.item_tipo + i1.item_sucursal + i1.item_numero) > 10   -- "más de 10 veces" = más de 10 facturas
ORDER BY total_vendido DESC;
```

```sql
-- CLASE 4 - EJERCICIO 33: Se requiere obtener una estadistica de venta de productos que sean componentes. Para
-- ello se solicita que realice la siguiente consulta que retorne la venta de los componentes del producto mas
-- vendido del año 2012. Se debera mostrar: a. Codigo de producto; b. Nombre del producto; c. Cantidad de unidades
-- vendidas; d. Cantidad de facturas en la cual se facturo; e. Precio promedio facturado de ese producto; f. Total
-- facturado para ese producto. El resultado debera ser ordenado por el total vendido por producto para el año
-- 2012.
-- NOTAS: Mas vendido = mas unidades en 2012 (00001718, que es compuesto); las columnas se calculan sobre 2012.
-- Devuelve 1 fila (00001420 PILAS: 5860 u., 245 facturas): el otro componente, 00001475, nunca se vendio.
SELECT
    p.prod_codigo,
    p.prod_detalle,
    SUM(i.item_cantidad)                                                   AS unidades_vendidas,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)          AS cant_facturas,
    AVG(i.item_precio)                                                     AS precio_promedio,
    SUM(i.item_cantidad * i.item_precio)                                   AS total_facturado
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE YEAR(f.fact_fecha) = 2012
  AND p.prod_codigo IN ( SELECT c.comp_componente                         -- componentes del...
                         FROM Composicion c
                         WHERE c.comp_producto = ( SELECT TOP 1 i2.item_producto   -- ...producto más vendido (unidades) de 2012
                                                   FROM Item_Factura i2
                                                   JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
                                                   WHERE YEAR(f2.fact_fecha) = 2012
                                                   GROUP BY i2.item_producto
                                                   ORDER BY SUM(i2.item_cantidad) DESC ) )
GROUP BY p.prod_codigo, p.prod_detalle
ORDER BY total_facturado DESC;
```

```sql
-- CLASE 4 - EJERCICIO 34: Escriba una consulta sql que retorne para todos los rubros la cantidad de facturas mal
-- facturadas por cada mes del año 2011. Se considera que una factura es incorrecta cuando en la misma factura se
-- facturan productos de dos rubros diferentes. Si no hay facturas mal hechas se debe retornar 0. Las columnas que
-- se deben mostrar son: 1- Codigo de Rubro; 2- Mes; 3- Cantidad de facturas mal realizadas.
-- NOTAS: Una factura mal hecha cuenta para un rubro si tiene un producto de ese rubro y otro de un rubro distinto.
-- Los 12 meses salen de una tabla armada con UNION ALL en el FROM (patrón de los apuntes, cap. 7.8: el enunciado
-- no deja otra), en CROSS JOIN con todos los rubros. Devuelve 372 filas (31 rubros x 12 meses); enero a junio dan
-- 0 porque 2011 tiene facturas recién desde julio. Verificado a mano: rubro 0001 en diciembre da 43.
SELECT
    r.rubr_id,
    m.mes,
    ( SELECT COUNT(*)                                                      -- mandadero: facturas mal hechas del rubro en ese mes (0 si no hay)
      FROM Factura f
      WHERE YEAR(f.fact_fecha) = 2011
        AND MONTH(f.fact_fecha) = m.mes
        AND EXISTS ( SELECT 1 FROM Item_Factura ia JOIN Producto pa ON pa.prod_codigo = ia.item_producto
                     WHERE ia.item_tipo = f.fact_tipo AND ia.item_sucursal = f.fact_sucursal AND ia.item_numero = f.fact_numero
                       AND pa.prod_rubro = r.rubr_id )                    -- tiene un producto del rubro...
        AND EXISTS ( SELECT 1 FROM Item_Factura ib JOIN Producto pb ON pb.prod_codigo = ib.item_producto
                     WHERE ib.item_tipo = f.fact_tipo AND ib.item_sucursal = f.fact_sucursal AND ib.item_numero = f.fact_numero
                       AND pb.prod_rubro <> r.rubr_id ) )                 -- ...y otro de un rubro distinto
                                                                           AS facturas_mal_hechas
FROM Rubro r
CROSS JOIN ( SELECT 1 AS mes UNION ALL SELECT 2 UNION ALL SELECT 3 UNION ALL SELECT 4
             UNION ALL SELECT 5 UNION ALL SELECT 6 UNION ALL SELECT 7 UNION ALL SELECT 8
             UNION ALL SELECT 9 UNION ALL SELECT 10 UNION ALL SELECT 11 UNION ALL SELECT 12 ) AS m   -- la tabla de los 12 meses que no existe
ORDER BY r.rubr_id, m.mes;
```

```sql
-- CLASE 4 - EJERCICIO 35: Se requiere realizar una estadistica de ventas por año y producto, para ello se solicita
-- que escriba una consulta sql que retorne las siguientes columnas: Año; Codigo de producto; Detalle del producto;
-- Cantidad de facturas emitidas a ese producto ese año; Cantidad de vendedores diferentes que compraron ese
-- producto ese año; Cantidad de productos a los cuales compone ese producto, si no compone a ninguno se debera
-- retornar 0; Porcentaje de la venta de ese producto respecto a la venta total de ese año. Los datos deberan ser
-- ordenados por año y por producto con mayor cantidad vendida.
-- NOTAS: Vendedores = COUNT(DISTINCT fact_vendedor). Venta = SUM(item_cantidad*item_precio); % contra el mismo
-- calculo para todo el año (mandadero). Devuelve 1016 filas; los % suman ~100 por año.
SELECT
    YEAR(f.fact_fecha)                                                     AS anio,
    p.prod_codigo,
    p.prod_detalle,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)          AS cant_facturas,
    COUNT(DISTINCT f.fact_vendedor)                                        AS cant_vendedores,
    ( SELECT COUNT(*) FROM Composicion c
      WHERE c.comp_componente = p.prod_codigo )                            AS compone_a,        -- COUNT ya da 0 si no compone a ninguno
    SUM(i.item_cantidad * i.item_precio) * 100
      / ( SELECT SUM(i2.item_cantidad * i2.item_precio)                   -- venta total del año (mandadero por año)
          FROM Item_Factura i2
          JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
          WHERE YEAR(f2.fact_fecha) = YEAR(f.fact_fecha) )                 AS porc_venta_anio
FROM Producto p
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura f      ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
GROUP BY YEAR(f.fact_fecha), p.prod_codigo, p.prod_detalle
ORDER BY anio, SUM(i.item_cantidad) DESC;                                 -- por año y dentro, el más vendido primero
```

```sql
-- CLASE 5 - EJERCICIO 10: Mostrar los 10 productos más vendidos en la historia y también los 10 productos menos
-- vendidos en la historia. Además mostrar de esos productos, quién fue el cliente que mayor compra realizó.
-- NOTAS: Versión con ROW_NUMBER(): dos numeraciones (DESC y ASC) por unidades vendidas en un subselect del FROM
-- (único caso permitido: filtrar por el número de fila); empates desempatados por código de producto. 'Mayor
-- compra' = cliente con más unidades de ese producto (mandadero TOP 1, desempata menor código). Devuelve 20 filas:
-- arriba 00001718 (31085 u., cliente 01742); los 10 menos vendidos tienen 1 unidad cada uno.
SELECT r.item_producto, r.prod_detalle, r.unidades,
       CASE WHEN r.puesto_mas <= 10 THEN 'MAS VENDIDO' ELSE 'MENOS VENDIDO' END AS ranking,
       ( SELECT TOP 1 f2.fact_cliente                                -- mandadero: el cliente que mas unidades compro
         FROM Item_Factura i2
         JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
         WHERE i2.item_producto = r.item_producto
         GROUP BY f2.fact_cliente
         ORDER BY SUM(i2.item_cantidad) DESC, f2.fact_cliente ) AS cliente_mayor_compra
FROM ( SELECT i.item_producto, p.prod_detalle,
              SUM(i.item_cantidad)                                                    AS unidades,
              ROW_NUMBER() OVER ( ORDER BY SUM(i.item_cantidad) DESC, i.item_producto ) AS puesto_mas,    -- 1 = el mas vendido
              ROW_NUMBER() OVER ( ORDER BY SUM(i.item_cantidad) ASC,  i.item_producto ) AS puesto_menos   -- 1 = el menos vendido
       FROM Item_Factura i
       JOIN Producto p ON p.prod_codigo = i.item_producto
       GROUP BY i.item_producto, p.prod_detalle ) r                -- el subselect en el FROM vale para filtrar por ROW_NUMBER
WHERE r.puesto_mas <= 10 OR r.puesto_menos <= 10
ORDER BY r.unidades DESC
```

```sql
-- CLASE 5 - EJERCICIO 16: Con el fin de lanzar una nueva campaña comercial para los clientes que menos compran en
-- la empresa, se pide una consulta SQL que retorne aquellos clientes cuyas ventas son inferiores a 1/3 del
-- promedio de ventas del producto que más se vendió en el 2012. Además mostrar: 1. Nombre del Cliente. 2. Cantidad
-- de unidades totales vendidas en el 2012 para ese cliente. 3. Código de producto que mayor venta tuvo en el 2012
-- (en caso de existir más de 1, mostrar solamente el de menor código) para ese cliente. Aclaraciones: la
-- composición es de 2 niveles, es decir, un producto compuesto solo se compone de productos no compuestos. Los
-- clientes deben ser ordenados por código de provincia ascendente.
-- NOTAS: "Ventas" = unidades vendidas al cliente en 2012; "promedio de ventas del producto más vendido" = unidades
-- de 2012 del producto con más unidades (00001718) dividido la cantidad de clientes que lo compraron (1440 u.), y
-- el umbral es 1/3 de eso (480 u.); con AVG por renglón (46 u.) el umbral queda en 15 u. y no sale ningún cliente
-- (el mínimo compró 72). Solo clientes con compras en 2012; la aclaración de composición no cambia nada porque se
-- cuentan las unidades facturadas. Cliente no tiene provincia (y la zona del departamento del vendedor es la misma
-- para los 3665 clientes), así que se ordena por clie_codigo. Devuelve 13 filas en la base.
SELECT
    c.clie_razon_social                                          AS cliente,
    SUM(i.item_cantidad)                                         AS unidades_2012,
    ( SELECT TOP 1 i2.item_producto                              -- mandadero: el producto que más compró en 2012
      FROM Item_Factura i2
      JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
      WHERE f2.fact_cliente = c.clie_codigo AND YEAR(f2.fact_fecha) = 2012
      GROUP BY i2.item_producto
      ORDER BY SUM(i2.item_cantidad) DESC, i2.item_producto )    -- desempate: menor código
                                                                 AS producto_mas_comprado
FROM Cliente c
JOIN Factura f      ON f.fact_cliente = c.clie_codigo
JOIN Item_Factura i ON i.item_tipo = f.fact_tipo AND i.item_sucursal = f.fact_sucursal AND i.item_numero = f.fact_numero
WHERE YEAR(f.fact_fecha) = 2012
GROUP BY c.clie_codigo, c.clie_razon_social
HAVING SUM(i.item_cantidad) <
       ( SELECT TOP 1 SUM(i3.item_cantidad) / COUNT(DISTINCT f3.fact_cliente)   -- promedio por cliente del más vendido de 2012
         FROM Item_Factura i3
         JOIN Factura f3 ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
         WHERE YEAR(f3.fact_fecha) = 2012
         GROUP BY i3.item_producto
         ORDER BY SUM(i3.item_cantidad) DESC ) / 3
ORDER BY c.clie_codigo;                                          -- no hay provincia en el esquema
```

```sql
-- CLASE 5 - EJERCICIO 18: Escriba una consulta que retorne una estadística de ventas para todos los rubros. La
-- consulta debe retornar: DETALLE_RUBRO: detalle del rubro; VENTAS: suma de las ventas en pesos de productos
-- vendidos de dicho rubro; PROD1: código del producto más vendido de dicho rubro; PROD2: código del segundo
-- producto más vendido de dicho rubro; CLIENTE: código del cliente que compró más productos del rubro en los
-- últimos 30 días. La consulta no puede mostrar NULL en ninguna de sus columnas y debe estar ordenada por cantidad
-- de productos diferentes vendidos del rubro.
-- NOTAS: "Más vendido" = más unidades (desempate por código). Los "últimos 30 días" se toman respecto de la última
-- fecha facturada (los datos terminan en 2012; en el parcial sería GETDATE()). Los NULL se reemplazan por 0 o '-'.
-- Orden descendente por productos distintos vendidos. Devuelve 31 filas (todos los rubros) en la base.
SELECT
    r.rubr_detalle                                                       AS DETALLE_RUBRO,
    ISNULL(( SELECT SUM(i.item_cantidad * i.item_precio)                 -- monto por producto: cantidad * precio
             FROM Item_Factura i
             JOIN Producto p ON p.prod_codigo = i.item_producto
             WHERE p.prod_rubro = r.rubr_id ), 0)                        AS VENTAS,
    ISNULL(( SELECT TOP 1 i.item_producto
             FROM Item_Factura i
             JOIN Producto p ON p.prod_codigo = i.item_producto
             WHERE p.prod_rubro = r.rubr_id
             GROUP BY i.item_producto
             ORDER BY SUM(i.item_cantidad) DESC, i.item_producto ), '-') AS PROD1,
    ISNULL(( SELECT TOP 1 i.item_producto                                -- el más vendido excluyendo al primero
             FROM Item_Factura i
             JOIN Producto p ON p.prod_codigo = i.item_producto
             WHERE p.prod_rubro = r.rubr_id
               AND i.item_producto NOT IN ( SELECT TOP 1 i2.item_producto
                                            FROM Item_Factura i2
                                            JOIN Producto p2 ON p2.prod_codigo = i2.item_producto
                                            WHERE p2.prod_rubro = r.rubr_id
                                            GROUP BY i2.item_producto
                                            ORDER BY SUM(i2.item_cantidad) DESC, i2.item_producto )
             GROUP BY i.item_producto
             ORDER BY SUM(i.item_cantidad) DESC, i.item_producto ), '-') AS PROD2,
    ISNULL(( SELECT TOP 1 f.fact_cliente
             FROM Item_Factura i
             JOIN Factura f  ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
             JOIN Producto p ON p.prod_codigo = i.item_producto
             WHERE p.prod_rubro = r.rubr_id
               AND f.fact_fecha >= DATEADD(DAY, -30, ( SELECT MAX(fact_fecha) FROM Factura ))  -- últimos 30 días
             GROUP BY f.fact_cliente
             ORDER BY SUM(i.item_cantidad) DESC, f.fact_cliente ), '-')  AS CLIENTE
FROM Rubro r
ORDER BY ( SELECT COUNT(DISTINCT i.item_producto)                        -- productos diferentes vendidos del rubro
           FROM Item_Factura i
           JOIN Producto p ON p.prod_codigo = i.item_producto
           WHERE p.prod_rubro = r.rubr_id ) DESC;
```

```sql
-- CLASE 5 - EJERCICIO 19: En virtud de una recategorización de productos referida a la familia de los mismos se
-- solicita que desarrolle una consulta SQL que retorne para todos los productos: código de producto, detalle del
-- producto, código de la familia del producto, detalle de la familia actual del producto, código de la familia
-- sugerido para el producto, detalle de la familia sugerido para el producto. La familia sugerida para un producto
-- es la que poseen la mayoría de los productos cuyo detalle coinciden en los primeros 5 caracteres. En caso que 2
-- o más familias pudieran ser sugeridas se deberá seleccionar la de menor código. Solo se deben mostrar los
-- productos para los cuales la familia actual sea diferente a la sugerida. Los resultados deben ser ordenados por
-- detalle de producto de manera ascendente.
-- NOTAS: El grupo de "productos con los mismos 5 primeros caracteres" incluye al propio producto. La familia
-- sugerida se calcula con un mandadero TOP 1 usado en el ON del JOIN a Familia (no es subselect en el FROM).
-- Devuelve 344 filas (ej. los "ALF. " de familia 8 2 pasan a 8 1, que tiene 35 de 55) en la base.
SELECT
    p.prod_codigo,
    p.prod_detalle,
    fa.fami_id       AS familia_actual,
    fa.fami_detalle  AS detalle_familia_actual,
    fs.fami_id       AS familia_sugerida,
    fs.fami_detalle  AS detalle_familia_sugerida
FROM Producto p
JOIN Familia fa ON fa.fami_id = p.prod_familia
JOIN Familia fs ON fs.fami_id = ( SELECT TOP 1 p2.prod_familia          -- la familia mayoritaria entre los "parecidos"
                                  FROM Producto p2
                                  WHERE LEFT(p2.prod_detalle, 5) = LEFT(p.prod_detalle, 5)
                                  GROUP BY p2.prod_familia
                                  ORDER BY COUNT(*) DESC, p2.prod_familia )   -- desempate: menor código
WHERE fs.fami_id <> p.prod_familia                                      -- solo los que cambian
ORDER BY p.prod_detalle;
```

```sql
-- CLASE 5 - EJERCICIO 23: Realizar una consulta SQL que para cada año muestre: año; el producto con composición
-- más vendido para ese año; cantidad de productos que componen directamente al producto más vendido; la cantidad
-- de facturas en las cuales aparece ese producto; el código de cliente que más compró ese producto; el porcentaje
-- que representa la venta de ese producto respecto al total de venta del año. El resultado deberá ser ordenado por
-- el total vendido por año en forma descendente.
-- NOTAS: "Más vendido" = más unidades en el año (desempate por código); el % es en pesos (cantidad * precio). Se
-- agrupa por (año, producto) y el HAVING deja solo el compuesto más vendido de cada año. Devuelve 3 filas (2010,
-- 2011, 2012) en la base.
SELECT
    YEAR(f.fact_fecha)                                                    AS anio,
    i.item_producto                                                       AS producto,
    ( SELECT COUNT(*) FROM Composicion c
      WHERE c.comp_producto = i.item_producto )                           AS cant_componentes,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)         AS cant_facturas,
    ( SELECT TOP 1 f2.fact_cliente                                        -- el cliente que más lo compró ese año
      FROM Item_Factura i2
      JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
      WHERE i2.item_producto = i.item_producto AND YEAR(f2.fact_fecha) = YEAR(f.fact_fecha)
      GROUP BY f2.fact_cliente
      ORDER BY SUM(i2.item_cantidad) DESC, f2.fact_cliente )              AS cliente_top,
    SUM(i.item_cantidad * i.item_precio) * 100.0
      / ( SELECT SUM(i3.item_cantidad * i3.item_precio)                   -- total vendido en el año
          FROM Item_Factura i3
          JOIN Factura f3 ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
          WHERE YEAR(f3.fact_fecha) = YEAR(f.fact_fecha) )                AS porcentaje_anio
FROM Item_Factura i
JOIN Factura f ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
GROUP BY YEAR(f.fact_fecha), i.item_producto
HAVING i.item_producto = ( SELECT TOP 1 i4.item_producto                  -- el compuesto más vendido de ese año
                           FROM Item_Factura i4
                           JOIN Factura f4 ON f4.fact_tipo = i4.item_tipo AND f4.fact_sucursal = i4.item_sucursal AND f4.fact_numero = i4.item_numero
                           WHERE YEAR(f4.fact_fecha) = YEAR(f.fact_fecha)
                             AND i4.item_producto IN ( SELECT comp_producto FROM Composicion )
                           GROUP BY i4.item_producto
                           ORDER BY SUM(i4.item_cantidad) DESC, i4.item_producto )
ORDER BY ( SELECT SUM(i5.item_cantidad * i5.item_precio)                  -- total vendido del año
           FROM Item_Factura i5
           JOIN Factura f5 ON f5.fact_tipo = i5.item_tipo AND f5.fact_sucursal = i5.item_sucursal AND f5.fact_numero = i5.item_numero
           WHERE YEAR(f5.fact_fecha) = YEAR(f.fact_fecha) ) DESC;
```

```sql
-- CLASE 5 - EJERCICIO 25: Realizar una consulta SQL que para cada año y familia muestre: a. Año. b. El código de
-- la familia más vendida en ese año. c. Cantidad de rubros que componen esa familia. d. Cantidad de productos que
-- componen directamente al producto más vendido de esa familia. e. La cantidad de facturas en las cuales aparecen
-- productos pertenecientes a esa familia. f. El código de cliente que más compró productos de esa familia. g. El
-- porcentaje que representa la venta de esa familia respecto al total de venta del año. El resultado deberá ser
-- ordenado por el total vendido por año y familia en forma descendente.
-- NOTAS: Una fila por año con su familia más vendida (en unidades, desempate por código). Rubros = rubros
-- distintos de los productos de la familia (Producto); el producto más vendido de la familia y el cliente top se
-- calculan dentro del año; el % es en pesos. Orden: total del año y luego total de la familia, descendente.
-- Devuelve 3 filas en la base.
SELECT
    YEAR(f.fact_fecha)                                                    AS anio,
    p.prod_familia                                                        AS familia,
    ( SELECT COUNT(DISTINCT p2.prod_rubro) FROM Producto p2
      WHERE p2.prod_familia = p.prod_familia )                            AS cant_rubros,
    ( SELECT COUNT(*) FROM Composicion c                                  -- componentes del más vendido de la familia
      WHERE c.comp_producto = ( SELECT TOP 1 i2.item_producto
                                FROM Item_Factura i2
                                JOIN Factura f2  ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
                                JOIN Producto p2 ON p2.prod_codigo = i2.item_producto
                                WHERE p2.prod_familia = p.prod_familia AND YEAR(f2.fact_fecha) = YEAR(f.fact_fecha)
                                GROUP BY i2.item_producto
                                ORDER BY SUM(i2.item_cantidad) DESC, i2.item_producto ) ) AS cant_componentes,
    COUNT(DISTINCT f.fact_tipo + f.fact_sucursal + f.fact_numero)         AS cant_facturas,
    ( SELECT TOP 1 f3.fact_cliente                                        -- el cliente que más compró de la familia
      FROM Item_Factura i3
      JOIN Factura f3  ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
      JOIN Producto p3 ON p3.prod_codigo = i3.item_producto
      WHERE p3.prod_familia = p.prod_familia AND YEAR(f3.fact_fecha) = YEAR(f.fact_fecha)
      GROUP BY f3.fact_cliente
      ORDER BY SUM(i3.item_cantidad) DESC, f3.fact_cliente )              AS cliente_top,
    SUM(i.item_cantidad * i.item_precio) * 100.0
      / ( SELECT SUM(i4.item_cantidad * i4.item_precio)                   -- total vendido en el año
          FROM Item_Factura i4
          JOIN Factura f4 ON f4.fact_tipo = i4.item_tipo AND f4.fact_sucursal = i4.item_sucursal AND f4.fact_numero = i4.item_numero
          WHERE YEAR(f4.fact_fecha) = YEAR(f.fact_fecha) )                AS porcentaje_anio
FROM Item_Factura i
JOIN Factura f  ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
JOIN Producto p ON p.prod_codigo = i.item_producto
GROUP BY YEAR(f.fact_fecha), p.prod_familia
HAVING p.prod_familia = ( SELECT TOP 1 p5.prod_familia                    -- la familia más vendida de ese año
                          FROM Item_Factura i5
                          JOIN Factura f5  ON f5.fact_tipo = i5.item_tipo AND f5.fact_sucursal = i5.item_sucursal AND f5.fact_numero = i5.item_numero
                          JOIN Producto p5 ON p5.prod_codigo = i5.item_producto
                          WHERE YEAR(f5.fact_fecha) = YEAR(f.fact_fecha)
                          GROUP BY p5.prod_familia
                          ORDER BY SUM(i5.item_cantidad) DESC, p5.prod_familia )
ORDER BY ( SELECT SUM(i6.item_cantidad * i6.item_precio)                  -- 1ro: total del año
           FROM Item_Factura i6
           JOIN Factura f6 ON f6.fact_tipo = i6.item_tipo AND f6.fact_sucursal = i6.item_sucursal AND f6.fact_numero = i6.item_numero
           WHERE YEAR(f6.fact_fecha) = YEAR(f.fact_fecha) ) DESC,
         SUM(i.item_cantidad * i.item_precio) DESC;                       -- 2do: total de la familia
```

```sql
-- CLASE 5 - EJERCICIO 26: Escriba una consulta SQL que retorne un ranking de empleados devolviendo las siguientes
-- columnas: empleado; depósitos que tiene a cargo; monto total facturado en el año corriente; código de cliente al
-- que más le vendió; producto más vendido; porcentaje de la venta de ese empleado sobre el total vendido ese año.
-- Los datos deberán ser ordenados por venta del empleado de mayor a menor.
-- NOTAS: "Año corriente" = 2012 (último año con datos; en el parcial sería YEAR(GETDATE())). Se listan todos los
-- empleados (los que no vendieron, con monto 0 y cliente/producto NULL). Monto y % por fact_total (sin JOIN a
-- items, no se infla); cliente top por monto y producto top por unidades vendidas por el empleado en 2012.
-- Devuelve 9 filas en la base (los 9 empleados).
SELECT
    e.empl_codigo,
    RTRIM(e.empl_apellido) + ', ' + RTRIM(e.empl_nombre)                  AS empleado,
    ( SELECT COUNT(*) FROM DEPOSITO d
      WHERE d.depo_encargado = e.empl_codigo )                            AS depositos_a_cargo,
    ISNULL(SUM(f.fact_total), 0)                                          AS monto_anio,
    ( SELECT TOP 1 f2.fact_cliente                                        -- cliente al que más le vendió
      FROM Factura f2
      WHERE f2.fact_vendedor = e.empl_codigo AND YEAR(f2.fact_fecha) = 2012
      GROUP BY f2.fact_cliente
      ORDER BY SUM(f2.fact_total) DESC, f2.fact_cliente )                 AS cliente_top,
    ( SELECT TOP 1 i3.item_producto                                       -- producto que más vendió
      FROM Item_Factura i3
      JOIN Factura f3 ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
      WHERE f3.fact_vendedor = e.empl_codigo AND YEAR(f3.fact_fecha) = 2012
      GROUP BY i3.item_producto
      ORDER BY SUM(i3.item_cantidad) DESC, i3.item_producto )             AS producto_top,
    ISNULL(SUM(f.fact_total), 0) * 100.0
      / ( SELECT SUM(f4.fact_total) FROM Factura f4
          WHERE YEAR(f4.fact_fecha) = 2012 )                              AS porcentaje_anio
FROM Empleado e
LEFT JOIN Factura f ON f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2012   -- el filtro de la dominada va en el ON
GROUP BY e.empl_codigo, e.empl_apellido, e.empl_nombre
ORDER BY monto_anio DESC;
```

```sql
-- CLASE 5 - EJERCICIO 27: Escriba una consulta SQL que retorne una estadística basada en la facturación por año y
-- envase devolviendo las siguientes columnas: año; código de envase; detalle del envase; cantidad de productos que
-- tienen ese envase; cantidad de productos facturados de ese envase; producto más vendido de ese envase; monto
-- total de venta de ese envase en ese año; porcentaje de la venta de ese envase respecto al total vendido de ese
-- año. Los datos deberán ser ordenados por año y dentro del año por el envase con más facturación de mayor a
-- menor.
-- NOTAS: "Productos facturados" = productos distintos de ese envase vendidos en el año (COUNT DISTINCT); "más
-- vendido" = más unidades en el año (desempate por código); montos en pesos (cantidad * precio). Solo
-- combinaciones año/envase con ventas. Devuelve 7 filas en la base (3 envases x 3 años, salvo 2010 que solo vendió
-- BOLSAS); los % de cada año suman 100.
SELECT
    YEAR(f.fact_fecha)                                                    AS anio,
    en.enva_codigo,
    en.enva_detalle,
    ( SELECT COUNT(*) FROM Producto p2
      WHERE p2.prod_envase = en.enva_codigo )                             AS cant_productos,
    COUNT(DISTINCT i.item_producto)                                       AS cant_productos_facturados,
    ( SELECT TOP 1 i3.item_producto                                       -- el más vendido del envase en el año
      FROM Item_Factura i3
      JOIN Factura f3  ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
      JOIN Producto p3 ON p3.prod_codigo = i3.item_producto
      WHERE p3.prod_envase = en.enva_codigo AND YEAR(f3.fact_fecha) = YEAR(f.fact_fecha)
      GROUP BY i3.item_producto
      ORDER BY SUM(i3.item_cantidad) DESC, i3.item_producto )             AS producto_top,
    SUM(i.item_cantidad * i.item_precio)                                  AS monto_anio,
    SUM(i.item_cantidad * i.item_precio) * 100.0
      / ( SELECT SUM(i4.item_cantidad * i4.item_precio)                   -- total vendido en el año
          FROM Item_Factura i4
          JOIN Factura f4 ON f4.fact_tipo = i4.item_tipo AND f4.fact_sucursal = i4.item_sucursal AND f4.fact_numero = i4.item_numero
          WHERE YEAR(f4.fact_fecha) = YEAR(f.fact_fecha) )                AS porcentaje_anio
FROM Item_Factura i
JOIN Factura f   ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
JOIN Producto p  ON p.prod_codigo = i.item_producto
JOIN Envases en  ON en.enva_codigo = p.prod_envase
GROUP BY YEAR(f.fact_fecha), en.enva_codigo, en.enva_detalle
ORDER BY anio, monto_anio DESC;
```

```sql
-- CLASE 5 - EJERCICIO 28: Escriba una consulta SQL que retorne una estadística por año y vendedor que retorne las
-- siguientes columnas: año; código de vendedor; detalle del vendedor; cantidad de facturas que realizó en ese año;
-- cantidad de clientes a los cuales les vendió en ese año; cantidad de productos facturados con composición en ese
-- año; cantidad de productos facturados sin composición en ese año; monto total vendido por ese vendedor en ese
-- año. Los datos deberán ser ordenados por año y dentro del año por el vendedor que haya vendido más productos
-- diferentes de mayor a menor. (El Ej. 31 tiene el mismo enunciado.)
-- NOTAS: El FROM es Factura (sin items) para que COUNT y SUM(fact_total) no se inflen; las cantidades de productos
-- con/sin composición son productos DISTINTOS vendidos (mandaderos con IN / NOT IN Composicion) y el orden usa los
-- productos distintos totales. Devuelve 14 filas en la base (2010-2012).
SELECT
    YEAR(f.fact_fecha)                                                    AS anio,
    e.empl_codigo                                                         AS vendedor,
    RTRIM(e.empl_apellido) + ', ' + RTRIM(e.empl_nombre)                  AS detalle_vendedor,
    COUNT(*)                                                              AS cant_facturas,
    COUNT(DISTINCT f.fact_cliente)                                        AS cant_clientes,
    ( SELECT COUNT(DISTINCT i2.item_producto)                             -- productos compuestos vendidos
      FROM Item_Factura i2
      JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
      WHERE f2.fact_vendedor = e.empl_codigo AND YEAR(f2.fact_fecha) = YEAR(f.fact_fecha)
        AND i2.item_producto IN ( SELECT comp_producto FROM Composicion ) )     AS cant_prod_con_composicion,
    ( SELECT COUNT(DISTINCT i3.item_producto)                             -- productos simples vendidos
      FROM Item_Factura i3
      JOIN Factura f3 ON f3.fact_tipo = i3.item_tipo AND f3.fact_sucursal = i3.item_sucursal AND f3.fact_numero = i3.item_numero
      WHERE f3.fact_vendedor = e.empl_codigo AND YEAR(f3.fact_fecha) = YEAR(f.fact_fecha)
        AND i3.item_producto NOT IN ( SELECT comp_producto FROM Composicion ) ) AS cant_prod_sin_composicion,
    SUM(f.fact_total)                                                     AS monto_total
FROM Factura f
JOIN Empleado e ON e.empl_codigo = f.fact_vendedor
GROUP BY YEAR(f.fact_fecha), e.empl_codigo, e.empl_apellido, e.empl_nombre
ORDER BY anio,
         ( SELECT COUNT(DISTINCT i4.item_producto)                        -- productos diferentes vendidos
           FROM Item_Factura i4
           JOIN Factura f4 ON f4.fact_tipo = i4.item_tipo AND f4.fact_sucursal = i4.item_sucursal AND f4.fact_numero = i4.item_numero
           WHERE f4.fact_vendedor = e.empl_codigo AND YEAR(f4.fact_fecha) = YEAR(f.fact_fecha) ) DESC;
```

```sql
-- CLASE 5 - EJERCICIO 30: Se desea obtener una estadística de ventas del año 2012, para los empleados que sean
-- jefes, o sea, que tengan empleados a su cargo; para ello se requiere que realice la consulta que retorne las
-- siguientes columnas: nombre del jefe; cantidad de empleados a cargo; monto total vendido de los empleados a
-- cargo; cantidad de facturas realizadas por los empleados a cargo; nombre del empleado con mejor ventas de ese
-- jefe. Debido a la performance requerida, solo se permite el uso de una subconsulta si fuese necesario. Los datos
-- deberán ser ordenados de mayor a menor por el total vendido y solo se deben mostrar los jefes cuyos subordinados
-- hayan realizado más de 10 facturas.
-- NOTAS: Subordinados = directos (empl_jefe). Única subconsulta: el mandadero del mejor empleado (por monto
-- vendido en 2012). El LEFT JOIN a Factura no infla: cada factura tiene un solo vendedor, así que aparece una vez;
-- empleados a cargo con COUNT(DISTINCT) y "empleados a cargo" cuenta también a los que no vendieron. Devuelve 2
-- filas en la base (Quiroga: 2026 facturas de sus subordinados; Perez: 657, verificado a mano).
SELECT
    RTRIM(j.empl_apellido) + ', ' + RTRIM(j.empl_nombre)                  AS jefe,
    COUNT(DISTINCT e.empl_codigo)                                         AS empleados_a_cargo,
    ISNULL(SUM(f.fact_total), 0)                                          AS monto_total,
    COUNT(f.fact_numero)                                                  AS cant_facturas,
    ( SELECT TOP 1 RTRIM(e2.empl_apellido) + ', ' + RTRIM(e2.empl_nombre) -- la única subconsulta: el mejor subordinado
      FROM Empleado e2
      JOIN Factura f2 ON f2.fact_vendedor = e2.empl_codigo
      WHERE e2.empl_jefe = j.empl_codigo AND YEAR(f2.fact_fecha) = 2012
      GROUP BY e2.empl_codigo, e2.empl_apellido, e2.empl_nombre
      ORDER BY SUM(f2.fact_total) DESC, e2.empl_codigo )                  AS mejor_empleado
FROM Empleado j
JOIN Empleado e     ON e.empl_jefe = j.empl_codigo                        -- self-join: jefe / subordinado
LEFT JOIN Factura f ON f.fact_vendedor = e.empl_codigo AND YEAR(f.fact_fecha) = 2012
GROUP BY j.empl_codigo, j.empl_apellido, j.empl_nombre
HAVING COUNT(f.fact_numero) > 10
ORDER BY monto_total DESC;
```

```sql
-- CLASE 6 - EJERCICIO 1: vista con las ventas de TODOS los clientes
CREATE VIEW dbo.mi_vw_ventas_por_cliente
AS
SELECT
    c.clie_codigo,
    c.clie_razon_social,
    COUNT(f.fact_numero)            AS cantidad_facturas,       -- COUNT(campo del LEFT JOIN) da 0 si nunca compro
    ISNULL(SUM(f.fact_total), 0)    AS total_facturado
FROM Cliente c
LEFT JOIN Factura f ON f.fact_cliente = c.clie_codigo
GROUP BY c.clie_codigo, c.clie_razon_social
GO

SELECT * 
FROM dbo.mi_vw_ventas_por_cliente
WHERE total_facturado > 1000
ORDER BY total_facturado DESC
```

```sql
-- CLASE 6 - EJERCICIO 2: funcion escalar, lo facturado por un cliente en un año
CREATE FUNCTION dbo.mi_total_facturado ( @cliente CHAR(6), @anio INT )
RETURNS DECIMAL(12,2)
AS
BEGIN
    RETURN ISNULL( ( SELECT SUM(fact_total)
                     FROM Factura
                     WHERE fact_cliente = @cliente
                       AND YEAR(fact_fecha) = @anio ), 0 )      -- 0 si no compro
END
GO

SELECT 
    clie_codigo, 
    clie_razon_social, 
    dbo.mi_total_facturado(clie_codigo, 2012) AS total_2012
FROM Cliente
ORDER BY total_2012 DESC
```

```sql
-- CLASE 6 - EJERCICIO 3: en cuantos depositos tiene stock > 0 un producto
CREATE FUNCTION dbo.mi_depositos_con_stock ( @producto CHAR(8) )
RETURNS INT
AS
BEGIN
    RETURN ( SELECT COUNT(*)
             FROM Stock
             WHERE stoc_producto = @producto
               AND stoc_cantidad > 0 )
END
GO

SELECT prod_codigo, prod_detalle, dbo.mi_depositos_con_stock(prod_codigo) AS depositos
FROM Producto
WHERE dbo.mi_depositos_con_stock(prod_codigo) > 3
```

```sql
-- CLASE 6 - EJERCICIO 4: procedimiento con OUTPUT
CREATE PROCEDURE dbo.mi_resumen_vendedor ( @vendedor NUMERIC(6), @anio INT, 
                                           @facturas INT OUTPUT, @total DECIMAL(12,2) OUTPUT )
AS
BEGIN
    SELECT @facturas = COUNT(*),
           @total    = ISNULL(SUM(fact_total), 0)       -- sin GROUP BY el agregado siempre devuelve una fila
    FROM Factura
    WHERE fact_vendedor = @vendedor
      AND YEAR(fact_fecha) = @anio
END
GO

DECLARE @f INT
DECLARE @t DECIMAL(12,2)
EXECUTE dbo.mi_resumen_vendedor @vendedor = 4, @anio = 2012, @facturas = @f OUTPUT, @total = @t OUTPUT
SELECT @f AS facturas, @t AS total
```

```sql
-- CLASE 6 - EJERCICIO 5 (a): la consulta del Ej. 14 de la practica como vista
-- "ultimo año" = el ultimo año con facturas; sin NULLs (ISNULL)
CREATE VIEW dbo.mi_vw_estadistica_clientes
AS
SELECT
    c.clie_codigo,
    ( SELECT COUNT(*) 
      FROM Factura f 
      WHERE f.fact_cliente = c.clie_codigo AND YEAR(f.fact_fecha) = u.anio )                    AS compras,
    ( SELECT ISNULL(AVG(f.fact_total), 0) 
      FROM Factura f 
      WHERE f.fact_cliente = c.clie_codigo AND YEAR(f.fact_fecha) = u.anio )                    AS promedio_compra,
    ( SELECT COUNT(DISTINCT i.item_producto)
      FROM Factura f
      JOIN Item_Factura i ON i.item_tipo = f.fact_tipo AND i.item_sucursal = f.fact_sucursal AND i.item_numero = f.fact_numero
      WHERE f.fact_cliente = c.clie_codigo AND YEAR(f.fact_fecha) = u.anio )                    AS productos_diferentes,
    ( SELECT ISNULL(MAX(f.fact_total), 0) 
      FROM Factura f 
      WHERE f.fact_cliente = c.clie_codigo AND YEAR(f.fact_fecha) = u.anio )                    AS mayor_compra
FROM Cliente c
CROSS JOIN ( SELECT MAX(YEAR(fact_fecha)) AS anio FROM Factura ) u      -- el ORDER BY va al consultar la vista, no adentro
GO

SELECT * FROM dbo.mi_vw_estadistica_clientes ORDER BY compras DESC
```

```sql
-- CLASE 6 - EJERCICIO 5 (b): la consulta del Ej. 21 de la practica como procedimiento con el año de parametro
-- no se puede poner un subselect adentro de SUM(), por eso la suma de los items se calcula en un subselect del FROM
CREATE PROCEDURE dbo.mi_facturas_incorrectas ( @anio INT )
AS
BEGIN
    SELECT
        YEAR(x.fact_fecha)                  AS anio,
        COUNT(DISTINCT x.fact_cliente)      AS clientes,
        COUNT(*)                            AS facturas,
        SUM( CASE WHEN ABS( x.fact_total - x.fact_total_impuestos - ISNULL(x.suma_items, 0) ) > 1 
                  THEN 1 ELSE 0 END )       AS facturas_incorrectas
    FROM ( SELECT 
               f.fact_fecha, f.fact_cliente, f.fact_total, f.fact_total_impuestos,
               ( SELECT SUM(i.item_cantidad * i.item_precio)
                 FROM Item_Factura i
                 WHERE i.item_tipo = f.fact_tipo AND i.item_sucursal = f.fact_sucursal AND i.item_numero = f.fact_numero ) AS suma_items
           FROM Factura f
           WHERE YEAR(f.fact_fecha) = @anio ) x
    GROUP BY YEAR(x.fact_fecha)
END
GO

EXECUTE dbo.mi_facturas_incorrectas @anio = 2012
```

```sql
-- CLASE 7 - EJERCICIO 1: cursor de solo lectura
DECLARE @clie CHAR(6)
DECLARE @razon CHAR(100)
DECLARE @cant INT
DECLARE @procesados INT = 0

DECLARE mi_cursor CURSOR FOR
    SELECT clie_codigo, clie_razon_social
    FROM Cliente
    WHERE clie_limite_credito >= 1000

OPEN mi_cursor
FETCH NEXT FROM mi_cursor INTO @clie, @razon

WHILE @@FETCH_STATUS = 0
BEGIN
    SET @cant = ( SELECT COUNT(*) FROM Factura WHERE fact_cliente = @clie )
    PRINT RTRIM(@razon) + ' - facturas: ' + CAST(@cant AS VARCHAR(10))
    SET @procesados = @procesados + 1
    FETCH NEXT FROM mi_cursor INTO @clie, @razon
END

CLOSE mi_cursor
DEALLOCATE mi_cursor

PRINT 'Clientes procesados: ' + CAST(@procesados AS VARCHAR(10))
```

```sql
-- CLASE 7 - EJERCICIO 2: corte de control (un renglon por vendedor y año, subtotal por año)
-- Se recorren las FACTURAS de a una; el corte se detecta cuando cambia el año o el vendedor respecto de la fila anterior
DECLARE @anio INT
DECLARE @vend NUMERIC(6)
DECLARE @total DECIMAL(12,2)
DECLARE @anio_ant INT = NULL
DECLARE @vend_ant NUMERIC(6) = NULL
DECLARE @acum_vend DECIMAL(14,2) = 0
DECLARE @acum_anio DECIMAL(14,2) = 0

DECLARE mi_cursor CURSOR FOR
    SELECT YEAR(fact_fecha), fact_vendedor, fact_total
    FROM Factura
    ORDER BY YEAR(fact_fecha), fact_vendedor

OPEN mi_cursor
FETCH NEXT FROM mi_cursor INTO @anio, @vend, @total

WHILE @@FETCH_STATUS = 0
BEGIN
    IF @anio_ant IS NOT NULL AND ( @anio <> @anio_ant OR @vend <> @vend_ant )       -- corte por vendedor
    BEGIN
        PRINT 'Año ' + CAST(@anio_ant AS VARCHAR(4)) + ' - vendedor ' + CAST(@vend_ant AS VARCHAR(10)) + ': ' + CAST(@acum_vend AS VARCHAR(20))
        SET @acum_vend = 0
    END

    IF @anio_ant IS NOT NULL AND @anio <> @anio_ant                                 -- corte por año
    BEGIN
        PRINT '== Subtotal ' + CAST(@anio_ant AS VARCHAR(4)) + ': ' + CAST(@acum_anio AS VARCHAR(20))
        SET @acum_anio = 0
    END

    SET @acum_vend = @acum_vend + @total
    SET @acum_anio = @acum_anio + @total
    SET @anio_ant = @anio
    SET @vend_ant = @vend

    FETCH NEXT FROM mi_cursor INTO @anio, @vend, @total
END

IF @anio_ant IS NOT NULL                                                            -- el ultimo grupo no dispara corte, se imprime al salir
BEGIN
    PRINT 'Año ' + CAST(@anio_ant AS VARCHAR(4)) + ' - vendedor ' + CAST(@vend_ant AS VARCHAR(10)) + ': ' + CAST(@acum_vend AS VARCHAR(20))
    PRINT '== Subtotal ' + CAST(@anio_ant AS VARCHAR(4)) + ': ' + CAST(@acum_anio AS VARCHAR(20))
END

CLOSE mi_cursor
DEALLOCATE mi_cursor
```

```sql
-- CLASE 7 - EJERCICIO 3 (b): cursor para modificar, +10% de limite a los clientes que facturaron en 2012
DECLARE @clie CHAR(6)

DECLARE mi_cursor CURSOR FOR
    SELECT clie_codigo
    FROM mi_cliente
    WHERE clie_codigo IN ( SELECT fact_cliente FROM Factura WHERE YEAR(fact_fecha) = 2012 )
    FOR UPDATE OF clie_limite_credito

OPEN mi_cursor
FETCH NEXT FROM mi_cursor INTO @clie

WHILE @@FETCH_STATUS = 0
BEGIN
    UPDATE mi_cliente
    SET clie_limite_credito = clie_limite_credito * 1.10
    WHERE CURRENT OF mi_cursor                          -- solo la fila donde esta parado el cursor

    FETCH NEXT FROM mi_cursor INTO @clie
END

CLOSE mi_cursor
DEALLOCATE mi_cursor

-- verifico contra la tabla original: solo cambian los clientes de 2012
SELECT c.clie_codigo, c.clie_limite_credito AS limite_original, m.clie_limite_credito AS limite_nuevo
FROM Cliente c
JOIN mi_cliente m ON m.clie_codigo = c.clie_codigo
WHERE m.clie_limite_credito <> c.clie_limite_credito
```

**Clase 7, ejercicio 5 (aislamiento, dos pestañas).** Cada celda de abajo dice en qué pestaña va y en qué orden se ejecuta. Lo que se espera:
- **(a) `READ COMMITTED`:** la segunda lectura de A trae el valor nuevo que B confirmó entre medio (lectura no repetible).
- **(a) `REPEATABLE READ`:** A bloquea las filas que leyó, así que el `UPDATE` de B se queda **esperando** hasta que A termine, y la segunda lectura de A da igual que la primera.
- **(b) deadlock:** A tiene la fila 1 y pide la 2, B tiene la 2 y pide la 1; ninguna puede avanzar. SQL Server elige una víctima, le hace `ROLLBACK` y le tira el error 1205; la otra termina.

```sql
-- CLASE 7 - EJERCICIO 5 (a) - PESTAÑA A, paso 1 (cambiar a REPEATABLE READ para la segunda corrida)
SET TRANSACTION ISOLATION LEVEL READ COMMITTED
BEGIN TRANSACTION
    SELECT clie_codigo, clie_limite_credito FROM mi_cliente WHERE clie_codigo = '00001'     -- primera lectura
    -- ahora ir a la pestaña B y correr su celda; despues volver y seguir con el paso 2
```

```sql
-- CLASE 7 - EJERCICIO 5 (a) - PESTAÑA B (se corre entre el paso 1 y el 2 de A)
UPDATE mi_cliente
SET clie_limite_credito = clie_limite_credito + 1
WHERE clie_codigo = '00001'
-- con READ COMMITTED termina enseguida; con REPEATABLE READ queda esperando hasta que A haga COMMIT/ROLLBACK
```

```sql
-- CLASE 7 - EJERCICIO 5 (a) - PESTAÑA A, paso 2
    SELECT clie_codigo, clie_limite_credito FROM mi_cliente WHERE clie_codigo = '00001'     -- segunda lectura: ¿igual o distinta?
COMMIT
```

```sql
-- CLASE 7 - EJERCICIO 5 (b) - PESTAÑA A (correr y, dentro de los 10 segundos, correr la B)
BEGIN TRANSACTION
    UPDATE mi_cliente SET clie_limite_credito = clie_limite_credito + 1 WHERE clie_codigo = '00001'    -- fila 1
    WAITFOR DELAY '00:00:10'
    UPDATE mi_cliente SET clie_limite_credito = clie_limite_credito + 1 WHERE clie_codigo = '00002'    -- fila 2
COMMIT
```

### EJERCICIO T6
Realizar un procedimiento que si en alguna factura se facturaron componentes que conforman un combo determinado (o sea que juntos componen otro producto de mayor nivel), en cuyo caso deberá reemplazar las filas correspondientes a dichos productos por una sola fila con el producto que componen con la cantidad de dicho producto que corresponda.

```sql
-- EJERCICIO T6
-- NOTAS: Hay combo si la factura tiene TODOS los componentes del producto; cantidad de combos =
-- MIN(FLOOR(item_cantidad / comp_cantidad)). A los componentes se les resta lo usado y se borran si quedan en 0
-- (el sobrante queda como renglón suelto). Precio del combo = lo que se pagó por sus componentes
-- (SUM(comp_cantidad*item_precio)), así el monto de la factura se mantiene; si el combo ya estaba en la factura se
-- suma la cantidad y se promedia el precio (el redondeo a 2 decimales puede mover centavos). Se procesa de a un
-- combo por vuelta (WHILE) para que dos combos no usen el mismo componente y para armar combos de varios niveles.
-- Probado: 6 facturas tenían 00006411+00006404 (y ya tenían 00006402): quedan con un solo renglón de 00006402 (ej.
-- 00096572: 24+23 = 47 unidades, sobra 1 de 00006411); los montos de 5 facturas no cambian y en una varían 0.14
-- por redondeo.
CREATE PROCEDURE dbo.sp_t6_armar_combos
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @tipo CHAR(1), @suc CHAR(4), @num CHAR(8), @combo CHAR(8),
            @cant DECIMAL(12,2), @precio DECIMAL(12,2);
    BEGIN TRY
        BEGIN TRANSACTION;
        WHILE 1 = 1
        BEGIN
            SET @combo = NULL;
            -- 1) busco una factura que tenga todos los componentes de algún combo
            SELECT TOP 1 @tipo = i.item_tipo, @suc = i.item_sucursal, @num = i.item_numero, @combo = c.comp_producto,
                         @cant = MIN(FLOOR(i.item_cantidad / c.comp_cantidad))     -- combos completos que se pueden armar
            FROM Item_Factura i
            JOIN Composicion c ON c.comp_componente = i.item_producto
            GROUP BY i.item_tipo, i.item_sucursal, i.item_numero, c.comp_producto
            HAVING COUNT(*) = ( SELECT COUNT(*) FROM Composicion c2 WHERE c2.comp_producto = c.comp_producto )  -- están todos
               AND MIN(FLOOR(i.item_cantidad / c.comp_cantidad)) >= 1
            ORDER BY i.item_tipo, i.item_sucursal, i.item_numero, c.comp_producto;

            IF @combo IS NULL BREAK;                                        -- no quedan combos por armar

            -- 2) precio de un combo = lo que se pagó por sus componentes en esa factura
            SELECT @precio = SUM(c.comp_cantidad * i.item_precio)
            FROM Composicion c
            JOIN Item_Factura i ON i.item_producto = c.comp_componente
                               AND i.item_tipo = @tipo AND i.item_sucursal = @suc AND i.item_numero = @num
            WHERE c.comp_producto = @combo;

            -- 3) descuento los componentes usados y borro los que quedan en 0
            UPDATE i
            SET i.item_cantidad = i.item_cantidad - @cant * c.comp_cantidad
            FROM Item_Factura i
            JOIN Composicion c ON c.comp_componente = i.item_producto AND c.comp_producto = @combo
            WHERE i.item_tipo = @tipo AND i.item_sucursal = @suc AND i.item_numero = @num;

            DELETE FROM Item_Factura
            WHERE item_tipo = @tipo AND item_sucursal = @suc AND item_numero = @num AND item_cantidad <= 0;

            -- 4) agrego el combo (si ya estaba en la factura, sumo cantidad y promedio el precio)
            UPDATE Item_Factura
            SET item_precio = (item_cantidad * item_precio + @cant * @precio) / (item_cantidad + @cant),
                item_cantidad = item_cantidad + @cant
            WHERE item_tipo = @tipo AND item_sucursal = @suc AND item_numero = @num AND item_producto = @combo;

            IF @@ROWCOUNT = 0
                INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
                VALUES (@tipo, @suc, @num, @combo, @cant, @precio);
        END
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
        PRINT ERROR_MESSAGE();
    END CATCH
END
GO

-- Prueba (modifica facturas: dentro de una transacción que se deshace)
BEGIN TRAN;
    SELECT item_numero, item_producto, item_cantidad, item_precio, item_cantidad * item_precio AS monto
    FROM Item_Factura WHERE item_tipo = 'A' AND item_sucursal = '0003' AND item_numero = '00096572'
      AND item_producto IN ('00006402', '00006404', '00006411');                -- antes: 6402 + los componentes 6404 y 6411
    EXEC dbo.sp_t6_armar_combos;
    SELECT item_numero, item_producto, item_cantidad, item_precio, item_cantidad * item_precio AS monto
    FROM Item_Factura WHERE item_tipo = 'A' AND item_sucursal = '0003' AND item_numero = '00096572'
      AND item_producto IN ('00006402', '00006404', '00006411');                -- después: 6402 con 47 u. y sobra 1 de 6411
ROLLBACK;
GO

-- DROP PROCEDURE dbo.sp_t6_armar_combos;
```

```sql
-- EJERCICIO T7
-- NOTAS: Como la tabla no existe en la base, el script la crea. Cant. Mov. = cantidad de renglones de Item_Factura
-- del artículo en el período; Precio de Venta = AVG(item_precio); Renglón = ROW_NUMBER por código; Ganancia = lo
-- vendido (SUM(cantidad*precio)) - cantidad vendida * costo actual, tomando prod_precio como costo actual. Fechas
-- inclusive. El SP vacía la tabla antes de cargar. Probado con 2012-01-01 a 2012-12-31: inserta 629 filas (=
-- artículos distintos vendidos en 2012); verificado a mano 00000102: 27 mov., 271.63 - 35 * 6.26 = 52.53.
IF OBJECT_ID('dbo.Ventas') IS NULL
    CREATE TABLE dbo.Ventas
    ( vent_codigo    CHAR(8) NOT NULL PRIMARY KEY,
      vent_detalle   CHAR(50),
      vent_cant_mov  INT,
      vent_precio    DECIMAL(12,2),
      vent_renglon   INT,
      vent_ganancia  DECIMAL(12,2) );
GO

CREATE PROCEDURE dbo.sp_t7_cargar_ventas (@desde SMALLDATETIME, @hasta SMALLDATETIME)
AS
BEGIN
    SET NOCOUNT ON;
    IF @desde IS NULL OR @hasta IS NULL OR @desde > @hasta                 -- defensivo
    BEGIN
        RAISERROR('Rango de fechas invalido', 16, 1);
        RETURN;
    END
    BEGIN TRY
        BEGIN TRANSACTION;
        DELETE FROM dbo.Ventas;
        INSERT INTO dbo.Ventas (vent_codigo, vent_detalle, vent_cant_mov, vent_precio, vent_renglon, vent_ganancia)
        SELECT
            p.prod_codigo, p.prod_detalle,
            COUNT(*),                                                       -- movimientos = renglones de venta
            AVG(i.item_precio),                                             -- precio promedio
            ROW_NUMBER() OVER (ORDER BY p.prod_codigo),                     -- nro. de renglón
            SUM(i.item_cantidad * i.item_precio) - SUM(i.item_cantidad) * p.prod_precio   -- vendido - costo actual
        FROM Item_Factura i
        JOIN Factura f  ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
        JOIN Producto p ON p.prod_codigo = i.item_producto
        WHERE f.fact_fecha BETWEEN @desde AND @hasta
        GROUP BY p.prod_codigo, p.prod_detalle, p.prod_precio;
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
        PRINT ERROR_MESSAGE();
    END CATCH
END
GO

-- Prueba
EXEC dbo.sp_t7_cargar_ventas '20120101', '20121231';
SELECT TOP 10 * FROM dbo.Ventas ORDER BY vent_renglon;
SELECT COUNT(*) AS filas FROM dbo.Ventas;
SELECT COUNT(DISTINCT i.item_producto) AS control                           -- tiene que dar lo mismo
FROM Item_Factura i
JOIN Factura f ON f.fact_tipo = i.item_tipo AND f.fact_sucursal = i.item_sucursal AND f.fact_numero = i.item_numero
WHERE f.fact_fecha BETWEEN '20120101' AND '20121231';
GO

-- DROP PROCEDURE dbo.sp_t7_cargar_ventas;
-- DROP TABLE dbo.Ventas;
```

### EJERCICIO T8
Realizar un procedimiento que complete la tabla `Diferencias` de precios, para los productos facturados que tengan composición y en los cuales el precio de facturación sea diferente al precio del cálculo de los precios unitarios por cantidad de sus componentes. Se aclara que un producto que compone a otro, también puede estar compuesto por otros y así sucesivamente. La tabla se debe crear y está formada por las siguientes columnas:

| Código | Detalle | Cantidad | Precio_generado | Precio_facturado |
|---|---|---|---|---|
| Código del artículo | Detalle del artículo | Cantidad de productos que conforman el combo | Precio que se compone a través de sus componentes | Precio del producto |

```sql
-- EJERCICIO T8
-- NOTAS: Precio_generado con función recursiva (un compuesto vale la suma de comp_cantidad * precio de cada
-- componente, y un componente compuesto se calcula igual, a cualquier nivel; los simples valen prod_precio).
-- Precio_facturado = cada item_precio distinto con que se facturó el producto (una fila por producto y precio
-- distinto que difiera del generado). Cantidad = cantidad de componentes directos del combo (filas de
-- Composicion). El SP vacía la tabla antes de cargar. Probado: 00006402 da precio generado 86.02 (=
-- 1*(2*1.34+2*2.32) + 1*(10*3.94+10*3.93)), verificado a mano; se insertan 19 filas (los 5 productos compuestos
-- facturados, todos con precios de factura distintos al generado).
CREATE TABLE dbo.Diferencias
( dife_codigo           CHAR(8)       NOT NULL,
  dife_detalle          CHAR(50),
  dife_cantidad         INT,
  dife_precio_generado  DECIMAL(12,2),
  dife_precio_facturado DECIMAL(12,2) NOT NULL,
  CONSTRAINT PK_Diferencias PRIMARY KEY (dife_codigo, dife_precio_facturado) );
GO

-- Precio armado por componentes, a cualquier nivel de composición
CREATE FUNCTION dbo.fn_t8_precio_generado (@prod CHAR(8))
RETURNS DECIMAL(12,2)
AS
BEGIN
    IF NOT EXISTS ( SELECT 1 FROM Composicion WHERE comp_producto = @prod )       -- caso base: es simple
        RETURN ( SELECT prod_precio FROM Producto WHERE prod_codigo = @prod );
    RETURN ( SELECT SUM(c.comp_cantidad * dbo.fn_t8_precio_generado(c.comp_componente))   -- caso recursivo
             FROM Composicion c
             WHERE c.comp_producto = @prod );
END
GO

CREATE PROCEDURE dbo.sp_t8_cargar_diferencias
AS
BEGIN
    SET NOCOUNT ON;
    BEGIN TRY
        BEGIN TRANSACTION;
        DELETE FROM dbo.Diferencias;
        INSERT INTO dbo.Diferencias (dife_codigo, dife_detalle, dife_cantidad, dife_precio_generado, dife_precio_facturado)
        SELECT DISTINCT
            p.prod_codigo, p.prod_detalle,
            ( SELECT COUNT(*) FROM Composicion c WHERE c.comp_producto = p.prod_codigo ),   -- componentes del combo
            dbo.fn_t8_precio_generado(p.prod_codigo),
            i.item_precio
        FROM Producto p
        JOIN Item_Factura i ON i.item_producto = p.prod_codigo                           -- solo los facturados
        WHERE p.prod_codigo IN ( SELECT comp_producto FROM Composicion )                 -- solo los que tienen composición
          AND i.item_precio <> dbo.fn_t8_precio_generado(p.prod_codigo);
        COMMIT TRANSACTION;
    END TRY
    BEGIN CATCH
        IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
        PRINT ERROR_MESSAGE();
    END CATCH
END
GO

-- Prueba
EXEC dbo.sp_t8_cargar_diferencias;
SELECT * FROM dbo.Diferencias ORDER BY dife_codigo, dife_precio_facturado;
GO

-- DROP PROCEDURE dbo.sp_t8_cargar_diferencias;
-- DROP FUNCTION dbo.fn_t8_precio_generado;
-- DROP TABLE dbo.Diferencias;
```

### EJERCICIO T9
Crear el/los objetos de base de datos que ante alguna modificación de un ítem de factura de un artículo con composición realice el movimiento de sus correspondientes componentes.

```sql
-- EJERCICIO T9
-- NOTAS: "Movimiento" = movimiento de STOCK de los componentes directos. Trigger AFTER INSERT, UPDATE, DELETE
-- sobre Item_Factura (molde "propagar"): lo de inserted resta stock (cantidad * comp_cantidad) y lo de deleted lo
-- devuelve; en un UPDATE se neutralizan y queda solo la diferencia. Se mueve el depósito que más stock tiene de
-- cada componente (desempate: menor código de depósito); si el componente no tiene fila en STOCK no se mueve nada.
-- Funciona con varias filas a la vez (acumula en una tabla variable). Probado: insertar 3 combos 00006411
-- (2x00005703 + 2x00001516) baja 6 unidades del depósito con más stock de cada componente (00001516 dep. 03: 154
-- -> 148; 00005703 dep. 11: 130 -> 124) y un UPDATE de 3 a 1 devuelve 4 (152 y 128).
CREATE TRIGGER dbo.tr_t9_mover_componentes ON dbo.Item_Factura
AFTER INSERT, UPDATE, DELETE
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @mov TABLE (componente CHAR(8), cantidad DECIMAL(12,2));

    -- lo vendido sale del stock (negativo) y lo que se deshace vuelve (positivo)
    INSERT INTO @mov
    SELECT c.comp_componente, -ins.item_cantidad * c.comp_cantidad
    FROM inserted ins
    JOIN Composicion c ON c.comp_producto = ins.item_producto           -- solo artículos con composición
    UNION ALL
    SELECT c.comp_componente, del.item_cantidad * c.comp_cantidad
    FROM deleted del
    JOIN Composicion c ON c.comp_producto = del.item_producto;

    IF NOT EXISTS ( SELECT 1 FROM @mov ) RETURN;

    UPDATE s
    SET s.stoc_cantidad = s.stoc_cantidad + ( SELECT SUM(m.cantidad) FROM @mov m WHERE m.componente = s.stoc_producto )
    FROM STOCK s
    WHERE s.stoc_producto IN ( SELECT componente FROM @mov )
      AND s.stoc_deposito = ( SELECT TOP 1 s2.stoc_deposito                 -- el depósito con más stock del componente
                              FROM STOCK s2
                              WHERE s2.stoc_producto = s.stoc_producto
                              ORDER BY s2.stoc_cantidad DESC, s2.stoc_deposito );
END
GO

-- Prueba (dentro de una transacción que se deshace)
BEGIN TRAN;
    SELECT stoc_producto, stoc_deposito, stoc_cantidad FROM STOCK
    WHERE stoc_producto IN ('00005703', '00001516') ORDER BY stoc_producto, stoc_cantidad DESC;
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '00068710', '00006411', 3, 6.24);                     -- 3 combos: -6 de cada componente
    SELECT stoc_producto, stoc_deposito, stoc_cantidad FROM STOCK
    WHERE stoc_producto IN ('00005703', '00001516') ORDER BY stoc_producto, stoc_cantidad DESC;
    UPDATE Item_Factura SET item_cantidad = 1                                 -- de 3 a 1: devuelve 4 de cada uno
    WHERE item_tipo = 'A' AND item_sucursal = '0003' AND item_numero = '00068710' AND item_producto = '00006411';
    SELECT stoc_producto, stoc_deposito, stoc_cantidad FROM STOCK
    WHERE stoc_producto IN ('00005703', '00001516') ORDER BY stoc_producto, stoc_cantidad DESC;
ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t9_mover_componentes;
```

### EJERCICIO T10
Crear el/los objetos de base de datos que ante el intento de borrar un artículo verifique que no exista stock y si es así lo borre en caso contrario que emita un mensaje de error.

```sql
-- EJERCICIO T10
-- NOTAS: Trigger INSTEAD OF DELETE sobre Producto (con AFTER no sirve: la FK de STOCK rechazaría el borrado
-- antes). "Tiene stock" = alguna fila de STOCK con stoc_cantidad > 0. Con borrado masivo: se borran los que no
-- tienen stock (primero sus filas de STOCK en 0, por la FK) y para los que sí tienen se emite el error sin
-- borrarlos. Si el artículo tiene ventas o es parte de una composición, la FK de Item_Factura/Composicion sigue
-- rechazando el borrado (no se pide borrar en cascada). Probado: con dos artículos de prueba, uno sin stock y otro
-- con stock, un único DELETE borra el primero y da error para el segundo.
CREATE TRIGGER dbo.tr_t10_borrar_producto ON dbo.Producto
INSTEAD OF DELETE
AS
BEGIN
    SET NOCOUNT ON;

    -- 1) los que no tienen stock: borro sus filas de STOCK (todas en 0) y después el producto
    DELETE s
    FROM STOCK s
    JOIN deleted d ON d.prod_codigo = s.stoc_producto
    WHERE NOT EXISTS ( SELECT 1 FROM STOCK s2 WHERE s2.stoc_producto = d.prod_codigo AND s2.stoc_cantidad > 0 );

    DELETE p
    FROM Producto p
    JOIN deleted d ON d.prod_codigo = p.prod_codigo
    WHERE NOT EXISTS ( SELECT 1 FROM STOCK s WHERE s.stoc_producto = d.prod_codigo AND s.stoc_cantidad > 0 );

    -- 2) los que tienen stock no se borran: mensaje de error
    IF EXISTS ( SELECT 1 FROM deleted d
                WHERE EXISTS ( SELECT 1 FROM STOCK s WHERE s.stoc_producto = d.prod_codigo AND s.stoc_cantidad > 0 ) )
        RAISERROR('No se puede borrar el articulo: todavia tiene stock', 16, 1);
END
GO

-- Prueba (dentro de una transacción que se deshace)
BEGIN TRAN;
    INSERT INTO Producto (prod_codigo, prod_detalle, prod_precio) VALUES ('ZZ000001', 'PRUEBA SIN STOCK', 1), ('ZZ000002', 'PRUEBA CON STOCK', 1);
    INSERT INTO STOCK (stoc_producto, stoc_deposito, stoc_cantidad) VALUES ('ZZ000001', '00', 0), ('ZZ000002', '00', 5);
    DELETE FROM Producto WHERE prod_codigo IN ('ZZ000001', 'ZZ000002');       -- borra el 1 y da error por el 2
    SELECT prod_codigo, prod_detalle FROM Producto WHERE prod_codigo LIKE 'ZZ%';   -- queda solo ZZ000002
ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t10_borrar_producto;
```

```sql
-- EJERCICIO T12
-- NOTAS: Trigger AFTER INSERT, UPDATE sobre Composicion (n aplicaciones => la regla vive en la base) + función que
-- recorre la composición de forma iterativa, nivel por nivel, guardando los productos ya alcanzados (así termina
-- aunque haya un ciclo, cosa que una función recursiva no garantiza: se pasaría de 32 niveles). Como hoy la regla
-- se cumple, todo ciclo nuevo pasa por una fila de inserted: alcanza con preguntar si cada comp_producto insertado
-- aparece entre sus propios componentes a cualquier nivel (incluye el caso directo A compone A). DELETE no puede
-- generar ciclos. Probado: rechaza 00006408 compuesto por sí mismo, rechaza 00006408 compuesto por 00006402 (ciclo
-- de 3 niveles 6402->6404->6408->6402) y un UPDATE que arma un ciclo; acepta una composición válida; Composicion
-- queda con sus 12 filas.
CREATE FUNCTION dbo.fn_t12_se_compone (@prod CHAR(8))
RETURNS BIT
AS
BEGIN
    DECLARE @alcanzados TABLE (prod CHAR(8) PRIMARY KEY);

    INSERT INTO @alcanzados                                    -- nivel 1: componentes directos
    SELECT DISTINCT comp_componente FROM Composicion WHERE comp_producto = @prod;

    WHILE @@ROWCOUNT > 0                                       -- mientras el nivel anterior agregue algo
        INSERT INTO @alcanzados
        SELECT DISTINCT c.comp_componente
        FROM Composicion c
        JOIN @alcanzados a ON a.prod = c.comp_producto
        WHERE NOT EXISTS ( SELECT 1 FROM @alcanzados a2 WHERE a2.prod = c.comp_componente );   -- solo los nuevos

    IF EXISTS ( SELECT 1 FROM @alcanzados WHERE prod = @prod ) -- ¿me encontré entre mis componentes?
        RETURN 1;
    RETURN 0;
END
GO

CREATE TRIGGER dbo.tr_t12_composicion_ciclica ON dbo.Composicion
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1 FROM inserted ins
                WHERE dbo.fn_t12_se_compone(ins.comp_producto) = 1 )    -- la tabla ya tiene las filas nuevas
    BEGIN
        RAISERROR('Un producto no puede estar compuesto por si mismo (a ningun nivel)', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba (los rechazos van en TRY/CATCH para que el script siga corriendo)
BEGIN TRY
    INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00006408', '00006408');   -- directo: rechaza
END TRY
BEGIN CATCH
    IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
    PRINT ERROR_MESSAGE();
END CATCH
GO
BEGIN TRY
    INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00006408', '00006402');   -- 3 niveles: rechaza
END TRY
BEGIN CATCH
    IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
    PRINT ERROR_MESSAGE();
END CATCH
GO
BEGIN TRY
    UPDATE Composicion SET comp_componente = '00006402'                                                      -- UPDATE con ciclo: rechaza
    WHERE comp_producto = '00006404' AND comp_componente = '00006408';
END TRY
BEGIN CATCH
    IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
    PRINT ERROR_MESSAGE();
END CATCH
GO
BEGIN TRAN;
    INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00006408', '00001516'); -- válido: entra
    SELECT * FROM Composicion WHERE comp_producto = '00006408';
ROLLBACK;
GO
SELECT COUNT(*) AS filas_composicion FROM Composicion;                                                        -- siguen 12
GO

-- DROP TRIGGER dbo.tr_t12_composicion_ciclica;
-- DROP FUNCTION dbo.fn_t12_se_compone;
```

```sql
-- EJERCICIO T13
-- NOTAS: "n aplicaciones" => trigger (no se puede confiar en el código de cada app). Función recursiva para la
-- suma de salarios a cargo + trigger AFTER INSERT/UPDATE/DELETE sobre Empleado que valida solo los jefes afectados
-- (el propio empleado tocado y toda su cadena de jefes hacia arriba) y rechaza con RAISERROR + ROLLBACK. "Jefe" =
-- empleado con al menos un subordinado. OJO: en esta base la regla NO se cumple hoy (gerente 25000 vs 20% de
-- 72200), así que la prueba arma una jerarquía nueva aislada (100 -> 101, 102): el alta válida pasa, subir el
-- sueldo del jefe a 2000 y borrar un subordinado son rechazados. Probado en TEST_G5.
GO
-- Suma de salarios de TODOS los empleados a cargo (directos + indirectos)
CREATE FUNCTION dbo.fn_t13_salarios_a_cargo (@jefe NUMERIC(6))
RETURNS DECIMAL(14,2)
AS
BEGIN
    RETURN ( SELECT ISNULL(SUM(e.empl_salario + dbo.fn_t13_salarios_a_cargo(e.empl_codigo)), 0)  -- directo + lo que tiene él a cargo
             FROM Empleado e
             WHERE e.empl_jefe = @jefe );
END
GO

CREATE TRIGGER dbo.tr_t13_salario_jefe ON dbo.Empleado
AFTER INSERT, UPDATE, DELETE
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @afectados TABLE (empl NUMERIC(6) PRIMARY KEY);

    -- 1) empleados tocados (por si son jefes y les cambió el sueldo) y sus jefes viejos/nuevos
    INSERT INTO @afectados
    SELECT empl_codigo FROM inserted
    UNION SELECT empl_jefe FROM inserted WHERE empl_jefe IS NOT NULL
    UNION SELECT empl_jefe FROM deleted  WHERE empl_jefe IS NOT NULL;

    -- 2) subo por la cadena de jefes hasta el gerente (la suma de todos ellos cambió)
    WHILE @@ROWCOUNT > 0
        INSERT INTO @afectados
        SELECT DISTINCT e.empl_jefe
        FROM Empleado e
        JOIN @afectados a ON a.empl = e.empl_codigo
        WHERE e.empl_jefe IS NOT NULL
          AND e.empl_jefe NOT IN ( SELECT empl FROM @afectados );

    -- 3) valido solo los afectados que hoy son jefes
    IF EXISTS ( SELECT 1
                FROM Empleado j
                JOIN @afectados a ON a.empl = j.empl_codigo
                WHERE EXISTS ( SELECT 1 FROM Empleado e WHERE e.empl_jefe = j.empl_codigo )   -- es jefe
                  AND j.empl_salario > 0.20 * dbo.fn_t13_salarios_a_cargo(j.empl_codigo) )
    BEGIN
        RAISERROR('Un jefe no puede ganar mas del 20%% de la suma de los salarios de sus empleados', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba: jerarquía nueva aislada (en la base actual la regla no se cumple)
BEGIN TRANSACTION;
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe)
    VALUES (100, 'Jefe', 'Prueba', 1000, NULL);                       -- todavía no es jefe: pasa
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe)
    VALUES (101, 'Emp', 'Uno', 3000, 100), (102, 'Emp', 'Dos', 3000, 100);  -- 20% de 6000 = 1200 >= 1000: pasa
    SELECT empl_codigo, empl_salario, dbo.fn_t13_salarios_a_cargo(empl_codigo) AS salarios_a_cargo
    FROM Empleado WHERE empl_codigo IN (1, 3, 100);                   -- 1 -> 72200, 3 -> 43700, 100 -> 6000
ROLLBACK;
GO
BEGIN TRANSACTION;
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe) VALUES (100, 'Jefe', 'Prueba', 1000, NULL);
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe) VALUES (101, 'Emp', 'Uno', 3000, 100), (102, 'Emp', 'Dos', 3000, 100);
    UPDATE Empleado SET empl_salario = 2000 WHERE empl_codigo = 100;  -- 2000 > 1200: RECHAZADO (rollback)
IF @@TRANCOUNT > 0 ROLLBACK;
GO
BEGIN TRANSACTION;
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe) VALUES (100, 'Jefe', 'Prueba', 1000, NULL);
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_salario, empl_jefe) VALUES (101, 'Emp', 'Uno', 3000, 100), (102, 'Emp', 'Dos', 3000, 100);
    DELETE FROM Empleado WHERE empl_codigo = 102;                     -- queda 20% de 3000 = 600 < 1000: RECHAZADO
IF @@TRANCOUNT > 0 ROLLBACK;
GO
SELECT COUNT(*) AS quedaron FROM Empleado WHERE empl_codigo >= 100;  -- 0
GO

-- DROP TRIGGER dbo.tr_t13_salario_jefe;
-- DROP FUNCTION dbo.fn_t13_salarios_a_cargo;
```

```sql
-- EJERCICIO T14
-- NOTAS: Trigger AFTER INSERT, UPDATE sobre Item_Factura. "Suma de los precios de sus componentes" =
-- SUM(comp_cantidad * prod_precio) de los componentes directos (precio unitario del compuesto vs. lo que costarían
-- sus partes sueltas); se compara contra item_precio (precio unitario). Primero valida (si algún renglón está por
-- debajo de la mitad rechaza TODA la sentencia con RAISERROR + ROLLBACK) y después imprime con un cursor los
-- renglones vendidos por debajo de la suma (PRINT no se puede hacer "en conjunto"). Probado en TEST_G5 con
-- 00001104 (componentes suman 6.91): a 5.00 imprime el aviso, a 3.00 (< 3.455) lo rechaza.
GO
-- Suma de los precios de los componentes (por cantidad) de un producto compuesto
CREATE FUNCTION dbo.fn_t14_suma_componentes (@prod CHAR(8))
RETURNS DECIMAL(12,2)
AS
BEGIN
    RETURN ( SELECT SUM(c.comp_cantidad * p.prod_precio)
             FROM Composicion c
             JOIN Producto p ON p.prod_codigo = c.comp_componente
             WHERE c.comp_producto = @prod );          -- NULL si no es compuesto
END
GO

CREATE TRIGGER dbo.tr_t14_precio_compuesto ON dbo.Item_Factura
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;

    -- 1) Validar: ninguno por debajo de la mitad de la suma de componentes
    IF EXISTS ( SELECT 1
                FROM inserted ins
                WHERE ins.item_precio < 0.5 * dbo.fn_t14_suma_componentes(ins.item_producto) )  -- simples: NULL => no entra
    BEGIN
        RAISERROR('No se puede vender un producto compuesto a menos de la mitad de la suma de sus componentes', 16, 1);
        ROLLBACK TRANSACTION;
        RETURN;
    END

    -- 2) Avisar: los vendidos por debajo de la suma (puede haber varios renglones => cursor)
    DECLARE @fecha SMALLDATETIME, @cliente CHAR(6), @prod CHAR(8), @precio DECIMAL(12,2), @suma DECIMAL(12,2);
    DECLARE c_aviso CURSOR FOR
        SELECT f.fact_fecha, f.fact_cliente, ins.item_producto, ins.item_precio, dbo.fn_t14_suma_componentes(ins.item_producto)
        FROM inserted ins
        JOIN Factura f ON f.fact_tipo = ins.item_tipo AND f.fact_sucursal = ins.item_sucursal AND f.fact_numero = ins.item_numero
        WHERE ins.item_precio < dbo.fn_t14_suma_componentes(ins.item_producto);
    OPEN c_aviso;
    FETCH NEXT FROM c_aviso INTO @fecha, @cliente, @prod, @precio, @suma;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        PRINT 'Fecha: ' + CONVERT(VARCHAR(10), @fecha, 103) + ' - Cliente: ' + RTRIM(@cliente)
            + ' - Producto: ' + RTRIM(@prod) + ' - Precio: ' + CAST(@precio AS VARCHAR(15))
            + ' (componentes: ' + CAST(@suma AS VARCHAR(15)) + ')';
        FETCH NEXT FROM c_aviso INTO @fecha, @cliente, @prod, @precio, @suma;
    END
    CLOSE c_aviso;
    DEALLOCATE c_aviso;
END
GO

-- Prueba 1: compuesto por debajo de la suma pero arriba de la mitad => imprime y graba
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 5.00),            -- 5.00 < 6.91 => aviso
           ('A', '0003', '99999999', '00001109', 1, 1.00);            -- simple => nada
    SELECT COUNT(*) AS renglones_grabados FROM Item_Factura WHERE item_numero = '99999999';   -- 2
ROLLBACK;
GO
-- Prueba 2: por debajo de la mitad => RECHAZADO
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 3.00);            -- 3.00 < 3.455
IF @@TRANCOUNT > 0 ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t14_precio_compuesto;
-- DROP FUNCTION dbo.fn_t14_suma_componentes;
```

```sql
-- EJERCICIO T15
-- NOTAS: Es la misma función que SolucionesParcial.sql (4.a, dbo.fn_precio_real): función escalar recursiva (sirve
-- en un WHERE, un procedimiento no). Caso base: producto simple => prod_precio; caso recursivo: SUM(comp_cantidad
-- * precio real del componente). Probado en TEST_G5: 00006402 (compuesto de compuestos, 2 niveles: 1 x 00006411 +
-- 1 x 00006404) da 7.32 + 78.70 = 86.02 contra un precio de lista de 80.00; el SELECT con la función en el WHERE
-- devuelve 5 de los 6 compuestos (todos menos 00001718 se venden más baratos que sus partes).
GO
CREATE FUNCTION dbo.fn_precio_real (@prod CHAR(8))
RETURNS DECIMAL(12,2)
AS
BEGIN
    IF NOT EXISTS ( SELECT 1 FROM Composicion WHERE comp_producto = @prod )      -- caso base: es simple
        RETURN ( SELECT prod_precio FROM Producto WHERE prod_codigo = @prod );
    RETURN ( SELECT SUM(c.comp_cantidad * dbo.fn_precio_real(c.comp_componente))  -- caso recursivo
             FROM Composicion c
             WHERE c.comp_producto = @prod );
END
GO

-- Prueba 1: precio de cada compuesto
SELECT prod_codigo, prod_precio, dbo.fn_precio_real(prod_codigo) AS precio_por_componentes
FROM Producto
WHERE prod_codigo IN ( SELECT comp_producto FROM Composicion );
GO
-- Prueba 2: usada como FILTRO en el WHERE (compuestos que se venden más baratos que sus partes)
SELECT p.prod_codigo, p.prod_detalle, p.prod_precio
FROM Producto p
WHERE p.prod_codigo IN ( SELECT comp_producto FROM Composicion )
  AND p.prod_precio < dbo.fn_precio_real(p.prod_codigo);
GO

-- DROP FUNCTION dbo.fn_precio_real;
```

En ultima instancia se dejara stock negativo en el ultimo deposito que se desconto.

```sql
-- EJERCICIO T16
-- NOTAS: Trigger AFTER INSERT sobre Item_Factura ("ante una venta"). Se agrupa inserted por producto (sirve para
-- inserts de muchas filas) y, por cada producto, un cursor recorre sus depósitos de mayor a menor stock (desempate
-- por código de depósito) descontando lo que haya; los compuestos se descuentan como un producto más (no se
-- explotan en componentes). Lo que falta al final queda negativo en el último depósito del que se descontó algo
-- (si ninguno tenía stock positivo, en el de mayor stock). Si el producto no tiene ninguna fila en STOCK no se
-- descuenta nada (supuesto). No se contemplan bajas/modificaciones de renglones (el enunciado habla solo de
-- ventas). Probado en TEST_G5 con 00006404 (stock 3,2,2,1,1,1,0,0 = 10): vender 4 deja 03=0 y 04=1; vender 15 (en
-- dos renglones de una misma sentencia) deja todos en 0 y el 07 en -5.
GO
CREATE TRIGGER dbo.tr_t16_descontar_stock ON dbo.Item_Factura
AFTER INSERT
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @prod CHAR(8), @cant DECIMAL(12,2), @depo CHAR(2), @stock DECIMAL(12,2),
            @descuento DECIMAL(12,2), @ultimo_depo CHAR(2), @primer_depo CHAR(2);

    DECLARE c_prod CURSOR FOR
        SELECT item_producto, SUM(item_cantidad)        -- un producto puede venir en varios renglones
        FROM inserted
        GROUP BY item_producto;
    OPEN c_prod;
    FETCH NEXT FROM c_prod INTO @prod, @cant;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        SET @ultimo_depo = NULL;
        SET @primer_depo = NULL;

        DECLARE c_depo CURSOR FOR
            SELECT stoc_deposito, stoc_cantidad
            FROM STOCK
            WHERE stoc_producto = @prod
            ORDER BY stoc_cantidad DESC, stoc_deposito;  -- el que más tiene primero
        OPEN c_depo;
        FETCH NEXT FROM c_depo INTO @depo, @stock;
        WHILE @@FETCH_STATUS = 0 AND @cant > 0
        BEGIN
            IF @primer_depo IS NULL SET @primer_depo = @depo;
            IF @stock > 0
            BEGIN
                SET @descuento = CASE WHEN @stock >= @cant THEN @cant ELSE @stock END;  -- lo que alcance
                UPDATE STOCK SET stoc_cantidad = stoc_cantidad - @descuento
                WHERE stoc_producto = @prod AND stoc_deposito = @depo;
                SET @cant = @cant - @descuento;
                SET @ultimo_depo = @depo;
            END
            FETCH NEXT FROM c_depo INTO @depo, @stock;
        END
        CLOSE c_depo;
        DEALLOCATE c_depo;

        IF @cant > 0                                     -- no alcanzó: negativo en el último que se descontó
            UPDATE STOCK SET stoc_cantidad = stoc_cantidad - @cant
            WHERE stoc_producto = @prod AND stoc_deposito = ISNULL(@ultimo_depo, @primer_depo);

        FETCH NEXT FROM c_prod INTO @prod, @cant;
    END
    CLOSE c_prod;
    DEALLOCATE c_prod;
END
GO

-- Prueba 1: vender 4 de 00006404 (03 tiene 3, 04 tiene 2) => 03 = 0, 04 = 1
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00006404', 4, 75.20);
    SELECT stoc_deposito, stoc_cantidad FROM STOCK WHERE stoc_producto = '00006404' ORDER BY stoc_deposito;
ROLLBACK;
GO
-- Prueba 2: vender 15 en dos renglones de una sola sentencia (hay 10) => todos en 0 y el 07 en -5
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999998', '2012-07-20', 3, 0, 0, '01502'), ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999998', '00006404', 7, 75.20), ('A', '0003', '99999999', '00006404', 8, 75.20);
    SELECT stoc_deposito, stoc_cantidad FROM STOCK WHERE stoc_producto = '00006404' ORDER BY stoc_deposito;
ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t16_descontar_stock;
```

```sql
-- EJERCICIO T17
-- NOTAS: "No se conoce la forma de acceso" => trigger AFTER INSERT, UPDATE sobre STOCK que rechaza (RAISERROR +
-- ROLLBACK) si alguna fila insertada/modificada queda fuera de [stoc_punto_reposicion, stoc_stock_maximo]; cubre
-- tanto cambios de cantidad como de los límites. Si un límite es NULL no se controla ese lado (la comparación con
-- NULL no da verdadero). Se valida solo lo que tocó la sentencia (inserted): en esta base hay 201 filas que ya
-- están por debajo del punto de reposición (y 25 negativas), que no bloquean otras operaciones pero sí cualquier
-- UPDATE que las deje fuera de rango. Probado en TEST_G5 con 00000030/depósito 00 (rango 5..50): poner 20 pasa, 60
-- y 3 son rechazados.
GO
CREATE TRIGGER dbo.tr_t17_rango_stock ON dbo.STOCK
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM inserted ins
                WHERE ins.stoc_cantidad < ins.stoc_punto_reposicion   -- por debajo del mínimo
                   OR ins.stoc_cantidad > ins.stoc_stock_maximo )     -- por encima del máximo
    BEGIN
        RAISERROR('El stock debe quedar entre el punto de reposicion y el stock maximo', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba: 00000030 en depósito 00 tiene 10 (punto de reposición 5, máximo 50)
BEGIN TRANSACTION;
    UPDATE STOCK SET stoc_cantidad = 20 WHERE stoc_producto = '00000030' AND stoc_deposito = '00';  -- dentro del rango: pasa
    SELECT stoc_cantidad FROM STOCK WHERE stoc_producto = '00000030' AND stoc_deposito = '00';      -- 20
ROLLBACK;
GO
PRINT 'Prueba: subir a 60';
BEGIN TRANSACTION;
    UPDATE STOCK SET stoc_cantidad = 60 WHERE stoc_producto = '00000030' AND stoc_deposito = '00';  -- > 50: RECHAZADO
IF @@TRANCOUNT > 0 ROLLBACK;
GO
PRINT 'Prueba: bajar a 3';
BEGIN TRANSACTION;
    UPDATE STOCK SET stoc_cantidad = stoc_cantidad - 7 WHERE stoc_producto = '00000030' AND stoc_deposito = '00';  -- 3 < 5: RECHAZADO
IF @@TRANCOUNT > 0 ROLLBACK;
GO
SELECT stoc_cantidad FROM STOCK WHERE stoc_producto = '00000030' AND stoc_deposito = '00';  -- sigue en 10
GO

-- DROP TRIGGER dbo.tr_t17_rango_stock;
```

```sql
-- EJERCICIO T18
-- NOTAS: Trigger AFTER INSERT, UPDATE sobre Factura: para cada (cliente, año, mes) que tocó la sentencia, si
-- SUM(fact_total) del mes supera clie_limite_credito, rechaza todo con RAISERROR + ROLLBACK. Se suma fact_total
-- sin JOIN a Item_Factura (no hay multiplicación de filas). Cubre altas, cambios de importe y facturas que se
-- mueven de cliente o de mes (se mira el estado nuevo, inserted). Las bajas no pueden violar la regla. Supuesto:
-- bajar el límite de un cliente no invalida lo ya facturado (no se pone trigger en Cliente). En esta base hay 89
-- (cliente, mes) que ya superan el límite; como se validan solo los meses tocados, no bloquean el resto. Probado
-- en TEST_G5 con el cliente 01502 (límite 7534, julio 2012 ya facturado 346.91): una factura de 7000 pasa, una de
-- 7200 y dos de 4000 en la misma sentencia son rechazadas.
GO
CREATE TRIGGER dbo.tr_t18_limite_credito ON dbo.Factura
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM Factura f
                JOIN Cliente c ON c.clie_codigo = f.fact_cliente
                WHERE EXISTS ( SELECT 1 FROM inserted ins               -- solo los (cliente, mes) tocados
                               WHERE ins.fact_cliente = f.fact_cliente
                                 AND YEAR(ins.fact_fecha) = YEAR(f.fact_fecha)
                                 AND MONTH(ins.fact_fecha) = MONTH(f.fact_fecha) )
                GROUP BY f.fact_cliente, YEAR(f.fact_fecha), MONTH(f.fact_fecha), c.clie_limite_credito
                HAVING SUM(f.fact_total) > c.clie_limite_credito )      -- lo facturado en el mes supera el límite
    BEGIN
        RAISERROR('El cliente supera su limite de credito mensual', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba: cliente 01502, límite 7534, en julio 2012 ya tiene 346.91 facturado
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 7000, 0, '01502');      -- 7346.91 <= 7534: pasa
    SELECT SUM(fact_total) AS julio_2012 FROM Factura WHERE fact_cliente = '01502' AND YEAR(fact_fecha) = 2012 AND MONTH(fact_fecha) = 7;
ROLLBACK;
GO
PRINT 'Prueba: una factura de 7200';
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 7200, 0, '01502');      -- 7546.91 > 7534: RECHAZADO
IF @@TRANCOUNT > 0 ROLLBACK;
GO
PRINT 'Prueba: dos facturas de 4000 en la misma sentencia';
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999998', '2012-07-20', 3, 4000, 0, '01502'),
           ('A', '0003', '99999999', '2012-07-21', 3, 4000, 0, '01502');      -- cada una entra, juntas no: RECHAZADO
IF @@TRANCOUNT > 0 ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t18_limite_credito;
```

### EJERCICIO T19
Cree el/los objetos de base de datos necesarios para que se cumpla la siguiente regla de negocio automáticamente: “Ningún jefe puede tener menos de 5 años de antigüedad y tampoco puede tener más del 50% del personal a su cargo (contando directos e indirectos) a excepción del gerente general”. Se sabe que en la actualidad la regla se cumple y existe un único gerente general.

```sql
-- EJERCICIO T19
-- NOTAS: Función recursiva (cantidad de empleados a cargo, directos + indirectos) + trigger AFTER INSERT, UPDATE,
-- DELETE sobre Empleado que rechaza con RAISERROR + ROLLBACK. Supuestos: "jefe" = tiene al menos un subordinado;
-- gerente general = el único con empl_jefe NULL (exceptuado de ambas condiciones); antigüedad = 5 años cumplidos
-- desde empl_ingreso a hoy (un jefe sin fecha de ingreso no cumple); "personal" = total de filas de Empleado. Se
-- revisan TODOS los jefes (son pocos) porque un alta/baja cambia el total de personal, que es el denominador del
-- porcentaje de todos. OJO: en esta base la regla NO se cumple hoy (el jefe 3 tiene 5 de 9 empleados = 55%), así
-- que la prueba primero pasa los empleados 4 y 5 al gerente (ese UPDATE deja la regla cumplida y el trigger lo
-- acepta) y después muestra dos rechazos: un jefe con ingreso 2024 y volver a darle 5 empleados al jefe 3. Probado
-- en TEST_G5.
GO
-- Cantidad de empleados a cargo (directos + indirectos)
CREATE FUNCTION dbo.fn_t19_cant_a_cargo (@jefe NUMERIC(6))
RETURNS INT
AS
BEGIN
    RETURN ( SELECT ISNULL(SUM(1 + dbo.fn_t19_cant_a_cargo(e.empl_codigo)), 0)   -- cada directo + los suyos
             FROM Empleado e
             WHERE e.empl_jefe = @jefe );
END
GO

CREATE TRIGGER dbo.tr_t19_reglas_jefe ON dbo.Empleado
AFTER INSERT, UPDATE, DELETE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM Empleado j
                WHERE j.empl_jefe IS NOT NULL                                              -- no es el gerente general
                  AND EXISTS ( SELECT 1 FROM Empleado e WHERE e.empl_jefe = j.empl_codigo ) -- es jefe
                  AND (    j.empl_ingreso IS NULL
                        OR DATEADD(YEAR, 5, j.empl_ingreso) > GETDATE()                    -- menos de 5 años
                        OR dbo.fn_t19_cant_a_cargo(j.empl_codigo) > 0.5 * ( SELECT COUNT(*) FROM Empleado ) ) )  -- más del 50%
    BEGIN
        RAISERROR('Un jefe (salvo el gerente general) debe tener 5 o mas anios de antiguedad y no mas del 50%% del personal a cargo', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba 1: dejar la regla cumplida (4 y 5 pasan al gerente) y dar de alta un empleado nuevo => pasa
BEGIN TRANSACTION;
    UPDATE Empleado SET empl_jefe = 1 WHERE empl_codigo IN (4, 5);    -- el jefe 3 queda con 3 de 9: pasa
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_ingreso, empl_salario, empl_jefe)
    VALUES (10, 'Nuevo', 'Prueba', '2024-01-01', 1000, 1);            -- no es jefe: pasa
    SELECT empl_codigo, empl_ingreso, dbo.fn_t19_cant_a_cargo(empl_codigo) AS a_cargo
    FROM Empleado WHERE empl_codigo IN (1, 2, 3, 10);                 -- 1 -> 9, 2 -> 1, 3 -> 3, 10 -> 0
ROLLBACK;
GO
-- Prueba 2: el empleado 10 (ingreso 2024) pasa a ser jefe => RECHAZADO
BEGIN TRANSACTION;
    UPDATE Empleado SET empl_jefe = 1 WHERE empl_codigo IN (4, 5);
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_ingreso, empl_salario, empl_jefe) VALUES (10, 'Nuevo', 'Prueba', '2024-01-01', 1000, 1);
    INSERT INTO Empleado (empl_codigo, empl_nombre, empl_apellido, empl_ingreso, empl_salario, empl_jefe) VALUES (11, 'Otro', 'Prueba', '2024-01-01', 900, 10);
IF @@TRANCOUNT > 0 ROLLBACK;
GO
-- Prueba 3: el jefe 3 vuelve a tener 5 de 9 (55%) => RECHAZADO
BEGIN TRANSACTION;
    UPDATE Empleado SET empl_jefe = 1 WHERE empl_codigo IN (4, 5);
    UPDATE Empleado SET empl_jefe = 3 WHERE empl_codigo IN (4, 5);
IF @@TRANCOUNT > 0 ROLLBACK;
GO
SELECT empl_codigo, empl_jefe FROM Empleado WHERE empl_codigo IN (4, 5);  -- siguen con jefe 3
GO

-- DROP TRIGGER dbo.tr_t19_reglas_jefe;
-- DROP FUNCTION dbo.fn_t19_cant_a_cargo;
```

Crear el/los objeto/s necesarios para mantener actualizadas las comisiones del vendedor. El cálculo de la comisión está dado por el 5% de la venta total efectuada por ese vendedor en ese mes, más un 3% adicional en caso de que ese vendedor haya vendido por lo menos 50 productos distintos en el mes.

```sql
-- EJERCICIO T20
-- NOTAS: Como la comisión es "por mes", no alcanza con empl_comision (un solo valor que además hoy es una tasa):
-- se crea la tabla dbo.Comision_Vendedor (anio, mes, vendedor) + función de cálculo + carga inicial + triggers en
-- las dos tablas fuente (Factura: altas/bajas/cambios de total, fecha o vendedor; Item_Factura: cambian los
-- productos distintos), con el molde "recalcular lo afectado". Venta total = SUM(fact_total) del vendedor en el
-- mes (sin JOIN, para no multiplicar filas); el 3% adicional es sobre esa misma venta total (8% en total). Probado
-- en TEST_G5: la carga inicial genera 65 filas (55 con >= 50 productos distintos); vendedor 5 en 06/2012: 0.08 *
-- 23809.28 = 1904.74; una factura nueva de 1000 en ese mes la sube a 1984.74; una factura de 1000 del vendedor 7
-- en 08/2012 con un solo producto crea la fila con 50.00 y al borrarla desaparece.
GO
CREATE TABLE dbo.Comision_Vendedor (
    anio      INT           NOT NULL,
    mes       INT           NOT NULL,
    vendedor  NUMERIC(6)    NOT NULL,
    comision  DECIMAL(12,2) NOT NULL,
    CONSTRAINT PK_Comision_Vendedor PRIMARY KEY (anio, mes, vendedor)
);
GO

-- Comisión de un vendedor en un mes
CREATE FUNCTION dbo.fn_t20_comision (@vendedor NUMERIC(6), @anio INT, @mes INT)
RETURNS DECIMAL(12,2)
AS
BEGIN
    DECLARE @total DECIMAL(14,2), @distintos INT;
    SELECT @total = ISNULL(SUM(f.fact_total), 0)                 -- venta total (solo cabecera: sin multiplicar filas)
    FROM Factura f
    WHERE f.fact_vendedor = @vendedor AND YEAR(f.fact_fecha) = @anio AND MONTH(f.fact_fecha) = @mes;

    SELECT @distintos = COUNT(DISTINCT i.item_producto)
    FROM Factura f
    JOIN Item_Factura i ON i.item_tipo = f.fact_tipo AND i.item_sucursal = f.fact_sucursal AND i.item_numero = f.fact_numero
    WHERE f.fact_vendedor = @vendedor AND YEAR(f.fact_fecha) = @anio AND MONTH(f.fact_fecha) = @mes;

    RETURN @total * 0.05 + CASE WHEN @distintos >= 50 THEN @total * 0.03 ELSE 0 END;
END
GO

-- Carga inicial (lo que ya existe)
CREATE PROCEDURE dbo.sp_t20_cargar_comisiones
AS
BEGIN
    SET NOCOUNT ON;
    DELETE FROM dbo.Comision_Vendedor;
    INSERT INTO dbo.Comision_Vendedor (anio, mes, vendedor, comision)
    SELECT YEAR(f.fact_fecha), MONTH(f.fact_fecha), f.fact_vendedor,
           dbo.fn_t20_comision(f.fact_vendedor, YEAR(f.fact_fecha), MONTH(f.fact_fecha))
    FROM Factura f
    WHERE f.fact_vendedor IS NOT NULL
    GROUP BY YEAR(f.fact_fecha), MONTH(f.fact_fecha), f.fact_vendedor;
END
GO

-- Trigger sobre Factura: cambia la venta total (o la factura cambia de mes / de vendedor)
CREATE TRIGGER dbo.tr_t20_comision_factura ON dbo.Factura
AFTER INSERT, UPDATE, DELETE
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @afectadas TABLE (anio INT, mes INT, vendedor NUMERIC(6));

    -- 1) claves afectadas: nuevas y viejas
    INSERT INTO @afectadas
    SELECT YEAR(fact_fecha), MONTH(fact_fecha), fact_vendedor FROM inserted WHERE fact_vendedor IS NOT NULL
    UNION
    SELECT YEAR(fact_fecha), MONTH(fact_fecha), fact_vendedor FROM deleted  WHERE fact_vendedor IS NOT NULL;

    -- 2) borrarlas
    DELETE cv
    FROM dbo.Comision_Vendedor cv
    JOIN @afectadas a ON a.anio = cv.anio AND a.mes = cv.mes AND a.vendedor = cv.vendedor;

    -- 3) recalcularlas (si el vendedor se quedó sin ventas ese mes, no se inserta)
    INSERT INTO dbo.Comision_Vendedor (anio, mes, vendedor, comision)
    SELECT a.anio, a.mes, a.vendedor, dbo.fn_t20_comision(a.vendedor, a.anio, a.mes)
    FROM @afectadas a
    WHERE EXISTS ( SELECT 1 FROM Factura f
                   WHERE f.fact_vendedor = a.vendedor AND YEAR(f.fact_fecha) = a.anio AND MONTH(f.fact_fecha) = a.mes );
END
GO

-- Trigger sobre Item_Factura: cambia la cantidad de productos distintos
CREATE TRIGGER dbo.tr_t20_comision_item ON dbo.Item_Factura
AFTER INSERT, UPDATE, DELETE
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @afectadas TABLE (anio INT, mes INT, vendedor NUMERIC(6));

    INSERT INTO @afectadas
    SELECT YEAR(f.fact_fecha), MONTH(f.fact_fecha), f.fact_vendedor
    FROM inserted ins
    JOIN Factura f ON f.fact_tipo = ins.item_tipo AND f.fact_sucursal = ins.item_sucursal AND f.fact_numero = ins.item_numero
    WHERE f.fact_vendedor IS NOT NULL
    UNION
    SELECT YEAR(f.fact_fecha), MONTH(f.fact_fecha), f.fact_vendedor
    FROM deleted del
    JOIN Factura f ON f.fact_tipo = del.item_tipo AND f.fact_sucursal = del.item_sucursal AND f.fact_numero = del.item_numero
    WHERE f.fact_vendedor IS NOT NULL;

    DELETE cv
    FROM dbo.Comision_Vendedor cv
    JOIN @afectadas a ON a.anio = cv.anio AND a.mes = cv.mes AND a.vendedor = cv.vendedor;

    INSERT INTO dbo.Comision_Vendedor (anio, mes, vendedor, comision)
    SELECT a.anio, a.mes, a.vendedor, dbo.fn_t20_comision(a.vendedor, a.anio, a.mes)
    FROM @afectadas a
    WHERE EXISTS ( SELECT 1 FROM Factura f
                   WHERE f.fact_vendedor = a.vendedor AND YEAR(f.fact_fecha) = a.anio AND MONTH(f.fact_fecha) = a.mes );
END
GO

-- Prueba
EXEC dbo.sp_t20_cargar_comisiones;
SELECT COUNT(*) AS filas FROM dbo.Comision_Vendedor;                                  -- 65
SELECT * FROM dbo.Comision_Vendedor WHERE anio = 2012 AND mes = 6 AND vendedor = 5;   -- 1904.74
GO
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999998', '2012-06-20', 5, 1000, 0, '01502'),      -- mes con >= 50 distintos: +8% de 1000
           ('A', '0003', '99999999', '2012-08-01', 7, 1000, 0, '01502');      -- mes nuevo
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00006404', 1, 1000);                    -- 1 producto distinto: solo 5%
    SELECT * FROM dbo.Comision_Vendedor WHERE (anio = 2012 AND mes = 6 AND vendedor = 5) OR (anio = 2012 AND mes = 8);  -- 1984.74 y 50.00
    DELETE FROM Item_Factura WHERE item_numero = '99999999';
    DELETE FROM Factura WHERE fact_numero = '99999999';
    SELECT COUNT(*) AS filas_agosto FROM dbo.Comision_Vendedor WHERE anio = 2012 AND mes = 8;   -- 0
ROLLBACK;
GO

-- DROP TRIGGER dbo.tr_t20_comision_item;
-- DROP TRIGGER dbo.tr_t20_comision_factura;
-- DROP PROCEDURE dbo.sp_t20_cargar_comisiones;
-- DROP FUNCTION dbo.fn_t20_comision;
-- DROP TABLE dbo.Comision_Vendedor;
```

```sql
-- EJERCICIO T21
-- NOTAS: Trigger AFTER INSERT, UPDATE sobre Item_Factura: si alguna factura tocada por la sentencia queda con
-- COUNT(DISTINCT prod_familia) > 1, RAISERROR + ROLLBACK (se deshace toda la transacción, así que si la cabecera
-- se grabó en la misma transacción tampoco queda). Segundo trigger sobre Producto (UPDATE de prod_familia):
-- cambiar la familia de un producto también puede romper la regla en las facturas donde aparece. En esta base 2494
-- facturas ya mezclan familias (la regla no se cumple hoy); como se valida solo lo tocado, no bloquean el resto.
-- Probado en TEST_G5: una factura con 00001104 y 00006404 (familia 121) se graba; agregarle 00006411 (familia 111)
-- o insertar 00001104 + 00001123 (121 y 123) se rechaza; pasar 00006404 (familia 121, en una factura con 00001104)
-- a la familia 111 también se rechaza.
GO
CREATE TRIGGER dbo.tr_t21_una_familia ON dbo.Item_Factura
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM Item_Factura i
                JOIN Producto p ON p.prod_codigo = i.item_producto
                WHERE EXISTS ( SELECT 1 FROM inserted ins                     -- solo las facturas tocadas
                               WHERE ins.item_tipo = i.item_tipo AND ins.item_sucursal = i.item_sucursal AND ins.item_numero = i.item_numero )
                GROUP BY i.item_tipo, i.item_sucursal, i.item_numero
                HAVING COUNT(DISTINCT p.prod_familia) > 1 )
    BEGIN
        RAISERROR('Una factura no puede contener productos de diferentes familias', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Si cambia la familia de un producto, revisar las facturas donde aparece
CREATE TRIGGER dbo.tr_t21_familia_producto ON dbo.Producto
AFTER UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF NOT UPDATE(prod_familia) RETURN;
    IF EXISTS ( SELECT 1
                FROM Item_Factura i
                JOIN Producto p ON p.prod_codigo = i.item_producto
                WHERE EXISTS ( SELECT 1 FROM Item_Factura i2                  -- facturas que tienen algún producto modificado
                               JOIN inserted ins ON ins.prod_codigo = i2.item_producto
                               WHERE i2.item_tipo = i.item_tipo AND i2.item_sucursal = i.item_sucursal AND i2.item_numero = i.item_numero )
                GROUP BY i.item_tipo, i.item_sucursal, i.item_numero
                HAVING COUNT(DISTINCT p.prod_familia) > 1 )
    BEGIN
        RAISERROR('El cambio de familia deja facturas con productos de diferentes familias', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- Prueba 1: dos productos de la familia 121 => se graba
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 3.51), ('A', '0003', '99999999', '00006404', 1, 75.20);
    SELECT COUNT(*) AS renglones FROM Item_Factura WHERE item_numero = '99999999';   -- 2
ROLLBACK;
GO
-- Prueba 2: agregarle un producto de la familia 111 => RECHAZADO
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 3.51), ('A', '0003', '99999999', '00006404', 1, 75.20);
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00006411', 1, 6.24);
IF @@TRANCOUNT > 0 ROLLBACK;
GO
-- Prueba 3: factura nueva con 121 y 123 en la misma sentencia => RECHAZADO
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 3.51), ('A', '0003', '99999999', '00001123', 1, 1.00);
IF @@TRANCOUNT > 0 ROLLBACK;
GO
-- Prueba 4: cambiar de familia un producto de la factura de prueba => RECHAZADO
BEGIN TRANSACTION;
    INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
    VALUES ('A', '0003', '99999999', '2012-07-20', 3, 0, 0, '01502');
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001104', 1, 3.51), ('A', '0003', '99999999', '00006404', 1, 75.20);
    UPDATE Producto SET prod_familia = '111' WHERE prod_codigo = '00006404';
IF @@TRANCOUNT > 0 ROLLBACK;
GO
SELECT COUNT(*) AS quedaron FROM Factura WHERE fact_numero = '99999999';   -- 0
GO

-- DROP TRIGGER dbo.tr_t21_familia_producto;
-- DROP TRIGGER dbo.tr_t21_una_familia;
```

### EJERCICIO T22
Se requiere recategorizar los rubros de productos, de forma tal que nigun rubro tenga más de 20 productos asignados, si un rubro tiene más de 20 productos asignados se deberan distribuir en otros rubros que no tengan mas de 20 productos y si no entran se debra crear un nuevo rubro en la misma familia con la descirpción “RUBRO REASIGNADO”, cree el/los objetos de base de datos necesarios para que dicha regla de negocio quede implementada.

```sql
-- EJERCICIO T22
-- NOTAS: Procedimiento para corregir lo existente + trigger para que la regla se siga cumpliendo. Cada rubro
-- conserva sus 20 primeros productos (por código) y los sobrantes se mueven uno a uno: primero a rubros existentes
-- con lugar (< 20), si no a un "RUBRO REASIGNADO" que tenga solo productos de la misma familia, y si no se crea
-- uno nuevo (código = máximo + 1). Rubro no tiene columna de familia, así que "en la misma familia" se interpreta
-- como: cada rubro nuevo agrupa productos de una sola familia. Prueba (tarda ~1 minuto por el cursor): antes 22 de
-- 31 rubros con más de 20 productos (2190 productos); después 0, los rubros chicos quedan llenos en 20 (ej. 0042
-- pasa de 3 a 20), se crean 114 "RUBRO REASIGNADO" (ninguno mezcla familias) y el trigger rechaza el INSERT de un
-- producto en un rubro con 20.
GO
CREATE PROCEDURE dbo.sp_t22_recategorizar_rubros
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @prod CHAR(8), @rubro CHAR(4), @fami CHAR(3), @destino CHAR(4);

    DECLARE c_sobrantes CURSOR STATIC FOR            -- STATIC: foto de los sobrantes antes de empezar a mover
        SELECT p.prod_codigo, p.prod_rubro, p.prod_familia
        FROM Producto p
        WHERE ( SELECT COUNT(*) FROM Producto p2
                WHERE p2.prod_rubro = p.prod_rubro AND p2.prod_codigo <= p.prod_codigo ) > 20   -- del 21 en adelante sobra
        ORDER BY p.prod_familia, p.prod_codigo;      -- por familia, así los rubros nuevos se llenan con una sola familia

    BEGIN TRANSACTION;
    OPEN c_sobrantes;
    FETCH NEXT FROM c_sobrantes INTO @prod, @rubro, @fami;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        SET @destino = NULL;
        SELECT TOP 1 @destino = r.rubr_id            -- un rubro con lugar
        FROM Rubro r
        WHERE r.rubr_id <> @rubro
          AND ( SELECT COUNT(*) FROM Producto p WHERE p.prod_rubro = r.rubr_id ) < 20
          AND ( r.rubr_detalle <> 'RUBRO REASIGNADO'  -- un reasignado solo sirve si es de la misma familia
                OR NOT EXISTS ( SELECT 1 FROM Producto p
                                WHERE p.prod_rubro = r.rubr_id AND p.prod_familia <> @fami ) )
        ORDER BY CASE WHEN r.rubr_detalle = 'RUBRO REASIGNADO' THEN 1 ELSE 0 END, r.rubr_id;  -- primero los rubros originales

        IF @destino IS NULL                          -- no entra en ninguno: rubro nuevo
        BEGIN
            SELECT @destino = RIGHT('0000' + CAST(MAX(CAST(rubr_id AS INT)) + 1 AS VARCHAR(4)), 4) FROM Rubro;
            INSERT INTO Rubro (rubr_id, rubr_detalle) VALUES (@destino, 'RUBRO REASIGNADO');
        END

        UPDATE Producto SET prod_rubro = @destino WHERE prod_codigo = @prod;
        FETCH NEXT FROM c_sobrantes INTO @prod, @rubro, @fami;
    END
    CLOSE c_sobrantes;
    DEALLOCATE c_sobrantes;
    COMMIT TRANSACTION;
END
GO

-- De acá en adelante la regla se valida sola: ningún alta/cambio puede dejar un rubro con más de 20
CREATE TRIGGER dbo.tr_t22_max_20_por_rubro ON dbo.Producto
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM Producto p
                WHERE p.prod_rubro IN ( SELECT prod_rubro FROM inserted )   -- solo los rubros tocados
                GROUP BY p.prod_rubro
                HAVING COUNT(*) > 20 )
    BEGIN
        RAISERROR('Un rubro no puede tener mas de 20 productos', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- PRUEBA (dentro de una transacción que se deshace)
BEGIN TRANSACTION;
SELECT COUNT(*) AS rubros_con_mas_de_20_ANTES
FROM Rubro r WHERE ( SELECT COUNT(*) FROM Producto p WHERE p.prod_rubro = r.rubr_id ) > 20;

EXEC dbo.sp_t22_recategorizar_rubros;

SELECT COUNT(*) AS rubros_con_mas_de_20_DESPUES
FROM Rubro r WHERE ( SELECT COUNT(*) FROM Producto p WHERE p.prod_rubro = r.rubr_id ) > 20;
SELECT COUNT(*) AS rubros_reasignados_nuevos,
       ( SELECT COUNT(*) FROM Rubro r2
         WHERE r2.rubr_detalle = 'RUBRO REASIGNADO'
           AND ( SELECT COUNT(DISTINCT p.prod_familia) FROM Producto p WHERE p.prod_rubro = r2.rubr_id ) > 1 ) AS nuevos_que_mezclan_familias
FROM Rubro r WHERE r.rubr_detalle = 'RUBRO REASIGNADO';
SELECT ( SELECT COUNT(*) FROM Producto WHERE prod_rubro = '0042' ) AS rubro_0042_tenia_3,   -- los rubros chicos se llenan hasta 20
       ( SELECT COUNT(*) FROM Producto WHERE prod_rubro = '0009' ) AS rubro_0009_tenia_294; -- los grandes quedan en 20
-- el trigger rechaza un producto nuevo en un rubro que ya tiene 20
BEGIN TRY
    INSERT INTO Producto (prod_codigo, prod_detalle, prod_precio, prod_familia, prod_rubro, prod_envase)
    VALUES ('99999999', 'PRUEBA T22', 1, '161', '0016', 1);
END TRY
BEGIN CATCH
    SELECT ERROR_MESSAGE() AS rechazado;
END CATCH
IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
GO
SELECT COUNT(*) AS rubros_tras_rollback FROM Rubro;   -- vuelve a 31
GO

-- DROP TRIGGER dbo.tr_t22_max_20_por_rubro;
-- DROP PROCEDURE dbo.sp_t22_recategorizar_rubros;
```

### EJERCICIO T24
Se requiere recategorizar los encargados asignados a los depositos. Para ello cree el o los objetos de bases de datos necesarios que lo resueva, teniendo en cuenta que un deposito no puede tener como encargado un empleado que pertenezca a un departamento que no sea de la misma zona que el deposito, si esto ocurre a dicho deposito debera asignársele el empleado con menos depositos asignados que pertenezca a un departamento de esa zona.

```sql
-- EJERCICIO T24
-- NOTAS: Procedimiento con cursor: recorre los depósitos mal asignados y a cada uno le da el empleado de un
-- departamento de su zona con menos depósitos, contando en el momento (así las asignaciones se van repartiendo);
-- desempate por código. Si en la zona del depósito no hay ningún departamento (o el depósito no tiene zona) no hay
-- a quién asignar y se deja como está: en la base solo hay departamentos en las zonas 003, 004 y 005, así que 24
-- de los 33 depósitos (16 en otras zonas + 8 sin zona) no se pueden corregir. Prueba: de los 9 depósitos de esas
-- zonas, 4 estaban mal (02, 06, 10, 11): 02 y 06 (zona 005) pasan al empleado 6 (0 depósitos, el 2 tenía 10) y 10
-- y 11 (zona 003) al 1, único empleado de Administración; los otros 5 no cambian.
GO
CREATE PROCEDURE dbo.sp_t24_reasignar_encargados
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @depo CHAR(2), @zona CHAR(3), @empl NUMERIC(6);

    DECLARE c_depos CURSOR FOR
        SELECT d.depo_codigo, d.depo_zona
        FROM DEPOSITO d
        WHERE NOT EXISTS ( SELECT 1                                  -- el encargado NO es de un depto de la zona
                           FROM Empleado e
                           JOIN Departamento dp ON dp.depa_codigo = e.empl_departamento
                           WHERE e.empl_codigo = d.depo_encargado AND dp.depa_zona = d.depo_zona )
          AND EXISTS ( SELECT 1 FROM Departamento dp WHERE dp.depa_zona = d.depo_zona )   -- y hay a quién asignarle
        ORDER BY d.depo_codigo;

    BEGIN TRANSACTION;
    OPEN c_depos;
    FETCH NEXT FROM c_depos INTO @depo, @zona;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        SELECT TOP 1 @empl = e.empl_codigo                           -- el de la zona con menos depósitos (contados ahora)
        FROM Empleado e
        JOIN Departamento dp ON dp.depa_codigo = e.empl_departamento
        WHERE dp.depa_zona = @zona
        ORDER BY ( SELECT COUNT(*) FROM DEPOSITO d WHERE d.depo_encargado = e.empl_codigo ), e.empl_codigo;

        UPDATE DEPOSITO SET depo_encargado = @empl WHERE depo_codigo = @depo;
        FETCH NEXT FROM c_depos INTO @depo, @zona;
    END
    CLOSE c_depos;
    DEALLOCATE c_depos;
    COMMIT TRANSACTION;
END
GO

-- PRUEBA (antes / después, y se deshace)
BEGIN TRANSACTION;
SELECT d.depo_codigo, d.depo_zona, d.depo_encargado AS encargado_ANTES, dp.depa_zona AS zona_del_encargado
FROM DEPOSITO d
LEFT JOIN Empleado e ON e.empl_codigo = d.depo_encargado
LEFT JOIN Departamento dp ON dp.depa_codigo = e.empl_departamento
WHERE d.depo_zona IN ( SELECT depa_zona FROM Departamento )
ORDER BY d.depo_codigo;

EXEC dbo.sp_t24_reasignar_encargados;

SELECT d.depo_codigo, d.depo_zona, d.depo_encargado AS encargado_DESPUES, dp.depa_zona AS zona_del_encargado
FROM DEPOSITO d
LEFT JOIN Empleado e ON e.empl_codigo = d.depo_encargado
LEFT JOIN Departamento dp ON dp.depa_codigo = e.empl_departamento
WHERE d.depo_zona IN ( SELECT depa_zona FROM Departamento )
ORDER BY d.depo_codigo;
ROLLBACK TRANSACTION;
GO

-- DROP PROCEDURE dbo.sp_t24_reasignar_encargados;
```

### EJERCICIO T25
Desarrolle el/los elementos de base de datos necesarios para que no se permita que la composición de los productos sea recursiva, o sea, que si el producto A compone al producto B, dicho producto B no pueda ser compuesto por el producto A, hoy la regla se cumple.

```sql
-- EJERCICIO T25
-- NOTAS: Función recursiva + trigger AFTER INSERT, UPDATE sobre Composicion. Una fila (producto P, componente C)
-- arma un ciclo si C es P o si C contiene a P a cualquier nivel; así se cubre el caso directo del enunciado (A-B /
-- B-A) y también los ciclos indirectos (A-B-C-A). Como la regla hoy se cumple, la recursión siempre termina (como
-- mucho 32 niveles de anidamiento en SQL Server). Prueba: (00001109 compuesto por 00001104) se rechaza por
-- directo; (00006411 compuesto por 00006402) se rechaza por indirecto (6402 -> 6411); (00006404 compuesto por
-- 00006411) se acepta.
GO
-- ¿@prod contiene a @buscado, directa o indirectamente?
CREATE FUNCTION dbo.fn_t25_contiene (@prod CHAR(8), @buscado CHAR(8))
RETURNS BIT
AS
BEGIN
    IF EXISTS ( SELECT 1 FROM Composicion WHERE comp_producto = @prod AND comp_componente = @buscado )   -- caso base: directo
        RETURN 1;
    IF EXISTS ( SELECT 1 FROM Composicion
                WHERE comp_producto = @prod AND dbo.fn_t25_contiene(comp_componente, @buscado) = 1 )      -- recursivo: algún componente lo contiene
        RETURN 1;
    RETURN 0;
END
GO

CREATE TRIGGER dbo.tr_t25_composicion_no_recursiva ON dbo.Composicion
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1 FROM inserted ins
                WHERE ins.comp_componente = ins.comp_producto                           -- A compuesto por A
                   OR dbo.fn_t25_contiene(ins.comp_componente, ins.comp_producto) = 1 ) -- el componente ya contiene al producto
    BEGIN
        RAISERROR('La composicion no puede ser recursiva', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- PRUEBA
SELECT * FROM Composicion WHERE comp_producto IN ('00001104', '00006402', '00006411', '00006404');
GO
BEGIN TRANSACTION;                                   -- 1) directo: 1104 se compone de 1109 => 1109 no puede componerse de 1104
BEGIN TRY
    INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00001109', '00001104');
END TRY
BEGIN CATCH
    SELECT ERROR_MESSAGE() AS rechazado;
END CATCH
IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
GO
BEGIN TRANSACTION;                                   -- 2) indirecto: 6402 -> 6411 => 6411 no puede componerse de 6402
BEGIN TRY
    INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00006411', '00006402');
END TRY
BEGIN CATCH
    SELECT ERROR_MESSAGE() AS rechazado;
END CATCH
IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
GO
BEGIN TRANSACTION;                                   -- 3) válido: 6404 + 6411 no forman ciclo
INSERT INTO Composicion (comp_cantidad, comp_producto, comp_componente) VALUES (1, '00006404', '00006411');
SELECT COUNT(*) AS aceptada FROM Composicion WHERE comp_producto = '00006404' AND comp_componente = '00006411';   -- 1
ROLLBACK TRANSACTION;
GO

-- DROP TRIGGER dbo.tr_t25_composicion_no_recursiva;
-- DROP FUNCTION dbo.fn_t25_contiene;
```

### EJERCICIO T27
Se requiere reasignar los encargados de stock de los diferentes depósitos. Para ello se solicita que realice el o los objetos de base de datos necesarios para asignar a cada uno de los depósitos el encargado que le corresponda, entendiendo que el encargado que le corresponde es cualquier empleado que no es jefe y que no es vendedor, o sea, que no está asignado a ningun cliente, se deberán ir asignando tratando de que un empleado solo tenga un deposito asignado, en caso de no poder se irán aumentando la cantidad de depósitos progresivamente para cada empleado.

```sql
-- EJERCICIO T27
-- NOTAS: Procedimiento: se reasignan TODOS los depósitos. Primero se limpian los encargados y después, depósito
-- por depósito (por código), se asigna el candidato con menos depósitos en ese momento (desempate por código): eso
-- reparte de a uno y, cuando no alcanzan, va subiendo de a uno la cantidad por empleado (round robin). Candidato =
-- no es jefe de nadie (NOT EXISTS, porque empl_jefe tiene NULL y NOT IN fallaría) y no es clie_vendedor de ningún
-- cliente. Si no hay candidatos no se toca nada. En la base el único candidato es el empleado 6 (1, 2 y 3 son
-- jefes; 3, 4, 5, 7, 8 y 9 son vendedores), así que los 33 depósitos quedan a su cargo; antes estaban repartidos
-- entre 1, 2, 3, 4 y 5.
GO
CREATE PROCEDURE dbo.sp_t27_reasignar_encargados
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @depo CHAR(2), @empl NUMERIC(6);
    DECLARE @candidatos TABLE (empl_codigo NUMERIC(6) PRIMARY KEY);

    INSERT INTO @candidatos
    SELECT e.empl_codigo
    FROM Empleado e
    WHERE NOT EXISTS ( SELECT 1 FROM Empleado s WHERE s.empl_jefe = e.empl_codigo )     -- no es jefe
      AND NOT EXISTS ( SELECT 1 FROM Cliente c WHERE c.clie_vendedor = e.empl_codigo ); -- no es vendedor

    IF NOT EXISTS ( SELECT 1 FROM @candidatos )      -- nadie puede ser encargado: no toco nada
        RETURN;

    BEGIN TRANSACTION;
    UPDATE DEPOSITO SET depo_encargado = NULL;       -- arranco de cero para contar solo lo nuevo

    DECLARE c_depos CURSOR FOR
        SELECT depo_codigo FROM DEPOSITO ORDER BY depo_codigo;
    OPEN c_depos;
    FETCH NEXT FROM c_depos INTO @depo;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        SELECT TOP 1 @empl = ca.empl_codigo          -- el que menos depósitos tiene hasta ahora
        FROM @candidatos ca
        ORDER BY ( SELECT COUNT(*) FROM DEPOSITO d WHERE d.depo_encargado = ca.empl_codigo ), ca.empl_codigo;

        UPDATE DEPOSITO SET depo_encargado = @empl WHERE depo_codigo = @depo;
        FETCH NEXT FROM c_depos INTO @depo;
    END
    CLOSE c_depos;
    DEALLOCATE c_depos;
    COMMIT TRANSACTION;
END
GO

-- PRUEBA (antes / después, y se deshace)
BEGIN TRANSACTION;
SELECT depo_encargado AS encargado_ANTES, COUNT(*) AS depositos FROM DEPOSITO GROUP BY depo_encargado;

EXEC dbo.sp_t27_reasignar_encargados;

SELECT depo_encargado AS encargado_DESPUES, COUNT(*) AS depositos FROM DEPOSITO GROUP BY depo_encargado;
GO
-- para ver el reparto progresivo: hago que 4 y 5 dejen de ser vendedores (dentro de la misma transacción)
UPDATE Cliente SET clie_vendedor = 3 WHERE clie_vendedor IN (4, 5);
EXEC dbo.sp_t27_reasignar_encargados;
SELECT depo_encargado AS encargado_CON_3_CANDIDATOS, COUNT(*) AS depositos FROM DEPOSITO GROUP BY depo_encargado;   -- 11 / 11 / 11
ROLLBACK TRANSACTION;
GO

-- DROP PROCEDURE dbo.sp_t27_reasignar_encargados;
```

```sql
-- EJERCICIO T28
-- NOTAS: Procedimiento con un solo UPDATE (no hace falta cursor): mandadero TOP 1 por cliente (más facturas;
-- desempate por monto vendido a ese cliente y después por código) y, si da NULL porque no tiene facturas, el
-- vendedor con mayor SUM(fact_total) de la empresa (calculado una vez en una variable). Facturas con fact_vendedor
-- NULL no cuentan. Prueba: 3687 clientes, 3617 sin facturas; el vendedor top de la empresa es el 4 (266819.83).
-- Antes: 3 -> 1999, 4 -> 1636, 22 NULL, resto 30. Después: 4 -> 3637, 3 -> 20, 7 -> 9, 8 -> 9, 5 -> 6, 9 -> 6
-- (ninguno en NULL); ej. el cliente 01742 queda con el 4, que le hizo 203 de sus 204 facturas.
GO
CREATE PROCEDURE dbo.sp_t28_reasignar_vendedores
AS
BEGIN
    SET NOCOUNT ON;
    DECLARE @vendedor_top NUMERIC(6);

    SELECT TOP 1 @vendedor_top = f.fact_vendedor     -- el que más vendió en monto en toda la empresa
    FROM Factura f
    WHERE f.fact_vendedor IS NOT NULL
    GROUP BY f.fact_vendedor
    ORDER BY SUM(f.fact_total) DESC, f.fact_vendedor;

    BEGIN TRANSACTION;
    UPDATE c
    SET c.clie_vendedor = ISNULL(
            ( SELECT TOP 1 f.fact_vendedor           -- el que más facturas le hizo a este cliente
              FROM Factura f
              WHERE f.fact_cliente = c.clie_codigo AND f.fact_vendedor IS NOT NULL
              GROUP BY f.fact_vendedor
              ORDER BY COUNT(*) DESC, SUM(f.fact_total) DESC, f.fact_vendedor ),
            @vendedor_top )                          -- sin facturas: el top de la empresa
    FROM Cliente c;
    COMMIT TRANSACTION;
END
GO

-- PRUEBA (antes / después, y se deshace)
BEGIN TRANSACTION;
SELECT clie_vendedor AS vendedor_ANTES, COUNT(*) AS clientes FROM Cliente GROUP BY clie_vendedor ORDER BY 2 DESC;

EXEC dbo.sp_t28_reasignar_vendedores;

SELECT clie_vendedor AS vendedor_DESPUES, COUNT(*) AS clientes FROM Cliente GROUP BY clie_vendedor ORDER BY 2 DESC;
-- control a mano de un cliente con facturas
SELECT f.fact_vendedor, COUNT(*) AS facturas, ( SELECT clie_vendedor FROM Cliente WHERE clie_codigo = '01742' ) AS asignado
FROM Factura f WHERE f.fact_cliente = '01742'
GROUP BY f.fact_vendedor ORDER BY 2 DESC;
ROLLBACK TRANSACTION;
GO

-- DROP PROCEDURE dbo.sp_t28_reasignar_vendedores;
```

```sql
-- EJERCICIO T29
-- NOTAS: Interpretación: los componentes que aparecen en una factura tienen que ser todos del mismo producto
-- compuesto; si la factura tiene componentes de 2 o más compuestos distintos se rechaza. Se cuenta COUNT(DISTINCT
-- comp_producto) sobre la factura entera ya modificada (los renglones pueden llegar en varios INSERT), solo para
-- las facturas tocadas; de paso también rechaza un producto que sea componente de dos compuestos distintos (la
-- otra lectura posible). Supone cabecera y renglones en la misma transacción, así el ROLLBACK no deja la factura.
-- En la base hay 5 facturas viejas que no cumplen y no se revisan. Prueba: 00001109 + 00001123 (ambos de 00001104)
-- se graban; agregar 00001475 (componente de 00001718) se rechaza.
GO
CREATE TRIGGER dbo.tr_t29_componentes_de_un_solo_producto ON dbo.Item_Factura
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM Item_Factura i
                JOIN Composicion c ON c.comp_componente = i.item_producto      -- solo los renglones que son componentes
                WHERE EXISTS ( SELECT 1 FROM inserted ins                     -- de las facturas tocadas
                               WHERE ins.item_tipo = i.item_tipo AND ins.item_sucursal = i.item_sucursal AND ins.item_numero = i.item_numero )
                GROUP BY i.item_tipo, i.item_sucursal, i.item_numero
                HAVING COUNT(DISTINCT c.comp_producto) > 1 )                  -- componentes de más de un compuesto
    BEGIN
        RAISERROR('Una factura no puede contener componentes de diferentes productos', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- PRUEBA
BEGIN TRANSACTION;
INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
VALUES ('A', '0003', '99999999', '2012-07-20', 4, 0, 0, '01669');
INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
VALUES ('A', '0003', '99999999', '00001109', 1, 10),     -- componente de 00001104
       ('A', '0003', '99999999', '00001123', 2, 10),     -- componente de 00001104 (mismo compuesto: ok)
       ('A', '0003', '99999999', '00001101', 1, 10);     -- no es componente
SELECT COUNT(*) AS renglones_grabados FROM Item_Factura WHERE item_numero = '99999999';   -- 3
BEGIN TRY
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001475', 1, 10); -- componente de 00001718: otro compuesto, se rechaza
END TRY
BEGIN CATCH
    SELECT ERROR_MESSAGE() AS rechazado;
END CATCH
IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
SELECT COUNT(*) AS factura_quedo FROM Factura WHERE fact_numero = '99999999';              -- 0
GO

-- DROP TRIGGER dbo.tr_t29_componentes_de_un_solo_producto;
```

```sql
-- EJERCICIO T30
-- NOTAS: Trigger AFTER INSERT sobre Item_Factura (como las facturas no se modifican, alcanza con el INSERT de
-- renglones; la cabecera sola no tiene unidades). Para cada (cliente, producto, año-mes) que tocó inserted se suma
-- todo lo comprado ese mes, incluidas las filas recién insertadas, y si pasa de 100 se rechaza. Ojo: en esta base
-- la regla NO se cumple (ej. el cliente 01742 compró 1700 unidades de 00001718 en 03/2012); el trigger solo mira
-- lo que se toca, así que eso no molesta salvo que se le venda más a ese cliente ese mes. Prueba: cliente 01669,
-- producto 00001101, 08/2012: una factura con 60 unidades se graba; otra del mismo mes con 50 (total 110) se
-- rechaza con el mensaje pedido.
GO
CREATE TRIGGER dbo.tr_t30_max_100_unidades_mes ON dbo.Item_Factura
AFTER INSERT
AS
BEGIN
    SET NOCOUNT ON;
    IF EXISTS ( SELECT 1
                FROM inserted ins
                JOIN Factura f ON f.fact_tipo = ins.item_tipo AND f.fact_sucursal = ins.item_sucursal AND f.fact_numero = ins.item_numero
                WHERE ( SELECT SUM(i2.item_cantidad)  -- todo lo que ese cliente compró de ese producto ese mes
                        FROM Item_Factura i2
                        JOIN Factura f2 ON f2.fact_tipo = i2.item_tipo AND f2.fact_sucursal = i2.item_sucursal AND f2.fact_numero = i2.item_numero
                        WHERE f2.fact_cliente = f.fact_cliente
                          AND i2.item_producto = ins.item_producto
                          AND YEAR(f2.fact_fecha) = YEAR(f.fact_fecha) AND MONTH(f2.fact_fecha) = MONTH(f.fact_fecha) ) > 100 )
    BEGIN
        RAISERROR('Se ha superado el límite máximo de compra de un producto', 16, 1);
        ROLLBACK TRANSACTION;
    END
END
GO

-- PRUEBA
BEGIN TRANSACTION;
INSERT INTO Factura (fact_tipo, fact_sucursal, fact_numero, fact_fecha, fact_vendedor, fact_total, fact_total_impuestos, fact_cliente)
VALUES ('A', '0003', '99999998', '2012-08-01', 4, 0, 0, '01669'),
       ('A', '0003', '99999999', '2012-08-20', 4, 0, 0, '01669');
INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
VALUES ('A', '0003', '99999998', '00001101', 60, 1);     -- 60 en el mes: se acepta
SELECT SUM(item_cantidad) AS unidades_en_el_mes FROM Item_Factura WHERE item_numero IN ('99999998', '99999999');   -- 60
BEGIN TRY
    INSERT INTO Item_Factura (item_tipo, item_sucursal, item_numero, item_producto, item_cantidad, item_precio)
    VALUES ('A', '0003', '99999999', '00001101', 50, 1); -- 60 + 50 = 110 en el mismo mes: se rechaza
END TRY
BEGIN CATCH
    SELECT ERROR_MESSAGE() AS rechazado;
END CATCH
IF @@TRANCOUNT > 0 ROLLBACK TRANSACTION;
GO

-- DROP TRIGGER dbo.tr_t30_max_100_unidades_mes;
```

### EJERCICIO T31
Desarrolle el o los objetos de base de datos necesarios, para que un jefe no pueda tener más de 20 empleados a cargo, directa o indirectamente, si esto ocurre debera asignarsele un jefe que cumpla esa condición, si no existe un jefe para asignarle se le deberá colocar como jefe al gerente general que es aquel que no tiene jefe.

```sql
-- EJERCICIO T31
-- NOTAS: Funciones recursivas + trigger AFTER INSERT, UPDATE sobre Empleado. Al empleado que entra (o cambia de
-- jefe) y deja a su jefe, o a algún jefe por encima, con más de 20 a cargo (directos + indirectos) se lo cuelga,
-- junto con todos los suyos, de otro empleado que YA es jefe y que, con ese agregado, sigue cumpliendo (él y sus
-- superiores); si no hay ninguno, queda bajo el gerente general. El gerente general no tiene tope (a la fuerza
-- tiene a todos a cargo). No se elige a nadie de su propio subárbol (evita ciclos). Desempate: jefe de menor
-- código; con varias filas se procesan de mayor a menor código, así se mueven los últimos. Supone
-- RECURSIVE_TRIGGERS OFF (el default) para que el UPDATE interno no redispare el trigger. Prueba (base: 1 gerente;
-- 2 tiene 1 a cargo, 3 tiene 5): alta de 20 empleados bajo el 3 => 3 queda con 20 y 5 pasan al 2 (queda con 6);
-- alta de 30 más bajo el 3 => 14 al 2 (llega a 20) y 16 al gerente.
GO
-- cantidad de empleados a cargo, directos + indirectos
CREATE FUNCTION dbo.fn_t31_cant_a_cargo (@jefe NUMERIC(6))
RETURNS INT
AS
BEGIN
    RETURN ISNULL(( SELECT COUNT(*) + SUM(dbo.fn_t31_cant_a_cargo(e.empl_codigo))   -- directos + los de cada directo
                    FROM Empleado e
                    WHERE e.empl_jefe = @jefe ), 0);                                -- sin subordinados: 0
END
GO

-- ¿@jefe (y todos sus superiores, salvo el gerente general) admite @extra empleados más sin pasar de 20?
CREATE FUNCTION dbo.fn_t31_admite (@jefe NUMERIC(6), @extra INT)
RETURNS BIT
AS
BEGIN
    DECLARE @superior NUMERIC(6) = ( SELECT empl_jefe FROM Empleado WHERE empl_codigo = @jefe );
    IF @superior IS NULL                             -- es el gerente general: sin tope
        RETURN 1;
    IF dbo.fn_t31_cant_a_cargo(@jefe) + @extra > 20
        RETURN 0;
    RETURN dbo.fn_t31_admite(@superior, @extra);     -- y que también le entre a su jefe
END
GO

-- ¿@empl está a cargo de @jefe, a cualquier nivel?
CREATE FUNCTION dbo.fn_t31_es_subordinado (@empl NUMERIC(6), @jefe NUMERIC(6))
RETURNS BIT
AS
BEGIN
    DECLARE @superior NUMERIC(6) = ( SELECT empl_jefe FROM Empleado WHERE empl_codigo = @empl );
    IF @superior IS NULL RETURN 0;
    IF @superior = @jefe RETURN 1;
    RETURN dbo.fn_t31_es_subordinado(@superior, @jefe);
END
GO

CREATE TRIGGER dbo.tr_t31_max_20_a_cargo ON dbo.Empleado
AFTER INSERT, UPDATE
AS
BEGIN
    SET NOCOUNT ON;
    IF NOT UPDATE(empl_jefe) RETURN;                 -- solo importa si cambió (o se cargó) el jefe

    DECLARE @emp NUMERIC(6), @jefe NUMERIC(6), @nuevo NUMERIC(6), @tam INT;
    DECLARE @gerente NUMERIC(6) = ( SELECT TOP 1 empl_codigo FROM Empleado
                                    WHERE empl_jefe IS NULL ORDER BY empl_codigo );

    DECLARE c_emp CURSOR LOCAL FOR                   -- uno por uno: cada movimiento cambia los conteos
        SELECT empl_codigo FROM inserted WHERE empl_jefe IS NOT NULL ORDER BY empl_codigo DESC;
    OPEN c_emp;
    FETCH NEXT FROM c_emp INTO @emp;
    WHILE @@FETCH_STATUS = 0
    BEGIN
        SELECT @jefe = empl_jefe FROM Empleado WHERE empl_codigo = @emp;
        IF dbo.fn_t31_admite(@jefe, 0) = 0           -- su jefe (o uno de arriba) se pasó de 20
        BEGIN
            SET @tam = 1 + dbo.fn_t31_cant_a_cargo(@emp);                         -- se muda con todos los suyos
            UPDATE Empleado SET empl_jefe = @gerente WHERE empl_codigo = @emp;  -- por defecto, al gerente general

            SET @nuevo = NULL;
            SELECT TOP 1 @nuevo = j.empl_codigo      -- un jefe que lo pueda recibir
            FROM Empleado j
            WHERE j.empl_codigo <> @emp
              AND j.empl_jefe IS NOT NULL            -- el gerente es el último recurso, ya está puesto
              AND EXISTS ( SELECT 1 FROM Empleado s WHERE s.empl_jefe = j.empl_codigo )   -- que ya sea jefe
              AND dbo.fn_t31_es_subordinado(j.empl_codigo, @emp) = 0                       -- que no sea uno de los suyos
              AND dbo.fn_t31_admite(j.empl_codigo, @tam) = 1
            ORDER BY j.empl_codigo;

            IF @nuevo IS NOT NULL
                UPDATE Empleado SET empl_jefe = @nuevo WHERE empl_codigo = @emp;
        END
        FETCH NEXT FROM c_emp INTO @emp;
    END
    CLOSE c_emp;
    DEALLOCATE c_emp;
END
GO

-- PRUEBA (se deshace)
BEGIN TRANSACTION;
SELECT empl_codigo AS jefe, dbo.fn_t31_cant_a_cargo(empl_codigo) AS a_cargo_ANTES
FROM Empleado WHERE empl_codigo IN (1, 2, 3);

INSERT INTO Empleado (empl_codigo, empl_nombre, empl_jefe)        -- 20 altas juntas bajo el 3 (tenía 5 => 25)
SELECT TOP 20 100 + ROW_NUMBER() OVER (ORDER BY prod_codigo), 'PRUEBA T31', 3 FROM Producto;
SELECT empl_codigo AS jefe, dbo.fn_t31_cant_a_cargo(empl_codigo) AS a_cargo_TRAS_20
FROM Empleado WHERE empl_codigo IN (1, 2, 3);

INSERT INTO Empleado (empl_codigo, empl_nombre, empl_jefe)        -- 30 más bajo el 3: el 2 se llena y el resto va al gerente
SELECT TOP 30 200 + ROW_NUMBER() OVER (ORDER BY prod_codigo), 'PRUEBA T31', 3 FROM Producto;
SELECT empl_codigo AS jefe, dbo.fn_t31_cant_a_cargo(empl_codigo) AS a_cargo_TRAS_50,
       ( SELECT COUNT(*) FROM Empleado e WHERE e.empl_jefe = j.empl_codigo AND e.empl_nombre = 'PRUEBA T31' ) AS altas_directas
FROM Empleado j WHERE empl_codigo IN (1, 2, 3);
ROLLBACK TRANSACTION;
GO

-- DROP TRIGGER dbo.tr_t31_max_20_a_cargo;
-- DROP FUNCTION dbo.fn_t31_es_subordinado;
-- DROP FUNCTION dbo.fn_t31_admite;
-- DROP FUNCTION dbo.fn_t31_cant_a_cargo;
```
