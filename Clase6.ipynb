# T-SQL: vistas, funciones y procedimientos
T-SQL es el SQL de SQL Server con extras de programación: variables (`@variable`), `IF`, `WHILE`, `BEGIN ... END`, funciones y procedimientos propios.

- `GO` no es SQL: separa **lotes**. `CREATE VIEW`, `CREATE FUNCTION` y `CREATE PROCEDURE` tienen que ser la primera sentencia de su lote, por eso se pone `GO` antes.
- Las variables se declaran con `DECLARE @nombre TIPO` y se asignan con `SET` o con `SELECT @nombre = campo`.

### Vistas
Una vista es una consulta guardada con un nombre. Se usa como si fuera una tabla pero **no guarda datos**: cada vez que la consulto se ejecuta el `SELECT`.

```sql
CREATE VIEW dbo.vw_productos_por_stock
AS
SELECT 
    prod.prod_codigo, 
    prod.prod_detalle, 
    SUM(st.stoc_cantidad) AS total_stock
FROM Producto prod
JOIN STOCK st ON st.stoc_producto = prod.prod_codigo
GROUP BY prod.prod_codigo, prod.prod_detalle
GO

SELECT * FROM dbo.vw_productos_por_stock
```

**Vista materializada (indexada):** guarda el resultado en disco. Requiere `WITH SCHEMABINDING` (la vista queda atada a las tablas, no se las puede modificar ni borrar sin borrar la vista), nombres con esquema (`dbo.Tabla`) y `COUNT_BIG(*)` si tiene `GROUP BY`. Para que realmente se materialice también hay que crearle un índice clúster único (`CREATE UNIQUE CLUSTERED INDEX ... ON dbo.vwm_productos_por_stock (...)`); ese paso no estaba en los apuntes de clase, lo agregué como nota.

```sql
CREATE VIEW dbo.vwm_productos_por_stock
WITH SCHEMABINDING
AS
SELECT 
    prod.prod_codigo, 
    prod.prod_detalle, 
    SUM(st.stoc_cantidad) AS total_stock,
    COUNT_BIG(*)          AS cantidad_filas 
FROM dbo.Producto prod
JOIN dbo.STOCK st ON st.stoc_producto = prod.prod_codigo
GROUP BY prod.prod_codigo, prod.prod_detalle
GO
```

### Funciones vs. procedimientos
| | Función | Procedimiento |
|---|---|---|
| Devuelve | un valor (o una tabla) con `RETURN` | nada obligatorio; puede devolver valores por parámetros `OUTPUT` |
| Se usa | dentro de un `SELECT` / `WHERE` | con `EXECUTE` |
| Puede modificar datos | No | Sí (`INSERT`, `UPDATE`, `DELETE`) |

### Función escalar
**Ejercicio 1:** dada un artículo y un depósito, devolver un string con el estado del depósito. Si la cantidad almacenada es menor al límite: `OCUPACION DEL DEPOSITO XX %` (XX = % de ocupación). Si es mayor o igual al límite: `DEPOSITO COMPLETO`.

Porcentaje de ocupación = cantidad × 100 / stock máximo.

```sql
CREATE FUNCTION estado_deposito_segun_articulo ( @articulo CHAR(8), @deposito CHAR(2) )
RETURNS VARCHAR(500)
AS
BEGIN
    DECLARE @estado VARCHAR(500)

    SET @estado = ( SELECT 
                        CASE WHEN s.stoc_cantidad < s.stoc_stock_maximo
                             THEN 'OCUPACION DEL DEPOSITO ' 
                                  + RTRIM(CAST(CAST(s.stoc_cantidad * 100.0 / s.stoc_stock_maximo AS DECIMAL(5,2)) AS CHAR)) + '%'
                             ELSE 'DEPOSITO COMPLETO'
                        END
                    FROM STOCK s
                    WHERE s.stoc_producto = @articulo
                      AND s.stoc_deposito = @deposito )
    RETURN @estado
END
GO

SELECT dbo.estado_deposito_segun_articulo('00000102', '11')      -- las funciones se llaman con el esquema (dbo.)
```

**Ejercicio 2:** dada un artículo y una fecha, devolver el stock que existía a esa fecha.

Stock a esa fecha = stock actual + lo que se vendió **después** de esa fecha (esas ventas ya lo habían restado).

```sql
CREATE FUNCTION stock_existente ( @articulo CHAR(8), @fecha DATE )
RETURNS DECIMAL(12,2)
AS
BEGIN
    DECLARE @stock_actual DECIMAL(12,2)
    DECLARE @ventas_posteriores DECIMAL(12,2)

    SET @stock_actual = ( SELECT SUM(ISNULL(stoc_cantidad, 0)) 
                          FROM STOCK 
                          WHERE stoc_producto = @articulo )

    SET @ventas_posteriores = ( SELECT SUM(ISNULL(item_cantidad, 0)) 
                                FROM Item_Factura 
                                JOIN Factura ON  item_numero   = fact_numero 
                                             AND item_sucursal = fact_sucursal 
                                             AND item_tipo     = fact_tipo
                                WHERE fact_fecha > @fecha 
                                  AND item_producto = @articulo )

    RETURN @stock_actual + ISNULL(@ventas_posteriores, 0)
END
GO

SELECT dbo.stock_existente('00001415', '2011-12-16')
```

### Procedimientos almacenados
Parámetros con `OUTPUT` devuelven un valor a quien llama. Para llamarlo se declara una variable y se pasa con `OUTPUT`.

**Ejercicio 3:** corregir la tabla `Empleado`. Tiene que haber un único empleado sin jefe (el gerente general). Si hay más de uno, el gerente general es el de **mayor salario** y, si empatan, el de **mayor antigüedad**. Al final la tabla tiene un solo empleado sin jefe y el procedimiento devuelve cuántos había antes.

```sql
CREATE PROCEDURE corregirTablaEmpleados ( @empleadosSinJefeAntes INT OUTPUT )
AS
BEGIN
    DECLARE @gerenteGeneral NUMERIC(6)

    SET @empleadosSinJefeAntes = ( SELECT COUNT(*) FROM Empleado WHERE empl_jefe IS NULL )

    IF @empleadosSinJefeAntes > 0
    BEGIN
        SET @gerenteGeneral = ( SELECT TOP 1 empl_codigo 
                                FROM Empleado
                                WHERE empl_jefe IS NULL
                                ORDER BY empl_salario DESC,     -- mayor salario
                                         empl_ingreso ASC )     -- si empatan, el que ingreso primero (mas antiguo)

        UPDATE Empleado
        SET empl_jefe = @gerenteGeneral
        WHERE empl_jefe IS NULL 
          AND empl_codigo != @gerenteGeneral                    -- todos los demas sin jefe pasan a tener al gerente como jefe
    END
END
GO

DECLARE @resultado INT
EXECUTE dbo.corregirTablaEmpleados @empleadosSinJefeAntes = @resultado OUTPUT
SELECT @resultado
```

**Ejercicio 4:** actualizar `empl_comision` con el total vendido por cada empleado en el último año y devolver el código del vendedor que más vendió en ese año.

```sql
CREATE PROCEDURE actualizarComision ( @masVendedor NUMERIC(6) OUTPUT )
AS
BEGIN
    DECLARE @ultimoAnio INT

    SET @ultimoAnio = ( SELECT TOP 1 YEAR(f2.fact_fecha)
                        FROM Factura f2
                        ORDER BY 1 DESC )

    UPDATE Empleado
    SET empl_comision = ( SELECT ISNULL(SUM(f1.fact_total), 0)
                          FROM Factura f1
                          WHERE YEAR(f1.fact_fecha) = @ultimoAnio 
                            AND f1.fact_vendedor = empl_codigo )

    SET @masVendedor = ( SELECT TOP 1 f1.fact_vendedor
                         FROM Factura f1
                         WHERE YEAR(f1.fact_fecha) = @ultimoAnio
                         GROUP BY f1.fact_vendedor
                         ORDER BY SUM(f1.fact_total) DESC )
END
GO

DECLARE @resultado INT
EXECUTE dbo.actualizarComision @masVendedor = @resultado OUTPUT
SELECT @resultado
```

### Guardar contraseñas con hash
En `Usuarios` (Clase 1) la columna `clave` es `VARBINARY(32)`: no guarda la contraseña sino su **hash**. `HASHBYTES('SHA2_256', texto)` devuelve justo 32 bytes. Es de una sola vía: no se puede volver al texto, solo comparar hash contra hash.

```sql
UPDATE Usuarios
SET clave = HASHBYTES('SHA2_256', 'miClave123')
WHERE usuario = 'user1'
GO

CREATE FUNCTION fx_valida_usuario ( @login CHAR(8), @clave VARCHAR(12) )
RETURNS BIT
AS
BEGIN
    DECLARE @retorno BIT = 0

    IF EXISTS ( SELECT 1 FROM Usuarios WHERE usuario = @login AND HASHBYTES('SHA2_256', @clave) = clave )
        SET @retorno = 1

    RETURN @retorno
END
GO

SELECT dbo.fx_valida_usuario('user1', 'miClave123')      -- 1 si el usuario y la clave coinciden, 0 si no
```
