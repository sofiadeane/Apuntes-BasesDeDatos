## Filtrado de datos (parte dos)
- **WHERE:** Muestra unicamente los datos que cumplen cierta condicion booleana
- **GROUP BY:** Agarra todos los registros que cumplan cierta condicion en una columna y los muestra como una sola fila, requiere una operacion de grupo como sum, max, min, avg, etc. para saber que hacer con el resto de los campos. 

Funciones de agregación (funciones de grupo) que podés usar:
- MIN() - devuelve el valor más pequeño de una columna
- MAX() - devuelve el valor más grande de una columna
- COUNT() - devuelve el número de filas de un conjunto
- SUM() - devuelve la suma de una columna numérica
- AVG() - devuelve el valor promedio de una columna numérica

#### COUNT(*) vs COUNT(campo)
- `COUNT(*)` cuenta filas
- `COUNT(campo)` cuenta solo las filas donde ese campo **no es NULL**
- `COUNT(DISTINCT campo)` cuenta cada valor distinto una sola vez

```sql
SELECT COUNT(*) FROM Cliente                 -- cuenta filas

SELECT COUNT(clie_telefono) FROM Cliente     -- cuenta solo los que tienen telefono (no null)

SELECT COUNT(*), COUNT(clie_telefono)
FROM Cliente
WHERE clie_telefono IS NULL                  -- N filas pero 0 telefonos: COUNT(campo) ignora los null
```

#### GROUP BY: una fila por grupo
Con `WHERE` filtro filas **antes** de agrupar. Con `ORDER BY` ordeno el resultado final.

```sql
SELECT 
    fact_cliente,
    SUM(fact_total) AS total_facturado,
    COUNT(*)        AS cantidad_facturas
FROM Factura
WHERE fact_fecha BETWEEN '2012-01-01' AND '2023-12-31'    -- filtra las facturas antes de agrupar
GROUP BY fact_cliente
ORDER BY fact_cliente
```

```sql
SELECT 
    fact_cliente,
    COUNT(*)                                                    AS cantidad_facturas,
    COUNT(DISTINCT fact_vendedor)       AS vendedores_distintos,        -- cada vendedor cuenta una sola vez
    COUNT(fact_vendedor)                          AS cantidad_vendedores        -- cuenta repetidos (si no es null)
FROM Factura
GROUP BY fact_cliente
ORDER BY fact_cliente
```

#### Agrupar por más de un campo
Queda una fila por cada **combinación** de los campos del `GROUP BY`. Todo campo del `SELECT` que no sea una función de grupo tiene que estar en el `GROUP BY`.

```sql
SELECT 
    YEAR(fact_fecha)  AS anio,
    fact_cliente,
    SUM(fact_total)   AS total_facturado,
    COUNT(*)          AS cantidad_facturas,
    AVG(fact_total)   AS promedio_factura,
    MAX(fact_total)   AS factura_mas_alta,
    MIN(fact_total)   AS factura_mas_baja
FROM Factura
GROUP BY YEAR(fact_fecha), fact_cliente     -- una fila por cada año + cliente
ORDER BY fact_cliente, YEAR(fact_fecha)
```

#### HAVING: filtrar grupos
- **WHERE** filtra filas, antes de agrupar. No puede usar funciones de grupo (`SUM`, `COUNT`...)
- **HAVING** filtra grupos, después de agrupar. Sí puede usar funciones de grupo

Orden en que se evalúa una consulta: `FROM` → `WHERE` → `GROUP BY` → `HAVING` → `SELECT` → `ORDER BY`. Por eso `WHERE` y `HAVING` no ven los alias del `SELECT` y `ORDER BY` sí.

```sql
SELECT 
    YEAR(fact_fecha)  AS anio,
    fact_cliente,
    SUM(fact_total)   AS total_facturado,
    COUNT(*)          AS cantidad_facturas
FROM Factura
GROUP BY YEAR(fact_fecha), fact_cliente
HAVING SUM(fact_total) > 5000               -- solo los grupos que facturaron mas de 5000
ORDER BY fact_cliente, YEAR(fact_fecha)
```
