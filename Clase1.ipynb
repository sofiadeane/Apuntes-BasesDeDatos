# Clase 1: Operaciones Basicas

#### Tipos de sentencias SQL
- **DML** (Data Manipulation Language): consultar y modificar los datos => `SELECT`, `INSERT`, `UPDATE`, `DELETE`
- **DDL** (Data Definition Language): crear, modificar y borrar la estructura => `CREATE`, `ALTER`, `DROP`
- **DCL** (Data Control Language): dar y sacar permisos => `GRANT`, `REVOKE`

### 1. Ver Datos con SELECT, FROM, WHERE, TOP N
- SELECT: Indico que campos mostrar
- FROM: Indico de que tablas
- WHERE: Aplico filtros
- TOP N: Pido que me traiga solo las primeras N filas

```sql
-- Selecciono todos los campos de la tabla producto y pido que me traiga solo las primeras 5 filas
SELECT TOP 5 *
FROM Producto
```

#### Tipos de datos en SQL
**Numéricos**
- **Bit:** Tipo logico / booleano. Guarda un 1 o un 0. En algunos motores admite Bit(N) para definir un arreglo de N bits.
- **Tinyint:** Numero entero muy chico. No lleva N, ocupa 1byte    => de 0 a 255 (en varios motores).
- **Smallint:** Numero entero chico. No lleva N, ocupa 2bytes      => desde -32.768 hasta 32.767.
- **Int:** numeros enteros. no tiene largo, ocupa 4bytes           => 10 digitos positivos o negativos.
- **Bigint:** Numero entero grande. No lleva N, ocupa 8bytes       => desde -2^63 hasta 2^63-1.
- **Decimal(t,d):** Decimal exacto, sin redondear. t es el total de digitos y d la cantidad de digitos decimales.
- **Float:** Decimal aproximado, soporta un rango enorme (alrededor de ±1,79 x 10^308).

**Texto**
- **Char(N):** String de longitud fija. N define el len del string. Se asume N=1 si no se aclara. Si len < N, rellena con espacios en blanco (siempre ocupa N).
- **Varchar(N):** String de longitud variable, hasta N caracteres. Solo ocupa los caracteres que realmente se usan. (Text es el viejo equivalente de Varchar(max)).
- **Nchar(N) / Nvarchar(N):** Igual que Char y Varchar pero guardan los caracteres en Unicode (varios idiomas o caracteres especiales).

**Fecha y hora**
- **Date:** Fecha sola (YYYY-MM-DD). No lleva parametro N.
- **Time(P):** Hora sola (HH:MM:SS). P es opcional e indica la precision de las fracciones de segundo.
- **Datetime:** Fecha y hora juntas, con precision de milisegundo. **Smalldatetime** es lo mismo con precision de minuto.
- **Timestamp(P):** Marca de fecha y hora. P es opcional para la precision de las fracciones de segundo. Se usa para marcar cuando se inserto o modifico un registro (se actualiza solo).

**Binarios y otros**
- **Binary(N):** Dato binario de longitud fija (N bytes).
- **Blob / Varbinary(N):** Datos binarios de longitud variable. N define la cantidad maxima de bytes. Guarda archivos, imagenes o PDFs en crudo.
- **Json:** Estructura de texto en formato JSON. No lleva parametro N. Valida la sintaxis del documento.

### 2. Manejar Tablas con CREATE, DROP, ALTER, INSERT, UPDATE y DELETE
**DDL (la estructura)**
- **CREATE:** Crea una tabla, requiere indicar las columnas y sus datos
- **DROP:** Borra la tabla entera (estructura + datos). No se puede borrar si otra tabla la referencia con una FK
- **ALTER:** Modifica la estructura de una tabla que ya existe: agregar o borrar columnas (`ADD`, `DROP COLUMN`) o restricciones

**DML (los datos)**
- **INSERT:** Agrega lineas a la tabla, le tengo que pasar los campos y parametros que deseo cargar
- **UPDATE:** Modifica valores de las filas que ya existen (`SET campo = valor`). Casi siempre va con `WHERE`, si no modifica TODAS las filas
- **DELETE:** Borra filas (no la tabla). Sin `WHERE` borra todas las filas

```sql

CREATE TABLE bancos(                                            -- {nombreColumna} {tipoDeDato} {restriccionDeEsosDatos}
    banc_codigo   INT         PRIMARY KEY,              -- Primary key agrega las restricciones: no null + no duplicada + unica por tabla (ver prox ejemplo)
    banc_nom      VARCHAR(50) NOT NULL,         -- tipoDato(LargoMaximoOpcional) + not null hace que no sea posible nullear el campo
    banc_estado   BIT                                                   -- si no agrego nada no hay restricciones
) 
```

```sql

DROP TABLE bancos                           -- borra la tabla
```

#### Restricciones
- **PRIMARY KEY:** identifica cada fila. Implica NOT NULL + no duplicada, y hay una sola por tabla (puede ser compuesta, ver `UsuariosXsegmentos`)
- **FOREIGN KEY / REFERENCES:** la columna solo acepta valores que existan en la PK de otra tabla (integridad referencial)
- **NOT NULL:** el campo no se puede dejar vacio
- **UNIQUE:** el valor no se puede repetir en la tabla
- **CHECK:** el valor tiene que cumplir una condicion (ej: `edad > 0 AND edad < 150`)
- **DEFAULT:** valor que se carga si no indico ninguno

```sql
CREATE TABLE Bancos(                        
    banc_codigo      INT                        PRIMARY KEY, 
    banc_cuit           CHAR(13)              NOT NULL,    
    banc_nom         VARCHAR(50)     NOT NULL,    
    banc_estado     BIT                       
)

CREATE TABLE Sucursales(
    sucu_id                 INT                             PRIMARY KEY, 
    banc_codigo       INT                              NOT NULL             REFERENCES bancos,                  -- REFERENCES bancos significa que banc_codigo es una FK de la tabla bancos
    sucu_dir               VARCHAR(50)          NOT NULL, 
    sucu_nombre     VARCHAR(50), 
    sucu_num           SMALLINT, 
    sucu_estado        BIT                              NOT NULL 
)

CREATE TABLE Usuarios(
    usuario                 CHAR(8)                    PRIMARY KEY,
    clave                      VARBINARY(32)     -- binario
)

CREATE TABLE Segmentos(
    idsegmento    INT PRIMARY KEY,
    DESCripcion   VARCHAR(20)
)

CREATE TABLE UsuariosXsegmentos(
    usuario             CHAR(8)     REFERENCES usuarios,
    idsegmento    INT               REFERENCES segmentos,
    CONSTRAINT pk_user_segm PRIMARY KEY (usuario, idsegmento)
)

INSERT INTO Usuarios (usuario, clave) 
    VALUES ('user1', NULL)

INSERT INTO Segmentos (idsegmento, Descripcion) 
    VALUES (1, 'segmento 1')
```

#### ALTER, UPDATE y la restriccion CHECK
Orden del ejemplo de abajo: borro la columna `antiguedad` (si existia), la agrego con `DEFAULT 0`, pongo todas las filas en 0 con `UPDATE` y despues agrego un `CHECK` para que nunca sea negativa. El ultimo `INSERT` falla justamente porque `antiguedad = -1` no cumple el `CHECK`.

```sql
-- OJO: DROP COLUMN falla si la columna todavia no existe (la primera vez que se corre el cuaderno)
-- borra la columna antiguedad
ALTER TABLE bancos 
    DROP COLUMN antiguedad

ALTER TABLE bancos 
    ADD antiguedad int DEFAULT 0

UPDATE bancos 
    SET antiguedad = 0

ALTER TABLE bancos 
    ADD CONSTRAINT check_antiguedad CHECK ( antiguedad >= 0 )

INSERT INTO bancos 
    VALUES ( 4, '12-45678901-5', 'nombre 4 ', 1, -1)
 
```

#### DELETE vs DROP
- **DELETE** borra filas (la tabla sigue existiendo). Hay que poner `WHERE`, si no se borran todas.
- **DROP TABLE** borra la tabla entera.

```sql
-- borra solo las filas que cumplen el WHERE
DELETE FROM Bancos
WHERE banc_codigo = 3

-- borra la tabla entera
DROP Bancos 
```

#### Esquemas
Un esquema es un conjunto de objetos (tablas, vistas, indices) de los que un usuario es el dueño (owner). En SQL Server el esquema por defecto es `dbo`, por eso despues se ve `dbo.Tabla` o `dbo.vista`.
- El nombre de un esquema (u objeto) nunca puede tener espacios, se usa guion bajo: `Grupo_Los_Mejores`
- **Separar ambientes:** en una misma base conviven Desarrollo, Integracion y Testing sin que choquen los nombres (`DESA.clientes` y `TEST.clientes`)
- **Aplicaciones SaaS:** distintas empresas comparten la misma estructura de tablas pero con datos y dueños separados

### 3. Filtrado de datos
- **WHERE:** muestra unicamente los datos que cumplen cierta condicion booleana
- **Operadores:** `=`, `<>`, `<`, `>`, `<=`, `>=`, `AND`, `OR`, `BETWEEN a AND b` (incluye los extremos)
- **ORDER BY:** ordena el resultado. `ASC` de menor a mayor (es el default), `DESC` de mayor a menor. Con varios campos, si hay empate en el primero desempata con el siguiente
- **NULL:** significa "desconocido", no se compara con `=`. Se usa `IS NULL` / `IS NOT NULL`

```sql
SELECT * FROM cliente
WHERE
 clie_codigo = '00009'

SELECT * FROM cliente
 WHERE
 clie_codigo BETWEEN '00009' AND '00012'

SELECT * FROM cliente
 WHERE
 clie_codigo > '00012' AND
 clie_limite_credito < 2000

SELECT * FROM cliente
 WHERE
 clie_codigo > '00012' AND
 clie_limite_credito < 2000
ORDER BY
 clie_codigo ASC

SELECT * FROM cliente
WHERE
 clie_codigo > '00012' AND
 clie_limite_credito < 2000
ORDER BY
 clie_codigo DESC
 
 
SELECT * FROM cliente
 WHERE
 clie_codigo > '00012' AND
 clie_limite_credito < 2000
ORDER BY
 clie_limite_credito DESC ,
 clie_codigo ASC

SELECT * FROM cliente
 ORDER BY
 clie_domicilio ASC

SELECT * FROM cliente
 WHERE
 clie_telefono IS NULL

SELECT * FROM cliente
WHERE
 clie_telefono IS NOT NULL
```

```sql
-- EJERCICIO 1
-- Código y razón social de los clientes con límite de crédito >= $1000, ordenado por código de cliente. *(Ya está resuelto en `EjerciciosGuia` N1: sirve de repaso.)*

SELECT clie_razon_social AS razon_social, clie_codigo AS codigo, clie_limite_credito AS limite
FROM Cliente
WHERE clie_limite_credito >= 1000
ORDER BY clie_codigo
```

```sql
-- EJERCICIO 2
-- Las 5 facturas de 2012 con mayor `fact_total`: mostrar fecha, vendedor y total, de mayor a menor.

SELECT TOP 5
    fact_fecha      AS fecha,
    fact_vendedor   AS vendedor,
    fact_total      AS total
FROM Factura
WHERE YEAR(fact_fecha) = 2012
ORDER BY total DESC
```

```sql
-- EJERCICIO 3
-- Los clientes que **no** tienen teléfono cargado, ordenados por razón social. Después mostrar solo los que **sí** tienen y cuyo límite de crédito está entre 500 y 2000 (probar con `BETWEEN` y con `>=` / `<=`).

SELECT          TOP 10
    clie_codigo AS codigo,
    clie_telefono AS cel,
    clie_razon_social AS nombre
FROM Cliente
WHERE clie_telefono IS NULL
ORDER BY nombre ASC

SELECT          TOP 10
    clie_codigo AS codigo,
    clie_telefono AS cel,
    clie_limite_credito AS llimite,
    clie_razon_social AS nombre
FROM Cliente
WHERE 
    ( clie_telefono IS NOT NULL )   AND
    ( clie_limite_credito BETWEEN 500 AND 2000 )
ORDER BY nombre ASC

SELECT          TOP 10
    clie_codigo AS codigo,
    clie_telefono AS cel,
    clie_limite_credito AS llimite,
    clie_razon_social AS nombre
FROM Cliente
WHERE 
    ( clie_telefono IS NOT NULL )   AND
    ( clie_limite_credito >= 500) AND 
    ( clie_limite_credito <= 2000 )
ORDER BY nombre ASC
```

```sql
-- EJERCICIO 4
-- Las facturas del primer semestre de 2011 emitidas por un vendedor a elección, ordenadas por fecha descendente y, si empatan, por total descendente. Mostrar el año con alias.

SELECT 
    YEAR(fact_fecha)    AS anio,
    fact_numero         AS numero,
    fact_fecha          AS fecha,
    fact_vendedor       AS vendedor,
    fact_total          AS total
FROM Factura
WHERE fact_fecha >= '2011-01-01' AND fact_fecha < '2011-07-01'     -- enero a junio (equivale a YEAR = 2011 AND MONTH <= 6)
  AND fact_vendedor = 4
ORDER BY fecha DESC, total DESC
```

```sql
-- EJERCICIO 5
-- DDL/DML: crear una tabla `Sucursales_Prueba` con PK, un campo `UNIQUE`, un campo con `DEFAULT` y un `CHECK`. Insertar 3 filas, probar que cada restricción rechaza una fila inválida, 

CREATE TABLE Sucursales_Prueba(
    sucu_id         INT           PRIMARY KEY,
    sucu_codigo     VARCHAR(10)   NOT NULL UNIQUE,                  -- UNIQUE: no se repite
    sucu_nombre     VARCHAR(50)   NOT NULL,
    sucu_estado     BIT           NOT NULL DEFAULT 1,               -- DEFAULT: si no lo indico vale 1
    sucu_empleados  INT           CHECK ( sucu_empleados >= 0 )     -- CHECK: no puede ser negativo
)
```
