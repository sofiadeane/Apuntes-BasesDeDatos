# Funciones de ventana (`OVER`)
A diferencia de `GROUP BY`, **no colapsan las filas**: cada fila se mantiene y se le agrega una columna calculada sobre una "ventana" de filas.

### ROW_NUMBER: numerar filas
`ROW_NUMBER()` le pone a cada fila su número de orden según el `ORDER BY` del `OVER`.

```sql
SELECT 
    ROW_NUMBER() OVER ( ORDER BY clie_codigo ASC )  AS numero_fila_asc,
    ROW_NUMBER() OVER ( ORDER BY clie_codigo DESC ) AS numero_fila_desc,
    *
FROM Cliente
```

### PARTITION BY: reiniciar por grupo
Con `PARTITION BY` la numeración arranca de nuevo en 1 en cada grupo (acá, en cada año).

```sql
SELECT 
    ROW_NUMBER() OVER ( PARTITION BY YEAR(fact_fecha) ORDER BY fact_cliente DESC ) AS nro_fila,
    fact_fecha, 
    fact_cliente, 
    fact_total
FROM Factura
```

### Filtrar por una función de ventana
No se puede usar en el `WHERE` (se calcula después de filtrar), así que se envuelve en un subselect en el `FROM`. Este es uno de los pocos casos donde el subselect en el `FROM` sí vale.

```sql
SELECT * 
FROM ( SELECT 
           ROW_NUMBER() OVER ( PARTITION BY YEAR(fact_fecha) ORDER BY fact_cliente DESC ) AS nro_fila,
           fact_fecha, 
           fact_cliente, 
           fact_total
       FROM Factura
     ) AS T
WHERE nro_fila BETWEEN 1 AND 30     -- las primeras 30 filas de cada año
```

### SUM, AVG, etc. con OVER
Las funciones de agregación también se pueden usar con `OVER`. Cada fila conserva su detalle y además muestra el total (o promedio) de su grupo.

```sql
SELECT 
    SUM(fact_total) OVER ( PARTITION BY YEAR(fact_fecha), fact_cliente ) AS suma_anio_cliente,
    AVG(fact_total) OVER ( PARTITION BY YEAR(fact_fecha), fact_cliente ) AS promedio_anio_cliente,
    fact_fecha,
    fact_cliente,
    fact_total
FROM Factura
```

### Acumulados: ROWS BETWEEN
Con `ORDER BY` + `ROWS BETWEEN` defino qué filas entran en el cálculo de cada fila. `UNBOUNDED PRECEDING AND CURRENT ROW` = desde la primera fila del grupo hasta la fila actual, o sea un **acumulado** (suma incremental).

```sql
SELECT 
    YEAR(fact_fecha) AS anio,
    fact_fecha,
    fact_cliente,
    fact_total,
    SUM(fact_total) OVER ( PARTITION BY YEAR(fact_fecha), fact_cliente
                           ORDER BY fact_fecha
                           ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW 
                         ) AS suma_incremental
FROM Factura
ORDER BY fact_cliente, anio, fact_fecha
```

Sumar `1` en cada fila con el mismo acumulado es lo mismo que numerar con `ROW_NUMBER()`:

```sql
SELECT 
    YEAR(fact_fecha) AS anio, 
    fact_fecha, 
    fact_cliente, 
    fact_total,
    ROW_NUMBER() OVER ( PARTITION BY fact_cliente ORDER BY fact_fecha ) AS nro_fila_grupo,
    SUM(1) OVER ( PARTITION BY fact_cliente 
                  ORDER BY fact_fecha 
                  ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW ) AS suma_incremental     -- da lo mismo que nro_fila_grupo
FROM Factura
ORDER BY fact_cliente, anio, fact_fecha
```

## Ejercicios para practicar
Con `ROW_NUMBER()`, `PARTITION BY`, `SUM/AVG ... OVER` y acumulados se pueden hacer estos de la práctica (todos tienen un "el que más..." por grupo, que es donde conviene la función de ventana). Todos están resueltos en `EjerciciosGuia`, sección *Respuestas > Clase 5*:

- **Ej. 10 (extra B):** los 10 productos más y menos vendidos de la historia y el cliente que más compró de cada uno. *(Resuelto en `EjerciciosGuia`: con `TOP` + `IN` en el extra B y con `ROW_NUMBER()` en Respuestas > Clase 5.)*
- **Ej. 16:** clientes con ventas inferiores a 1/3 del promedio de ventas del producto más vendido de 2012; mostrar nombre, unidades de 2012 y el producto que más compró (el de menor código si empatan), ordenados por provincia. *(Ojo: en este esquema `Cliente` no tiene provincia; si el profe no aclara otra cosa, ordená por `clie_codigo` o por zona, ver `Departamento`/`Zona`.)*
- **Ej. 18:** estadística por rubro: ventas en pesos, los dos productos más vendidos, el cliente que más compró en los últimos 30 días; sin NULLs, ordenado por productos diferentes vendidos.
- **Ej. 19:** sugerencia de recategorización de familia por producto (la familia mayoritaria entre los productos con los mismos 5 primeros caracteres de detalle; desempata el menor código); solo los que cambian.
- **Ej. 23:** por año: el producto con composición más vendido, cuántos productos lo componen, facturas en que aparece, cliente que más lo compró y % sobre la venta del año.
- **Ej. 25:** por año: la familia más vendida, rubros que la componen, facturas, cliente que más compró y % sobre la venta del año.
- **Ej. 26:** ranking de empleados: depósitos a cargo, monto facturado en el año corriente, cliente al que más vendió, producto más vendido y % sobre el total del año.
- **Ej. 27:** por año y envase: cantidad de productos, facturados, producto más vendido, monto y % sobre el total del año.
- **Ej. 28** (el **Ej. 31** es el mismo enunciado): por año y vendedor: facturas, clientes, productos con y sin composición y monto total, ordenado por productos diferentes vendidos.
- **Ej. 30:** estadística de 2012 de los jefes (empleados a cargo, monto, facturas, mejor empleado), usando como máximo **una** subconsulta y solo si sus subordinados hicieron más de 10 facturas.
