# Joins
Un JOIN une filas de dos tablas a través de un valor que tienen en común (normalmente PK de una = FK de la otra).

### Inner Joins
Devuelve solo las filas con valores que matchean en las dos tablas. `JOIN` solo es lo mismo que `INNER JOIN`.

**Sintaxis:** 

<!-- sql-src:U0VMRUNUIAogICAgby5PcmRlcklELCAKICAgIGMuQ3VzdG9tZXJOYW1lCkZST00gT3JkZXJzIG8KSU5ORVIgSk9JTiBDdXN0b21lcnMgYyBPTiBvLkN1c3RvbWVySUQgPSBjLkN1c3RvbWVySUQ7 -->
<pre class="sql"><code><span class="kw">SELECT</span> 
    o.OrderID, 
    c.CustomerName
<span class="kw">FROM</span> Orders o
<span class="kw">INNER</span> <span class="kw">JOIN</span> Customers c <span class="kw">ON</span> o.CustomerID = c.CustomerID;</code></pre>

### Outer Joins
Devuelve todas las filas de una tabla + los que matchean de la otra. Donde no hay match, las columnas de la otra tabla quedan en NULL. Se puede especificar:
- **LEFT (OUTER) JOIN:** Trae toda la tabla a la izquierda y solo las que matchean de la derecha
- **RIGHT (OUTER) JOIN:** Trae toda la tabla a la derecha y solo las que matchean de la izquierda
- **FULL (OUTER) JOIN:** Trae ambas tablas completas, unificadas en una sola

**Sintaxis:** 

<!-- sql-src:U0VMRUNUIAogICAgYy5DdXN0b21lck5hbWUsCiAgICBvLk9yZGVySUQKRlJPTSBDdXN0b21lcnMgYwpMRUZUIEpPSU4gT3JkZXJzIG8gT04gYy5DdXN0b21lcklEID0gby5DdXN0b21lcklEOw== -->
<pre class="sql"><code><span class="kw">SELECT</span> 
    c.CustomerName,
    o.OrderID
<span class="kw">FROM</span> Customers c
<span class="kw">LEFT</span> <span class="kw">JOIN</span> Orders o <span class="kw">ON</span> c.CustomerID = o.CustomerID;</code></pre>

**LEFT, RIGHT y FULL:** el mismo join pero conservando las filas sin match. Compará cuántas filas devuelve cada uno y dónde aparecen los `NULL`.

```sql
SELECT p.numero AS primo, e.numero AS entero          -- todos los primos, esten o no en enteros
FROM primos p 
LEFT JOIN enteros e ON p.numero = e.numero

SELECT p.numero AS primo, e.numero AS entero          -- todos los enteros, esten o no en primos
FROM primos p 
RIGHT JOIN enteros e ON p.numero = e.numero

SELECT p.numero AS primo, e.numero AS entero          -- todo de las dos tablas
FROM primos p 
FULL OUTER JOIN enteros e ON p.numero = e.numero
```

Para usar las columnas de las dos tablas se le pone un **alias** a cada tabla (`f`, `c`) y se escribe `alias.columna`. Si el nombre de la columna es único en las tablas del join, el alias es opcional (`fact_cliente` no se confunde con nada).

```sql
SELECT * 
FROM Factura f 
JOIN Cliente c ON f.fact_cliente = c.clie_codigo
```

**Join + GROUP BY + HAVING:** primero se une, después se agrupa. Como `clie_razon_social` no es una función de grupo, también va en el `GROUP BY`.

```sql
SELECT 
    fact_cliente,
    clie_razon_social,
    YEAR(fact_fecha)                                AS anio_factura,
    COUNT(*)                                               AS cantidad_facturas,
    COUNT(DISTINCT fact_vendedor)  AS vendedores_distintos,
    AVG(fact_total)                                     AS promedio_factura,
    MAX(fact_total)                                     AS factura_mas_alta,
    MIN(fact_total)                                      AS factura_mas_baja
FROM Factura f 
JOIN Cliente c ON f.fact_cliente = c.clie_codigo
GROUP BY YEAR(fact_fecha), fact_cliente, clie_razon_social
HAVING SUM(fact_total) > 5000
ORDER BY fact_cliente, YEAR(fact_fecha)
```

## Ejercicios para practicar
Con `INNER` / `LEFT JOIN`, varias tablas, clave compuesta de `Factura` + `Item_Factura`, `GROUP BY` y `HAVING` se pueden hacer estos de la práctica:

- **Ej. 2:** código y detalle de los artículos vendidos en 2012, ordenados por cantidad vendida. *(Ya resuelto en `EjerciciosGuia` N2.)*
- **Ej. 3:** código, nombre y stock total de cada producto (sin importar el depósito), ordenado por nombre. *(Ya resuelto en `EjerciciosGuia` N3.)*
- **Ej. 9:** código del jefe, código y nombre de cada empleado que lo tiene como jefe, y la cantidad de depósitos que ambos tienen asignados. *(Auto-join sobre `Empleado`.)*
- **Ej. 11:** por familia: detalle, cantidad de productos diferentes vendidos y monto de las ventas sin impuestos; ordenado por la familia con más productos diferentes vendidos, solo familias con ventas > $20000 en 2012.
- **Ej. 13:** para cada producto con composición: nombre, precio, y la suma de (precio del componente x cantidad) de lo que lo compone. Solo los compuestos por más de 2 productos, ordenados de mayor a menor por cantidad de componentes.
- **Ej. 15:** pares de productos vendidos juntos (misma factura) más de 500 veces: código y detalle de cada uno y cantidad de veces, ordenado por esa cantidad, sin repetir pares. *(Auto-join de `Item_Factura`.)*

Si aparece un resultado inflado en alguno, mirá la nota de cardinalidad de la Clase 4.

```sql
-- EJERCICIO 2: código y detalle de los artículos vendidos en 2012, ordenados por cantidad vendida. *(Ya resuelto en `EjerciciosGuia` N2.)*

SELECT
    p.prod_codigo               AS id_producto,
    p.prod_detalle              AS detalle,
    SUM(i.item_cantidad)        AS cant_total
FROM Producto p
JOIN Item_Factura i ON p.prod_codigo = i.item_producto
JOIN Factura f 
    ON  i.item_tipo     = f.fact_tipo
    AND i.item_sucursal = f.fact_sucursal
    AND i.item_numero   = f.fact_numero
WHERE YEAR(f.fact_fecha) = 2012
GROUP BY p.prod_codigo, p.prod_detalle
ORDER BY cant_total DESC
```

```sql
-- EJERCICIO 3: código, nombre y stock total de cada producto (sin importar el depósito), ordenado por nombre. *(Ya resuelto en `EjerciciosGuia` N3.)*
SELECT
    p.prod_codigo                    AS prod_codigo,
    p.prod_detalle                    AS nombre,
    SUM(s.stoc_cantidad)      AS cantidad
FROM Producto p
JOIN STOCK s ON  p.prod_codigo = s.stoc_producto
GROUP BY prod_codigo, prod_detalle
ORDER BY nombre ASC
```

```sql
-- EJERCICIO 9: código del jefe, código y nombre de cada empleado que lo tiene como jefe, y la cantidad de depósitos que ambos tienen asignados. 
SELECT 
    j.empl_codigo AS jefe,
    e.empl_codigo AS empleado,
    e.empl_nombre AS nombre_empleado,
    COUNT(d.depo_codigo) AS cant_depositos
FROM Empleado e -- en e busco datos de empleados
JOIN Empleado j ON j.empl_codigo = e.empl_jefe -- en j busco los datos del jefe
LEFT JOIN DEPOSITO d ON d.depo_encargado IN (e.empl_codigo, j.empl_codigo) -- depositos del empleado o del jefe
GROUP BY  j.empl_codigo, e.empl_codigo, e.empl_nombre

/**Por qué LEFT JOIN con Deposito
La diferencia está en qué pasa con los empleados que no tienen depósitos.

JOIN (INNER): solo devuelve filas que encuentran pareja en ambas tablas. Un empleado sin depósito, y cuyo jefe tampoco tiene, no matchea con nada en Deposito, así que desaparece del resultado.
LEFT JOIN: conserva todas las filas de la tabla de la izquierda (acá, la combinación empleado-jefe) aunque no tengan pareja. En esos casos las columnas de d vienen en NULL.

Por eso no va un WHERE sobre d (ni repetir la condicion del ON): lo convertiria en INNER JOIN y descartaria a los que tienen 0 depositos.*/
```

```sql
-- EJERCICIO 11: por familia: detalle, cantidad de productos diferentes 
-- vendidos y monto de las ventas sin impuestos; ordenado por la familia 
-- con más productos diferentes vendidos, solo familias con ventas > $20000 en 2012.

SELECT
    f.fami_detalle                                                      AS nombre_familia,
    COUNT(DISTINCT i.item_producto)                  AS productos_distintos,
    SUM(i.item_cantidad * i.item_precio)                 AS monto_sin_impuestos   -- item_precio es unitario
FROM Producto p
JOIN Familia f ON f.fami_id = p.prod_familia
JOIN Item_Factura i ON i.item_producto = p.prod_codigo
JOIN Factura fac
    ON  i.item_tipo            = fac.fact_tipo
    AND i.item_sucursal   = fac.fact_sucursal
    AND i.item_numero    = fac.fact_numero
WHERE YEAR(fac.fact_fecha) = 2012
GROUP BY f.fami_id, f.fami_detalle
HAVING SUM(i.item_cantidad * i.item_precio) > 20000
ORDER BY productos_distintos DESC
```

```sql
-- EJERCICIO 13: para cada producto con composición: nombre, precio, y la suma de (precio del componente x cantidad) de lo que lo compone. 
-- Solo los compuestos por más de 2 productos, ordenados de mayor a menor por cantidad de componentes.

SELECT 
    c.comp_producto                                                                            AS id_producto,
    SUM(c.comp_cantidad)                                                                 AS cant_de_comp,
    p.prod_detalle                                                                                  AS nombre,
    p.prod_precio                                                                                   AS precio,
    SUM(c.comp_cantidad *p_comp.prod_precio)                        AS precio_total
FROM Composicion c
JOIN Producto p                 ON p.prod_codigo = c.comp_producto
JOIN Producto p_comp    ON p_comp.prod_codigo = c.comp_componente
GROUP BY c.comp_producto, p.prod_detalle, p.prod_precio   
HAVING SUM(c.comp_cantidad) > 2
ORDER BY cant_de_comp DESC
```

```sql
-- EJERCICIO 15: pares de productos vendidos juntos (misma factura) más de 500 veces: código y detalle de cada uno y cantidad 
-- de veces, ordenado por esa cantidad, sin repetir pares. 

SELECT
    i1.item_producto    AS cod_producto_1,
    p1.prod_detalle     AS detalle_1,
    i2.item_producto    AS cod_producto_2,
    p2.prod_detalle     AS detalle_2,
    COUNT(*)            AS veces_juntos
FROM Item_Factura i1
JOIN Item_Factura i2 
    ON  i2.item_tipo     = i1.item_tipo
    AND i2.item_sucursal = i1.item_sucursal
    AND i2.item_numero   = i1.item_numero              -- misma factura (OJO: NO igualar item_producto, sino nunca hay dos distintos)
    AND i2.item_producto > i1.item_producto            -- > y no <>: con <> aparecen (A,B) y (B,A), con > cada par una sola vez
JOIN Producto p1 ON p1.prod_codigo = i1.item_producto
JOIN Producto p2 ON p2.prod_codigo = i2.item_producto
GROUP BY i1.item_producto, p1.prod_detalle, i2.item_producto, p2.prod_detalle   -- un grupo por par
HAVING COUNT(*) > 500                                                            -- cada fila del grupo es una factura con el par
ORDER BY COUNT(*) DESC
```
